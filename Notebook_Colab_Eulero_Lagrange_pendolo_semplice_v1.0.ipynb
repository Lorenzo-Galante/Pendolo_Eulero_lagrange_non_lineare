# Equazioni di Eulero–Lagrange: il modello del pendolo semplice

Per un sistema descritto da $n$ coordinate generalizzate

$$
q_1,q_2,\ldots,q_n,
$$

la forma generale delle equazioni di Eulero–Lagrange è

$$
\boxed{
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot q_j}\right)
-
\frac{\partial L}{\partial q_j}
=Q_j^{\mathrm{(nc)}},
\qquad j=1,\ldots,n,
}
$$

dove $Q_j^{\mathrm{(nc)}}$ rappresenta l'eventuale forza generalizzata non conservativa. Quando tutte le forze attive sono conservative,

$$
Q_j^{\mathrm{(nc)}}=0,
$$

e la lagrangiana è

$$
L(q,\dot q,t)=T(q,\dot q,t)-V(q,t).
$$

Nel pendolo ideale considerato qui non introduciamo forze non conservative, quindi useremo il membro destro nullo senza scrivere ancora l'equazione specifica del pendolo.

In questa attività applicheremo il formalismo al **pendolo semplice ideale**:

- un punto materiale di massa $m$;
- collegato a un punto fisso $O$ mediante un filo o un'asta ideale di lunghezza costante $\ell$;
- moto contenuto nel piano verticale $(x,y)$;
- gravità uniforme di accelerazione $g$;
- assenza di attrito e resistenza dell'aria.

Il vincolo geometrico è

$$
x^2+y^2=\ell^2.
$$

Il sistema ha un solo grado di libertà. Scegliamo come coordinata generalizzata

$$
\boxed{q=\theta},
$$

dove $\theta$ è l'angolo misurato dalla verticale rivolta verso il basso, positivo nel verso indicato in figura. La sua derivata temporale è

$$
\boxed{\dot q=\dot\theta=\frac{d\theta}{dt}},
$$

cioè la **velocità angolare** del pendolo; inoltre

$$
\ddot q=\ddot\theta=\frac{d^2\theta}{dt^2}
$$

è l'accelerazione angolare.

* l'energia potenziale gravitazionale $V$.

### Scelta dello zero dell'energia potenziale

Ricorda che l'energia potenziale è definita **a meno di una costante additiva**: possiamo quindi scegliere liberamente la configurazione alla quale assegnare energia potenziale nulla.

In questa attività scegliamo come riferimento la posizione di **minima energia potenziale**, cioè la configurazione in cui il pendolo si trova verticalmente verso il basso:

$$
\theta=0.
$$

Assumiamo quindi

$$
\boxed{V(0)=0}.
$$

Nel calcolare $V(\theta)$, considera dunque la variazione di altezza della massa rispetto a questa posizione di riferimento.

> Una diversa scelta dello zero dell'energia potenziale cambierebbe $V$ soltanto per una costante e non modificherebbe l'equazione del moto. Qui fissiamo $V(0)=0$ per utilizzare tutti la stessa convenzione.

Nella prossima cella inserirai il risultato dei tuoi calcoli nel codice.

La cella trasformerà automaticamente le tue espressioni in formule matematiche ben renderizzate. Non le sostituirà con quelle corrette del pendolo: la successiva soluzione numerica seguirà proprio l'equazione prodotta dalle tue scelte.


````python
# CELLA 3 — Inserimento di T e V e visualizzazione matematica
import sympy as sp
from IPython.display import display, Markdown, Math

sp.init_printing(use_latex="mathjax")

# Simboli disponibili allo studente.
# Le variabili Python restano m, l, g, theta, theta_dot,
# ma i nomi interni sono scelti per ottenere una notazione LaTeX da manuale.
m, g = sp.symbols('m g', positive=True, finite=True)
ell = sp.symbols('ell', positive=True, finite=True)
l = ell                         # lo studente continua a scrivere semplicemente: l

t = sp.symbols('t', real=True)
theta = sp.symbols('theta', real=True, finite=True)
theta_dot = sp.Symbol(r'\dot{\theta}', real=True, finite=True)
theta_ddot = sp.Symbol(r'\ddot{\theta}', real=True, finite=True)
sin, cos = sp.sin, sp.cos

# ==========================================================
# ===== INIZIO AREA STUDENTE: MODIFICA SOLO QUESTE 2 RIGHE ===
# ==========================================================
T = 0  # <-- SOSTITUISCI 0 con la tua espressione di T
V = 0  # <-- SOSTITUISCI 0 con la tua espressione di V
# ==========================================================
# ===== FINE AREA STUDENTE: NON MODIFICARE SOTTO ============
# ==========================================================

# Convertiamo eventuali decimali semplici (per esempio 0.5) in frazioni esatte:
# questo migliora la leggibilità delle formule senza cambiare il ragionamento dello studente.
T = sp.nsimplify(sp.sympify(T), rational=True)
V = sp.nsimplify(sp.sympify(V), rational=True)
L = sp.simplify(T - V)

# Mostriamo separatamente testo e formule: Math usa direttamente MathJax
# e rende le espressioni in modo affidabile anche in Google Colab.
display(Markdown("### Espressioni inserite"))
display(Math(r"\boxed{T(\theta,\dot{\theta}) = " + sp.latex(T) + r"}"))
display(Math(r"\boxed{V(\theta) = " + sp.latex(V) + r"}"))
display(Markdown("La lagrangiana risultante è"))
display(Math(r"\boxed{L(\theta,\dot{\theta}) = T-V = " + sp.latex(L) + r"}"))

````

## **Attività 3 — Dalla Lagrangiana all'equazione del moto**
<br>

<span style="
display:inline-block;
background:#fff3a3;
color:#111;
padding:3px 8px;
margin:12px 0;
border-radius:5px;
font-weight:1000;">
✍️ CARTA E PENNA
</span>

<br>
Nella cella precedente hai costruito le funzioni $T$ e $V$, adesso calcola la Lagrangiana del sistema

$$
L(\theta,\dot{\theta})=T-V.
$$

E, **senza utilizzare il codice** della cella successiva, scrivi  l'equazione di Eulero–Lagrange

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot{\theta}}
\right)
-
\frac{\partial L}{\partial \theta}
=0.
$$

Calcola separatamente:

$$
\frac{\partial L}{\partial \dot{\theta}},
\qquad
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot{\theta}}
\right),
\qquad
\frac{\partial L}{\partial \theta}.
$$

Combina quindi i risultati e ricava l'equazione differenziale del moto.

Infine prova a scriverla isolando l'accelerazione generalizzata:

$$
\ddot{\theta}=\ldots
$$

Solo dopo aver completato il calcolo, esegui la cella successiva e confronta il tuo risultato con quello ottenuto simbolicamente dal codice **a partire dalle stesse espressioni di $T$ e $V$ che hai inserito tu**.

Il confronto verifica quindi i passaggi della derivazione simbolica, non la correttezza fisica iniziale delle espressioni di energia che hai scelto.


````python
#@title Derivazione simbolica dalle espressioni dello studente  { display-mode: "form" }

# ==========================================================
# DERIVAZIONE SIMBOLICA DELL'EQUAZIONE DI EULERO–LAGRANGE
# ==========================================================
#
# Questa cella utilizza ESATTAMENTE le espressioni T e V
# inserite dallo studente nella cella precedente.
# ==========================================================

from IPython.display import display, Markdown, Math

# Derivate parziali della Lagrangiana
dL_dtheta = sp.diff(L, theta)
dL_dtheta_dot = sp.diff(L, theta_dot)

# Derivata totale rispetto al tempo di ∂L/∂theta_dot
d_dt_dL_dtheta_dot = (
    sp.diff(dL_dtheta_dot, theta) * theta_dot
    + sp.diff(dL_dtheta_dot, theta_dot) * theta_ddot
    + sp.diff(dL_dtheta_dot, t)
)

# Equazione di Eulero–Lagrange
EL = sp.simplify(
    d_dt_dL_dtheta_dot - dL_dtheta
)

display(Markdown("### Derivazione simbolica dalle espressioni inserite"))
display(Markdown(
    "Il codice utilizza esattamente le tue espressioni di $T$ e $V$: "
    "non le confronta con una soluzione attesa e non le sostituisce."
))

display(Math(
    r"\frac{\partial L}{\partial \dot{\theta}}="
    + sp.latex(sp.simplify(dL_dtheta_dot))
))

display(Markdown("<br>"))

display(Math(
    r"\frac{d}{dt}"
    r"\left(\frac{\partial L}{\partial \dot{\theta}}\right)="
    + sp.latex(sp.simplify(d_dt_dL_dtheta_dot))
))

display(Markdown("<br>"))

display(Math(
    r"\frac{\partial L}{\partial \theta}="
    + sp.latex(sp.simplify(dL_dtheta))
))

display(Markdown("<br><br>"))

display(Markdown("### Equazione di Eulero–Lagrange"))

display(Markdown("<br>"))

display(Math(
    r"\frac{d}{dt}"
    r"\left(\frac{\partial L}{\partial \dot{\theta}}\right)"
    r"-\frac{\partial L}{\partial\theta}=0"
    r"\qquad\Longrightarrow\qquad"
    + sp.latex(EL)
    + r"=0"
))

# Isoliamo theta_ddot
coeff_ddot = sp.simplify(sp.diff(EL, theta_ddot))
resto = sp.simplify(EL.subs(theta_ddot, 0))

if coeff_ddot != 0:

    accelerazione = sp.simplify(-resto / coeff_ddot)

    display(Markdown("<br><br>"))

    display(Markdown(
        "Scrivendo l'equazione in forma esplicita rispetto "
        "all'accelerazione generalizzata:"
    ))

    display(Markdown("<br>"))

    display(Math(
        r"\boxed{\ddot{\theta}="
        + sp.latex(accelerazione)
        + r"}"
    ))

    display(Markdown("<br>"))
    display(Markdown("### Osservazione sul parametro massa"))

    if m not in accelerazione.free_symbols:
        display(Markdown(
            "Nell'equazione esplicita appena ottenuta la massa $m$ **non compare**. "
            "La traiettoria prevista da questa equazione non dipende quindi dal valore della massa. "
            "Da questo punto $m$ non verrà trattata come parametro della **dinamica**. "
            "Resterà invece presente nelle espressioni di $T$ e $V$ e, quando necessario, "
            "verrà usata per valutare le energie in unità fisiche."
        ))
    else:
        display(Markdown(
            "Nell'equazione esplicita ottenuta dalle espressioni inserite la massa $m$ compare ancora. "
            "Il notebook non la elimina né modifica il modello: in questo caso $m$ verrà mantenuta "
            "come parametro anche nell'esplorazione dinamica."
        ))

else:

    display(Markdown("<br><br>"))

    display(Markdown(
        "**Non è possibile isolare $\\ddot{\\theta}$ "
        "dall'equazione ottenuta. Controlla le espressioni "
        "inserite per $T$ e $V$.**"
    ))
````

````python
#@title SIMULAZIONE numerica e animazione a partire dalla "Tua" equazione del moto  { display-mode: "form" }

# CELLA 4 — Equazione di Eulero–Lagrange, soluzione numerica e animazione
# Codice nascosto di default in Colab: usa “Mostra codice” per analizzarlo.
#
# VERSIONE VELOCE E AD ALTA LEGGIBILITÀ:
# - Python deriva l'equazione e integra l'ODE una sola volta;
# - i pannelli animati sono disegnati direttamente dal browser con Canvas;
# - Canvas ad alta densità di pixel per testi e curve più nitidi;
# - BLOCCO 1 (cinematica), due colonne: A = (x,y) + theta(t), B = theta_dot(t) + theta_ddot(t);
# - BLOCCO 2 (energie), tre colonne: C = (x,y), D = V(t) + V(theta), E = T(t) + traiettoria (theta,T);
# - i due blocchi hanno controlli di animazione indipendenti, inclusi avanzamento/arretramento di un frame.

import json
import uuid
import numpy as np
from scipy.integrate import solve_ivp
import ipywidgets as widgets
from IPython.display import display, Markdown, Math, HTML, Javascript, clear_output

# ---------- Derivazione simbolica dall'espressione scelta dallo studente ----------
def deriva_equazione_eulero_lagrange():
    """Restituisce l'equazione EL e, quando possibile, theta_ddot = F(...)."""
    dL_dtheta = sp.diff(L, theta)
    dL_dtheta_dot = sp.diff(L, theta_dot)

    # Derivata totale d/dt usando theta_dot=dtheta/dt e theta_ddot=d²theta/dt²
    d_dt_dL_dtheta_dot = (
        sp.diff(dL_dtheta_dot, theta) * theta_dot
        + sp.diff(dL_dtheta_dot, theta_dot) * theta_ddot
        + sp.diff(dL_dtheta_dot, t)
    )

    # L'equazione EL è lineare in theta_ddot: la isoliamo direttamente.
    EL_raw = sp.expand(d_dt_dL_dtheta_dot - dL_dtheta)
    coeff_ddot = sp.diff(EL_raw, theta_ddot)
    if sp.simplify(coeff_ddot) == 0:
        raise ValueError(
            "L'equazione ottenuta non permette di determinare theta_ddot. "
            "Controlla che T dipenda in modo appropriato da theta_dot."
        )

    resto = EL_raw.subs(theta_ddot, 0)
    accelerazione = sp.simplify(-resto / coeff_ddot)
    EL = sp.simplify(EL_raw)

    simboli_ammessi = {theta, theta_dot, t, m, l, g}
    simboli_extra = accelerazione.free_symbols - simboli_ammessi
    if simboli_extra:
        raise ValueError(
            "Nell'accelerazione compaiono simboli non previsti: "
            + ", ".join(sorted(map(str, simboli_extra)))
        )

    return EL, accelerazione


def limiti_verticali(y, margine=0.12):
    y = np.asarray(y, dtype=float)
    y = y[np.isfinite(y)]
    if y.size == 0:
        return [-1.0, 1.0]
    ymin, ymax = np.min(y), np.max(y)
    if np.isclose(ymin, ymax):
        amp = max(1.0, abs(ymin))
        return [float(ymin - 0.2 * amp), float(ymax + 0.2 * amp)]
    delta = ymax - ymin
    return [float(ymin - margine * delta), float(ymax + margine * delta)]


def valuta_lungo_traiettoria(expr, th, om, tt, m_val, l_val, g_val):
    """Valuta un'espressione simbolica lungo la traiettoria; restituisce None se non valutabile."""
    try:
        fun = sp.lambdify((theta, theta_dot, t, m, l, g), expr, modules="numpy")
        valori = np.asarray(fun(th, om, tt, m_val, l_val, g_val), dtype=float)
        if valori.ndim == 0:
            valori = np.full_like(np.asarray(th, dtype=float), float(valori))
        valori = np.broadcast_to(valori, np.asarray(th).shape).astype(float)
        if not np.all(np.isfinite(valori)):
            return None
        return valori
    except Exception:
        return None


def calcola_e_anima(_=None):
    with output:
        clear_output(wait=True)
        try:
            EL, accel_expr = deriva_equazione_eulero_lagrange()
            massa_nella_dinamica = (m in accel_expr.free_symbols)

            # Parametri numerici scelti dall'utente.
            # m_val serve sempre per valutare T e V; entra nella ODE solo
            # se compare realmente nell'accelerazione ricavata dallo studente.
            m_val = float(m_energy_widget.value)
            l_val = float(l_widget.value)
            g_val = float(g_widget.value)
            theta0 = np.deg2rad(float(theta0_widget.value))
            omega0 = float(omega0_widget.value)
            t_max = float(tmax_widget.value)
            n_frames = int(frames_widget.value)

            if m_val <= 0 or l_val <= 0 or g_val <= 0 or t_max <= 0:
                raise ValueError("m, ℓ, g e il tempo finale devono essere positivi.")

            # ---------- Rendering matematico ----------
            display(Markdown("### Equazione prodotta dalle espressioni inserite"))
            display(Markdown("L'equazione di Eulero–Lagrange calcolata è"))
            display(Math(r"\boxed{" + sp.latex(EL) + r" = 0}"))

            display(Markdown("Isolando l'accelerazione angolare, il codice ottiene"))
            display(Math(r"\boxed{\ddot{\theta} = " + sp.latex(accel_expr) + r"}"))

            display(Markdown("### Condizioni e parametri della simulazione"))
            display(Math(
                rf"\theta(0)={theta0_widget.value:.2f}^\circ,\qquad "
                rf"\dot{{\theta}}(0)={omega0_widget.value:.2f}\,\mathrm{{rad\,s^{{-1}}}}"
            ))
            if massa_nella_dinamica:
                display(Math(
                    rf"m={m_val:.2f}\,\mathrm{{kg}},\qquad "
                    rf"\ell={l_val:.2f}\,\mathrm{{m}},\qquad "
                    rf"g={g_val:.2f}\,\mathrm{{m\,s^{{-2}}}},\qquad "
                    rf"t_f={t_max:.2f}\,\mathrm{{s}},\qquad "
                    rf"N_{{\rm frame}}={n_frames}"
                ))
            else:
                display(Math(
                    rf"\ell={l_val:.2f}\,\mathrm{{m}},\qquad "
                    rf"g={g_val:.2f}\,\mathrm{{m\,s^{{-2}}}},\qquad "
                    rf"t_f={t_max:.2f}\,\mathrm{{s}},\qquad "
                    rf"N_{{\rm frame}}={n_frames}"
                ))

            display(HTML(
                '<div style="margin:8px 0 10px 0;padding:9px 11px;border-left:4px solid #f59e0b;'
                'background:#fffbeb;color:#78350f;border-radius:4px;font-size:13px;">'
                '<b>Nota:</b> la simulazione segue esattamente le espressioni di <b>T</b> e <b>V</b> inserite. '
                'Il programma non le sostituisce con quelle corrette del pendolo.</div>'
            ))

            # ---------- Equazione numerica ----------
            accel_fun = sp.lambdify(
                (theta, theta_dot, t, m, l, g), accel_expr, modules="numpy"
            )

            def sistema(t_num, stato):
                th, om = stato
                alpha = accel_fun(th, om, t_num, m_val, l_val, g_val)
                alpha = float(np.asarray(alpha))
                if not np.isfinite(alpha):
                    raise FloatingPointError(
                        f"Accelerazione non finita a t={t_num:.4g} s."
                    )
                return [om, alpha]

            soluzione = solve_ivp(
                sistema,
                (0.0, t_max),
                [theta0, omega0],
                method="DOP853",
                dense_output=True,
                rtol=1e-8,
                atol=1e-10,
            )

            if not soluzione.success:
                raise RuntimeError("Il solutore numerico segnala: " + soluzione.message)

            # Campionamento indipendente dal numero di frame.
            n_campioni = max(300, 5 * n_frames)
            tempi = np.linspace(0.0, t_max, n_campioni)
            stati = soluzione.sol(tempi)
            theta_t = np.asarray(stati[0], dtype=float)
            omega_t = np.asarray(stati[1], dtype=float)

            alpha_t = np.asarray(
                accel_fun(theta_t, omega_t, tempi, m_val, l_val, g_val),
                dtype=float,
            )
            if alpha_t.ndim == 0:
                alpha_t = np.full_like(theta_t, float(alpha_t))
            alpha_t = np.broadcast_to(alpha_t, theta_t.shape).astype(float)

            if (not np.all(np.isfinite(theta_t))
                    or not np.all(np.isfinite(omega_t))
                    or not np.all(np.isfinite(alpha_t))):
                raise FloatingPointError(
                    "La soluzione contiene valori non finiti. Controlla T, V o i parametri."
                )

            x_t = l_val * np.sin(theta_t)
            y_t = -l_val * np.cos(theta_t)

            # Energie lungo la traiettoria: usiamo sempre le espressioni dello studente.
            # T viene poi rappresentata come curva parametrica nel piano (theta mod 2pi, T).
            T_t = valuta_lungo_traiettoria(T, theta_t, omega_t, tempi, m_val, l_val, g_val)
            V_t = valuta_lungo_traiettoria(V, theta_t, omega_t, tempi, m_val, l_val, g_val)

            # Curva V(theta) su [0, 2pi]. È ben definita solo se V non dipende da theta_dot o t.
            theta_grid = np.linspace(0.0, 2.0 * np.pi, 361)
            simboli_v_non_config = V.free_symbols - {theta, m, l, g}
            V_curve = None
            V_curve_message = ""
            if simboli_v_non_config:
                V_curve_message = (
                    "V inserita non è una funzione della sola configurazione θ; "
                    "il paesaggio V(θ) non è quindi univocamente definito."
                )
            else:
                try:
                    vcurve_fun = sp.lambdify((theta, m, l, g), V, modules="numpy")
                    V_curve = np.asarray(vcurve_fun(theta_grid, m_val, l_val, g_val), dtype=float)
                    if V_curve.ndim == 0:
                        V_curve = np.full_like(theta_grid, float(V_curve))
                    V_curve = np.broadcast_to(V_curve, theta_grid.shape).astype(float)
                    if not np.all(np.isfinite(V_curve)):
                        V_curve = None
                        V_curve_message = "V(θ) contiene valori non finiti nell'intervallo [0, 2π]."
                except Exception:
                    V_curve = None
                    V_curve_message = "Non è stato possibile valutare numericamente V(θ) su [0, 2π]."

            # Coordinate del sistema nello spazio delle configurazioni S^1.
            theta_mod = np.mod(theta_t, 2.0 * np.pi)

            frame_idx = np.unique(
                np.linspace(0, len(tempi) - 1, n_frames, dtype=int)
            )

            def lista(a):
                return np.round(np.asarray(a, dtype=float), 8).tolist()

            dati = {
                "t": lista(tempi),
                "theta": lista(theta_t),
                "thetaMod": lista(theta_mod),
                "omega": lista(omega_t),
                "alpha": lista(alpha_t),
                "x": lista(x_t),
                "y": lista(y_t),
                "frameIdx": [int(v) for v in frame_idx],
                "l": l_val,
                "tmax": t_max,
                "thetaLim": limiti_verticali(theta_t),
                "omegaLim": limiti_verticali(omega_t),
                "alphaLim": limiti_verticali(alpha_t),
                "thetaGrid": lista(theta_grid),
                "Vcurve": None if V_curve is None else lista(V_curve),
                "VcurveLim": [-1.0, 1.0] if V_curve is None else limiti_verticali(V_curve),
                "VcurveMessage": V_curve_message,
                "Ttraj": None if T_t is None else lista(T_t),
                "Tlim": [-1.0, 1.0] if T_t is None else limiti_verticali(T_t),
                "Tmessage": "" if T_t is not None else "Non è stato possibile valutare T lungo la traiettoria.",
                "Vtraj": None if V_t is None else lista(V_t),
                "VtrajLim": [-1.0, 1.0] if V_t is None else limiti_verticali(V_t),
                "VtrajMessage": "" if V_t is not None else "Non è stato possibile valutare V lungo la traiettoria.",
            }

            uid = "pendolo_" + uuid.uuid4().hex[:10]

            html = f"""
            <div id="{uid}_box" style="margin:12px 0;padding:12px;border:1px solid #d1d5db;
                 border-radius:10px;background:#ffffff;color:#111827;">
              <div style="font-weight:700;font-size:18px;margin:0 0 12px 2px;">
                Evoluzione dinamica prodotta dall'equazione dello studente
              </div>

              <!-- ====================================================== -->
              <!-- BLOCCO 1: CINEMATICA                                   -->
              <!-- ====================================================== -->
              <div style="font-weight:800;font-size:16px;margin:4px 0 8px 2px;color:#0d47a1;">
                Blocco 1 — Cinematica
              </div>

              <div style="display:grid;grid-template-columns:repeat(auto-fit,minmax(410px,1fr));gap:12px;align-items:start;">

                <div style="display:flex;flex-direction:column;gap:10px;">
                  <div style="font-weight:700;font-size:13px;color:#4b5563;margin-left:2px;">A</div>
                  <canvas id="{uid}_p_kin" style="width:100%;height:220px;border:1px solid #e5e7eb;border-radius:6px;background:#fff;"></canvas>
                  <canvas id="{uid}_th" style="width:100%;height:185px;border:1px solid #e5e7eb;border-radius:6px;background:#fff;"></canvas>
                </div>

                <div style="display:flex;flex-direction:column;gap:10px;">
                  <div style="font-weight:700;font-size:13px;color:#4b5563;margin-left:2px;">B</div>
                  <canvas id="{uid}_om" style="width:100%;height:220px;border:1px solid #e5e7eb;border-radius:6px;background:#fff;"></canvas>
                  <canvas id="{uid}_al" style="width:100%;height:185px;border:1px solid #e5e7eb;border-radius:6px;background:#fff;"></canvas>
                </div>
              </div>

              <div style="display:flex;align-items:center;gap:7px;margin-top:12px;flex-wrap:wrap;padding:8px 0 3px 0;">
                <button id="{uid}_kin_prev" title="Frame precedente">◀︎</button>
                <button id="{uid}_kin_play" title="Play">▶︎</button>
                <button id="{uid}_kin_pause" title="Pausa">⏸︎</button>
                <button id="{uid}_kin_next" title="Frame successivo">▶︎|</button>
                <span style="margin-left:6px;">velocità:</span>
                <select id="{uid}_kin_speed">
                  <option value="0.5">0.5×</option>
                  <option value="1" selected>1×</option>
                  <option value="2">2×</option>
                  <option value="4">4×</option>
                </select>
                <input id="{uid}_kin_slider" type="range" min="0" max="{len(frame_idx)-1}" value="0"
                       step="1" style="flex:1;min-width:260px;">
                <span id="{uid}_kin_label" style="min-width:115px;text-align:right;font-variant-numeric:tabular-nums;"></span>
              </div>

              <div style="height:1px;background:#d1d5db;margin:18px 0 16px 0;"></div>

              <!-- ====================================================== -->
              <!-- BLOCCO 2: ENERGIE                                      -->
              <!-- ====================================================== -->
              <div style="font-weight:800;font-size:16px;margin:0 0 8px 2px;color:#0d47a1;">
                Blocco 2 — Energie
              </div>

              <div style="display:grid;grid-template-columns:repeat(auto-fit,minmax(300px,1fr));gap:12px;align-items:stretch;">

                <div style="display:flex;flex-direction:column;gap:10px;">
                  <div style="font-weight:700;font-size:13px;color:#4b5563;margin-left:2px;">C</div>
                  <canvas id="{uid}_p_ene" style="width:100%;height:380px;border:1px solid #e5e7eb;border-radius:6px;background:#fff;"></canvas>
                </div>

                <div style="display:flex;flex-direction:column;gap:10px;">
                  <div style="font-weight:700;font-size:13px;color:#4b5563;margin-left:2px;">D</div>
                  <canvas id="{uid}_Vt" style="width:100%;height:185px;border:1px solid #e5e7eb;border-radius:6px;background:#fff;"></canvas>
                  <canvas id="{uid}_V" style="width:100%;height:185px;border:1px solid #e5e7eb;border-radius:6px;background:#fff;"></canvas>
                </div>

                <div style="display:flex;flex-direction:column;gap:10px;">
                  <div style="font-weight:700;font-size:13px;color:#4b5563;margin-left:2px;">E</div>
                  <canvas id="{uid}_Tt" style="width:100%;height:185px;border:1px solid #e5e7eb;border-radius:6px;background:#fff;"></canvas>
                  <canvas id="{uid}_T" style="width:100%;height:185px;border:1px solid #e5e7eb;border-radius:6px;background:#fff;"></canvas>
                </div>
              </div>

              <div style="display:flex;align-items:center;gap:7px;margin-top:12px;flex-wrap:wrap;padding:8px 0 3px 0;">
                <button id="{uid}_ene_prev" title="Frame precedente">◀︎</button>
                <button id="{uid}_ene_play" title="Play">▶︎</button>
                <button id="{uid}_ene_pause" title="Pausa">⏸︎</button>
                <button id="{uid}_ene_next" title="Frame successivo">▶︎|</button>
                <span style="margin-left:6px;">velocità:</span>
                <select id="{uid}_ene_speed">
                  <option value="0.5">0.5×</option>
                  <option value="1" selected>1×</option>
                  <option value="2">2×</option>
                  <option value="4">4×</option>
                </select>
                <input id="{uid}_ene_slider" type="range" min="0" max="{len(frame_idx)-1}" value="0"
                       step="1" style="flex:1;min-width:260px;">
                <span id="{uid}_ene_label" style="min-width:115px;text-align:right;font-variant-numeric:tabular-nums;"></span>
              </div>

              <div style="font-size:12px;color:#4b5563;margin-top:8px;line-height:1.4;">
                I due blocchi hanno controlli indipendenti. Le curve sono calcolate in Python una sola volta;
                le animazioni sono disegnate al volo dal browser. Il grafico T(θ) è rappresentato come
                traiettoria parametrica nel piano (θ,T), perché T non è in generale una funzione univoca del solo angolo.
              </div>
            </div>
            """
            display(HTML(html))

            js_data = json.dumps(dati, separators=(",", ":"))
            js = f"""
            (() => {{
              const D = {js_data};
              const uid = {json.dumps(uid)};
              const byId = s => document.getElementById(uid+s);
              const cpKin=byId('_p_kin'), cpEne=byId('_p_ene');
              const cth=byId('_th'), com=byId('_om'), cal=byId('_al');
              const cV=byId('_V'), cT=byId('_T'), cVt=byId('_Vt'), cTt=byId('_Tt');
              const frameIdx=D.frameIdx;
              let kKin=0, timerKin=null, speedKin=1;
              let kEne=0, timerEne=null, speedEne=1;
              const baseInterval=Math.max(18, 1000 * D.tmax / Math.max(1,frameIdx.length-1));

              const FONT='-apple-system,BlinkMacSystemFont,"Segoe UI",Roboto,Arial,sans-serif';

              function fmt(v) {{
                if(!Number.isFinite(v)) return '—';
                if(Math.abs(v) > 0 && (Math.abs(v) >= 10000 || Math.abs(v) < 0.001))
                  return v.toExponential(2);
                let s=v.toFixed(2);
                s=s.replace(/[.]00$/,'').replace(/([.][0-9])0$/,'$1');
                if(s==='-0') s='0';
                return s;
              }}

              function prepare(canvas) {{
                const rect=canvas.getBoundingClientRect();
                const W=Math.max(280,rect.width), H=Math.max(165,rect.height);
                const dpr=Math.min(window.devicePixelRatio || 1, 2.5);
                const pw=Math.round(W*dpr), ph=Math.round(H*dpr);
                if(canvas.width!==pw || canvas.height!==ph) {{canvas.width=pw; canvas.height=ph;}}
                const ctx=canvas.getContext('2d');
                ctx.setTransform(dpr,0,0,dpr,0,0);
                ctx.imageSmoothingEnabled=true;
                ctx.clearRect(0,0,W,H);
                ctx.fillStyle='#ffffff';ctx.fillRect(0,0,W,H);
                return {{ctx,W,H}};
              }}

              function mapX(x,xmin,xmax,W,L,R) {{return L+(x-xmin)/(xmax-xmin)*(W-L-R);}}
              function mapY(y,ymin,ymax,H,T,B) {{return H-B-(y-ymin)/(ymax-ymin)*(H-T-B);}}

              function bubble(ctx,px,py,symbol,value,unit,W,H) {{
                ctx.save();
                const valueText=fmt(value)+(unit?' '+unit:'');
                ctx.font='600 12px '+FONT;
                const valueW=ctx.measureText(valueText).width;
                ctx.font='italic 17px Georgia,"Times New Roman",serif';
                const symbolW=ctx.measureText(symbol).width;
                const padX=9, boxH=48, boxW=Math.max(74,Math.max(valueW,symbolW)+2*padX);
                let x=px+9, y=py-boxH-9;
                if(x+boxW>W-5) x=px-boxW-9;
                if(y<5) y=py+9;
                if(y+boxH>H-5) y=H-boxH-5;
                ctx.fillStyle='rgba(255,255,255,0.98)';
                ctx.strokeStyle='#c62828';ctx.lineWidth=1.2;
                ctx.beginPath();ctx.roundRect(x,y,boxW,boxH,7);ctx.fill();ctx.stroke();
                ctx.textAlign='center';ctx.textBaseline='middle';
                ctx.fillStyle='#7f1d1d';ctx.font='italic 17px Georgia,"Times New Roman",serif';
                ctx.fillText(symbol,x+boxW/2,y+14.5);
                ctx.fillStyle='#111827';ctx.font='600 12px '+FONT;
                ctx.fillText(valueText,x+boxW/2,y+34);
                ctx.restore();
              }}

              function axes(canvas,xmin,xmax,ymin,ymax,title,xlabel,ylabel) {{
                const P=prepare(canvas); const ctx=P.ctx,W=P.W,H=P.H;
                const L=78,R=16,T=39,B=47;
                ctx.fillStyle='#111827';ctx.font='700 14px '+FONT;ctx.textAlign='center';ctx.textBaseline='alphabetic';
                ctx.fillText(title,W/2,21);
                ctx.font='12px '+FONT;

                ctx.strokeStyle='#e5e7eb';ctx.lineWidth=1;
                for(let j=0;j<=4;j++) {{
                  const yy=T+j*(H-T-B)/4;
                  ctx.beginPath();ctx.moveTo(L,yy);ctx.lineTo(W-R,yy);ctx.stroke();
                  const v=ymax-j*(ymax-ymin)/4;
                  ctx.fillStyle='#374151';ctx.textAlign='right';ctx.textBaseline='middle';
                  ctx.fillText(fmt(v),L-9,yy);
                }}
                for(let j=0;j<=4;j++) {{
                  const xx=L+j*(W-L-R)/4;
                  ctx.beginPath();ctx.moveTo(xx,T);ctx.lineTo(xx,H-B);ctx.stroke();
                  const v=xmin+j*(xmax-xmin)/4;
                  ctx.fillStyle='#374151';ctx.textAlign='center';ctx.textBaseline='top';
                  ctx.fillText(fmt(v),xx,H-B+8);
                }}
                ctx.strokeStyle='#374151';ctx.lineWidth=1.15;ctx.strokeRect(L,T,W-L-R,H-T-B);
                ctx.fillStyle='#111827';ctx.font='12px '+FONT;ctx.textAlign='center';ctx.textBaseline='alphabetic';
                ctx.fillText(xlabel,(L+W-R)/2,H-8);
                ctx.save();ctx.translate(20,(T+H-B)/2);ctx.rotate(-Math.PI/2);
                ctx.fillText(ylabel,0,0);ctx.restore();
                return {{ctx,W,H,L,R,T,B}};
              }}

              function plotSeries(canvas,arr,lim,title,ylabel,valueName,unit,i) {{
                const A=axes(canvas,0,D.tmax,lim[0],lim[1],title,'t [s]',ylabel);
                const ctx=A.ctx;
                ctx.strokeStyle='#1565c0';ctx.lineWidth=2.15;ctx.lineJoin='round';ctx.lineCap='round';
                ctx.beginPath();
                for(let j=0;j<=i;j++) {{
                  const xx=mapX(D.t[j],0,D.tmax,A.W,A.L,A.R);
                  const yy=mapY(arr[j],lim[0],lim[1],A.H,A.T,A.B);
                  if(j===0)ctx.moveTo(xx,yy);else ctx.lineTo(xx,yy);
                }}
                ctx.stroke();
                const px=mapX(D.t[i],0,D.tmax,A.W,A.L,A.R);
                const py=mapY(arr[i],lim[0],lim[1],A.H,A.T,A.B);
                ctx.fillStyle='#c62828';ctx.beginPath();ctx.arc(px,py,4.5,0,2*Math.PI);ctx.fill();
                bubble(ctx,px,py,valueName,arr[i],unit,A.W,A.H);
              }}

              function pendulum(canvas,i) {{
                const P=prepare(canvas); const ctx=P.ctx,W=P.W,H=P.H;
                const L=38,R=18,T=40,B=38;
                const rmax=1.25*D.l;
                const plotW=W-L-R, plotH=H-T-B;
                const scale=Math.min(plotW/(2*rmax),plotH/(2*rmax));
                const cx=L+plotW/2, cy=T+plotH/2;
                const X=x=>cx+x*scale, Y=y=>cy-y*scale;

                ctx.fillStyle='#111827';ctx.font='700 14px '+FONT;ctx.textAlign='center';
                ctx.fillText('Moto nel piano (x,y)',W/2,21);

                ctx.strokeStyle='#9ca3af';ctx.lineWidth=1;
                ctx.beginPath();ctx.moveTo(X(-rmax),Y(0));ctx.lineTo(X(rmax),Y(0));ctx.stroke();
                ctx.beginPath();ctx.moveTo(X(0),Y(-rmax));ctx.lineTo(X(0),Y(rmax));ctx.stroke();
                ctx.fillStyle='#374151';ctx.font='12px '+FONT;
                ctx.fillText('x',W-18,Y(0)-6);ctx.fillText('y',X(0)+11,T+4);

                ctx.save();ctx.setLineDash([4,4]);ctx.strokeStyle='#c7cdd4';ctx.lineWidth=1;
                ctx.beginPath();
                for(let a=0;a<=120;a++) {{
                  const ph=-Math.PI+2*Math.PI*a/120;
                  const xx=X(D.l*Math.sin(ph)),yy=Y(-D.l*Math.cos(ph));
                  if(a===0)ctx.moveTo(xx,yy);else ctx.lineTo(xx,yy);
                }}
                ctx.stroke();ctx.restore();

                const start=Math.max(0,i-Math.max(20,Math.floor(D.t.length/5)));
                ctx.strokeStyle='rgba(21,101,192,0.28)';ctx.lineWidth=1.5;ctx.beginPath();
                for(let j=start;j<=i;j++) {{
                  const xx=X(D.x[j]),yy=Y(D.y[j]);
                  if(j===start)ctx.moveTo(xx,yy);else ctx.lineTo(xx,yy);
                }}
                ctx.stroke();

                ctx.strokeStyle='#263238';ctx.lineWidth=3;ctx.beginPath();
                ctx.moveTo(X(0),Y(0));ctx.lineTo(X(D.x[i]),Y(D.y[i]));ctx.stroke();
                ctx.fillStyle='#111827';ctx.beginPath();ctx.arc(X(0),Y(0),4,0,2*Math.PI);ctx.fill();
                ctx.fillStyle='#1565c0';ctx.beginPath();ctx.arc(X(D.x[i]),Y(D.y[i]),9,0,2*Math.PI);ctx.fill();

                ctx.fillStyle='#ffffff';ctx.strokeStyle='#9ca3af';ctx.lineWidth=1;
                const bw=102,bh=28,bx=W-bw-15,by=35;
                ctx.fillRect(bx,by,bw,bh);ctx.strokeRect(bx,by,bw,bh);
                ctx.fillStyle='#111827';ctx.font='600 12px '+FONT;ctx.textAlign='center';ctx.textBaseline='middle';
                ctx.fillText('t = '+fmt(D.t[i])+' s',bx+bw/2,by+bh/2+0.5);
                ctx.textBaseline='alphabetic';ctx.fillText('x [m]',W/2,H-8);
              }}

              function potential(i) {{
                if(D.Vcurve===null) {{
                  const P=prepare(cV),ctx=P.ctx,W=P.W,H=P.H;
                  ctx.fillStyle='#111827';ctx.font='700 14px '+FONT;ctx.textAlign='center';ctx.fillText('Energia potenziale V(θ)',W/2,22);
                  ctx.fillStyle='#6b7280';ctx.font='13px '+FONT;ctx.textAlign='center';
                  const msg=D.VcurveMessage || 'V(θ) non disponibile.';
                  const words=msg.split(' ');let line='',yy=H/2-12;
                  for(const w of words){{const test=line+w+' ';if(ctx.measureText(test).width>W-80){{ctx.fillText(line,W/2,yy);line=w+' ';yy+=20;}}else line=test;}}
                  ctx.fillText(line,W/2,yy);return;
                }}
                const A=axes(cV,0,2*Math.PI,D.VcurveLim[0],D.VcurveLim[1],'Energia potenziale V(θ)','θ [rad]','V [J]');
                const ctx=A.ctx;
                ctx.strokeStyle='#1565c0';ctx.lineWidth=2.15;ctx.beginPath();
                for(let j=0;j<D.thetaGrid.length;j++){{
                  const xx=mapX(D.thetaGrid[j],0,2*Math.PI,A.W,A.L,A.R);
                  const yy=mapY(D.Vcurve[j],D.VcurveLim[0],D.VcurveLim[1],A.H,A.T,A.B);
                  if(j===0)ctx.moveTo(xx,yy);else ctx.lineTo(xx,yy);
                }}ctx.stroke();
                const th=D.thetaMod[i];
                // interpolazione lineare sulla curva V(theta)
                const u=th/(2*Math.PI)*(D.thetaGrid.length-1);const j0=Math.floor(u),j1=Math.min(D.thetaGrid.length-1,j0+1),a=u-j0;
                const vv=(1-a)*D.Vcurve[j0]+a*D.Vcurve[j1];
                const px=mapX(th,0,2*Math.PI,A.W,A.L,A.R),py=mapY(vv,D.VcurveLim[0],D.VcurveLim[1],A.H,A.T,A.B);
                ctx.fillStyle='#c62828';ctx.beginPath();ctx.arc(px,py,4.5,0,2*Math.PI);ctx.fill();
                bubble(ctx,px,py,'V',vv,'J',A.W,A.H);
              }}

              function kinetic(i) {{
                if(D.Ttraj===null) {{
                  const P=prepare(cT),ctx=P.ctx,W=P.W,H=P.H;
                  ctx.fillStyle='#111827';ctx.font='700 14px '+FONT;ctx.textAlign='center';ctx.fillText('Energia cinetica T(θ) lungo la traiettoria',W/2,22);
                  ctx.fillStyle='#6b7280';ctx.font='13px '+FONT;ctx.fillText(D.Tmessage || 'T non disponibile lungo la traiettoria.',W/2,H/2);return;
                }}
                const A=axes(cT,0,2*Math.PI,D.Tlim[0],D.Tlim[1],'Energia cinetica T(θ)','θ [rad]','T [J]');
                const ctx=A.ctx;

                // Curva parametrica t -> (theta mod 2pi, T). Non assumiamo che T sia una funzione univoca di theta.
                ctx.strokeStyle='#1565c0';ctx.lineWidth=2.0;ctx.lineJoin='round';ctx.lineCap='round';
                ctx.beginPath();
                let started=false;
                for(let j=0;j<=i;j++){{
                  const th=D.thetaMod[j];
                  const xx=mapX(th,0,2*Math.PI,A.W,A.L,A.R);
                  const yy=mapY(D.Ttraj[j],D.Tlim[0],D.Tlim[1],A.H,A.T,A.B);
                  const wrapJump=(j>0 && Math.abs(D.thetaMod[j]-D.thetaMod[j-1])>Math.PI);
                  if(!started || wrapJump){{ctx.moveTo(xx,yy);started=true;}} else ctx.lineTo(xx,yy);
                }}
                ctx.stroke();

                // Piccoli punti già visitati: rendono visibili eventuali valori diversi di T allo stesso angolo.
                ctx.fillStyle='rgba(21,101,192,0.28)';
                const stride=Math.max(1,Math.floor((i+1)/140));
                for(let j=0;j<=i;j+=stride){{
                  const xx=mapX(D.thetaMod[j],0,2*Math.PI,A.W,A.L,A.R);
                  const yy=mapY(D.Ttraj[j],D.Tlim[0],D.Tlim[1],A.H,A.T,A.B);
                  ctx.beginPath();ctx.arc(xx,yy,1.6,0,2*Math.PI);ctx.fill();
                }}

                const th=D.thetaMod[i];
                const px=mapX(th,0,2*Math.PI,A.W,A.L,A.R);
                const py=mapY(D.Ttraj[i],D.Tlim[0],D.Tlim[1],A.H,A.T,A.B);
                ctx.fillStyle='#c62828';ctx.beginPath();ctx.arc(px,py,4.8,0,2*Math.PI);ctx.fill();
                bubble(ctx,px,py,'T',D.Ttraj[i],'J',A.W,A.H);
              }}

              function energyTime(canvas,arr,lim,title,ylabel,symbol,unit,i,message) {{
                if(arr===null) {{
                  const P=prepare(canvas),ctx=P.ctx,W=P.W,H=P.H;
                  ctx.fillStyle='#111827';ctx.font='700 14px '+FONT;ctx.textAlign='center';ctx.fillText(title,W/2,22);
                  ctx.fillStyle='#6b7280';ctx.font='13px '+FONT;ctx.fillText(message || 'Energia non disponibile lungo la traiettoria.',W/2,H/2);return;
                }}
                const A=axes(canvas,0,D.tmax,lim[0],lim[1],title,'t [s]',ylabel);
                const ctx=A.ctx;
                ctx.strokeStyle='#1565c0';ctx.lineWidth=2.15;ctx.lineJoin='round';ctx.lineCap='round';ctx.beginPath();
                for(let j=0;j<=i;j++) {{
                  const xx=mapX(D.t[j],0,D.tmax,A.W,A.L,A.R);
                  const yy=mapY(arr[j],lim[0],lim[1],A.H,A.T,A.B);
                  if(j===0)ctx.moveTo(xx,yy);else ctx.lineTo(xx,yy);
                }}
                ctx.stroke();
                const px=mapX(D.t[i],0,D.tmax,A.W,A.L,A.R);
                const py=mapY(arr[i],lim[0],lim[1],A.H,A.T,A.B);
                ctx.fillStyle='#c62828';ctx.beginPath();ctx.arc(px,py,4.8,0,2*Math.PI);ctx.fill();
                bubble(ctx,px,py,symbol,arr[i],unit,A.W,A.H);
              }}

              function drawKinematics() {{
                const i=frameIdx[kKin];
                pendulum(cpKin,i);
                plotSeries(cth,D.theta,D.thetaLim,'Posizione angolare θ(t)','θ [rad]','θ','rad',i);
                plotSeries(com,D.omega,D.omegaLim,'Velocità angolare θ̇(t)','θ̇ [rad/s]','θ̇','rad/s',i);
                plotSeries(cal,D.alpha,D.alphaLim,'Accelerazione angolare θ̈(t)','θ̈ [rad/s²]','θ̈','rad/s²',i);
                byId('_kin_slider').value=kKin;
                byId('_kin_label').textContent='t = '+fmt(D.t[i])+' s';
              }}

              function drawEnergies() {{
                const i=frameIdx[kEne];
                pendulum(cpEne,i);
                energyTime(cVt,D.Vtraj,D.VtrajLim,'Energia potenziale V(t)','V [J]','V','J',i,D.VtrajMessage);
                potential(i);
                energyTime(cTt,D.Ttraj,D.Tlim,'Energia cinetica T(t)','T [J]','T','J',i,D.Tmessage);
                kinetic(i);
                byId('_ene_slider').value=kEne;
                byId('_ene_label').textContent='t = '+fmt(D.t[i])+' s';
              }}

              function stopKin() {{
                if(timerKin!==null){{clearTimeout(timerKin);timerKin=null;}}
              }}
              function tickKin() {{
                stopKin();
                timerKin=setTimeout(()=>{{
                  kKin=(kKin+1)%frameIdx.length;
                  drawKinematics();
                  tickKin();
                }},baseInterval/speedKin);
              }}

              function stopEne() {{
                if(timerEne!==null){{clearTimeout(timerEne);timerEne=null;}}
              }}
              function tickEne() {{
                stopEne();
                timerEne=setTimeout(()=>{{
                  kEne=(kEne+1)%frameIdx.length;
                  drawEnergies();
                  tickEne();
                }},baseInterval/speedEne);
              }}

              // Comandi BLOCCO 1 — cinematica
              byId('_kin_play').onclick=()=>tickKin();
              byId('_kin_pause').onclick=()=>stopKin();
              byId('_kin_prev').onclick=()=>{{stopKin();kKin=Math.max(0,kKin-1);drawKinematics();}};
              byId('_kin_next').onclick=()=>{{stopKin();kKin=Math.min(frameIdx.length-1,kKin+1);drawKinematics();}};
              byId('_kin_speed').onchange=e=>{{
                speedKin=parseFloat(e.target.value)||1;
                if(timerKin!==null)tickKin();
              }};
              byId('_kin_slider').oninput=e=>{{
                stopKin();kKin=parseInt(e.target.value);drawKinematics();
              }};

              // Comandi BLOCCO 2 — energie
              byId('_ene_play').onclick=()=>tickEne();
              byId('_ene_pause').onclick=()=>stopEne();
              byId('_ene_prev').onclick=()=>{{stopEne();kEne=Math.max(0,kEne-1);drawEnergies();}};
              byId('_ene_next').onclick=()=>{{stopEne();kEne=Math.min(frameIdx.length-1,kEne+1);drawEnergies();}};
              byId('_ene_speed').onchange=e=>{{
                speedEne=parseFloat(e.target.value)||1;
                if(timerEne!==null)tickEne();
              }};
              byId('_ene_slider').oninput=e=>{{
                stopEne();kEne=parseInt(e.target.value);drawEnergies();
              }};

              window.addEventListener('resize',()=>{{
                drawKinematics();
                drawEnergies();
              }},{{passive:true}});

              drawKinematics();
              drawEnergies();
              return 'ok';
            }})()
            """

            try:
                from google.colab import output as colab_output
                colab_output.eval_js(js)
            except Exception:
                display(Javascript(js))

        except Exception as errore:
            display(Markdown(f"""
### La simulazione non può essere avviata

Il codice ha prodotto il seguente messaggio:

```text
{type(errore).__name__}: {errore}
```

Questo non significa necessariamente che il programma sia guasto. Può indicare che le espressioni scelte per $T$ e $V$ non generano un'equazione differenziale del secondo ordine ben definita oppure diventano singolari con i parametri selezionati. Torna alla cella 3, controlla le formule e rieseguila.
"""))


# ---------- Interfaccia ----------
style = {'description_width': '185px'}
layout = widgets.Layout(width='360px')

theta0_widget = widgets.FloatText(
    value=30.0, description='θ(0) [°]:', style=style, layout=layout,
)
omega0_widget = widgets.FloatText(
    value=0.0,
    description='theta_punto(0) [rad/s]:',
    style=style,
    layout=layout,
)
m_energy_widget = widgets.FloatText(
    value=1.0, description='m [kg]:', style={'description_width': '70px'},
    layout=widgets.Layout(width='180px'),
)
m_energy_label = widgets.HTML(
    value='<b>Solo per i grafici energetici</b> &nbsp; (<i>T</i> e <i>V</i>)',
    layout=widgets.Layout(width='300px')
)
l_widget = widgets.FloatText(
    value=1.0, description='lunghezza l [m]:', style=style, layout=layout,
)
g_widget = widgets.FloatText(
    value=9.81, description='gravità g [m/s²]:', style=style, layout=layout,
)
tmax_widget = widgets.FloatSlider(
    value=10.0, min=2.0, max=30.0, step=1.0, description='tempo finale [s]:',
    style=style, layout=widgets.Layout(width='520px'),
    continuous_update=True, readout=True, readout_format='.1f',
)
tmax_manual_widget = widgets.BoundedFloatText(
    value=10.0, min=2.0, max=30.0, step=0.1, description='da tastiera:',
    style={'description_width': '95px'}, layout=widgets.Layout(width='210px'),
)
tmax_link = widgets.jslink(
    (tmax_widget, 'value'),
    (tmax_manual_widget, 'value')
)

frames_widget = widgets.IntSlider(
    value=60, min=30, max=120, step=10, description='fotogrammi:',
    style=style, layout=widgets.Layout(width='520px'),
    continuous_update=True, readout=True,
)
frames_manual_widget = widgets.BoundedIntText(
    value=60, min=30, max=120, step=1, description='da tastiera:',
    style={'description_width': '95px'}, layout=widgets.Layout(width='210px'),
)
frames_link = widgets.jslink(
    (frames_widget, 'value'),
    (frames_manual_widget, 'value')
)

pulsante = widgets.Button(
    description='Deriva, risolvi e anima', button_style='primary', icon='play',
    layout=widgets.Layout(width='250px', height='42px'),
)
pulsante.on_click(calcola_e_anima)
output = widgets.Output()

intro = widgets.HTML(
    """
    <div style="
        padding:14px 16px;border-left:6px solid #1565c0;border:1px solid #90caf9;
        border-left-width:6px;border-radius:6px;background-color:#eef5ff;color:#111827;
        font-size:15px;line-height:1.45;">
      <div style="font-size:17px;font-weight:700;color:#0d47a1;margin-bottom:5px;">
        Inserisci le condizioni iniziali e i parametri
      </div>
      L'angolo iniziale è richiesto in gradi per comodità; nel calcolo viene convertito in radianti.<br>
      Dopo ogni modifica di <b>T</b> o <b>V</b> nella cella 3, riesegui prima la cella 3 e poi premi
      <b>Deriva, risolvi e anima</b>.<br>
      Puoi scegliere da <b>30 a 120 fotogrammi</b>. I grafici sono disegnati direttamente dal browser.<br>
      Il <b>Blocco 1 — Cinematica</b> raccoglie il moto nel piano, θ(t), θ̇(t) e θ̈(t) in due colonne.<br>
      Il <b>Blocco 2 — Energie</b> raccoglie il moto nel piano e, separatamente, V(t), V(θ), T(t) e la traiettoria nel piano (θ,T).<br>
      Ogni blocco dispone di una <b>barra di animazione indipendente</b>, con Play, Pausa, frame precedente, frame successivo, velocità e cursore temporale.
    </div>
    """
)

initial_block = widgets.VBox([
    theta0_widget,
    omega0_widget,
    widgets.HBox([m_energy_label, m_energy_widget]),
], layout=widgets.Layout(width='500px'))

parameter_block = widgets.VBox(
    [l_widget, g_widget],
    layout=widgets.Layout(width='380px', margin='0 0 0 12px')
)

initial_and_parameters = widgets.HBox(
    [initial_block, parameter_block],
    layout=widgets.Layout(
        width='100%',
        flex_flow='row wrap',
        align_items='flex-start'
    )
)

display(
    intro,
    initial_and_parameters,
    widgets.HBox(
        [tmax_widget, tmax_manual_widget],
        layout=widgets.Layout(flex_flow='row wrap')
    ),
    widgets.HBox(
        [frames_widget, frames_manual_widget],
        layout=widgets.Layout(flex_flow='row wrap')
    ),
    pulsante,
    output,
)
````

# Dallo spazio delle configurazioni allo spazio degli stati

## **Attività 4 — Una configurazione identifica completamente lo stato del pendolo?**

Nella simulazione precedente hai osservato l'evoluzione temporale di $\theta$ e di $\dot\theta$.

La coordinata generalizzata $\theta$ individua la **configurazione geometrica** del pendolo.

Per mettere a fuoco la differenza tra configurazione ed evoluzione, immagina ora di **congelare la simulazione in un preciso istante** e di conservarne una sola **istantanea**: una fotografia del sistema in quell'istante. Da questa immagine puoi riconoscere la posizione del pendolo e quindi il valore di $\theta$.

Supponi ora di avere due pendoli identici e che, nelle due istantanee, essi abbiano esattamente lo stesso valore di $\theta$. Le due configurazioni geometriche sono quindi indistinguibili.

✍️ CARTA E PENNA
</span>

Rifletti brevemente, su carta: **quali informazioni sono sufficienti per prevedere univocamente come evolveranno?**
<br><br>
La prossima esplorazione numerica serve a mettere alla prova la tua deduzione. Nella cella interattiva seguente premi **Mostra informazione mancante**.

Poi:

- scegli lo stesso angolo iniziale per i due pendoli identici **A** e **B**;
- assegna liberamente due diversi valori alla ulteriore grandezza dinamica appena individuata;
- **prima di simulare**, traccia su carta uno schizzo qualitativo delle due curve previste $\theta_A(t)$ e $\theta_B(t)$ nello stesso piano $(t,\theta)$;
- solo dopo avvia la simulazione e confronta il risultato con la tua previsione.

La simulazione continuerà a utilizzare **l'equazione del moto ottenuta dalle espressioni di $T$ e $V$ che hai inserito**: il notebook non sostituisce il tuo modello con quello atteso.


````python
#@title Attività 4.1 — Stessa configurazione, condizioni dinamiche diverse? { display-mode: "form" }

import json
import uuid
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
import ipywidgets as widgets
from IPython.display import display, Markdown, Math, HTML, clear_output

# ---------------------------------------------------------------------
# Questa attività NON verifica se T e V sono quelle corrette del pendolo.
# Usa la Lagrangiana costruita dallo studente e ne esplora le conseguenze.
# ---------------------------------------------------------------------

def _att41_accelerazione_da_L():
    dL_dtheta = sp.diff(L, theta)
    dL_dtheta_dot = sp.diff(L, theta_dot)

    d_dt = (
        sp.diff(dL_dtheta_dot, theta) * theta_dot
        + sp.diff(dL_dtheta_dot, theta_dot) * theta_ddot
        + sp.diff(dL_dtheta_dot, t)
    )

    eq = sp.expand(d_dt - dL_dtheta)
    coeff = sp.simplify(sp.diff(eq, theta_ddot))

    if coeff == 0:
        raise ValueError(
            "Il modello inserito non permette di isolare l'accelerazione angolare. "
            "Torna ai tuoi calcoli di T e V e verifica che l'equazione del moto "
            "sia effettivamente del secondo ordine."
        )

    accel = sp.simplify(-eq.subs(theta_ddot, 0) / coeff)

    simboli_ammessi = {theta, theta_dot, t, m, l, g}
    extra = accel.free_symbols - simboli_ammessi
    if extra:
        raise ValueError(
            "Nell'accelerazione compaiono simboli non previsti: "
            + ", ".join(sorted(map(str, extra)))
        )

    return accel


def _att41_valore_widget_precedente(nome, default):
    obj = globals().get(nome, None)
    try:
        return float(obj.value)
    except Exception:
        return float(default)


def _att41_riga_controllo(label_html, slider, text, label_width='145px'):
    # Colab non renderizza in modo affidabile LaTeX dentro i widget.
    # Usiamo una griglia a colonne fisse: il controllo inizia subito dopo
    # la reale zona riservata all'etichetta, senza il grande spazio vuoto
    # che HBox può introdurre in Colab.
    label = widgets.HTML(
        value=label_html,
        layout=widgets.Layout(
            width=label_width, min_width=label_width, max_width=label_width,
            overflow='visible'
        )
    )
    return widgets.GridBox(
        children=[label, slider, text],
        layout=widgets.Layout(
            grid_template_columns=f'{label_width} 420px 210px',
            grid_gap='0px 6px',
            align_items='center',
            width='auto'
        )
    )


def _att41_theta_dot_label(indice):
    # Etichetta volutamente testuale: è stabile in Colab e non dipende dal rendering MathJax/CSS.
    return f'<span style="font-size:15px;white-space:nowrap;">theta_punto_{indice}(0) [rad/s]:</span>'


# ---------------------------------------------------------------------
# CONTROLLI — slider + inserimento da tastiera sincronizzato
# ---------------------------------------------------------------------

_slider_layout41 = widgets.Layout(width='420px')
_text_layout41 = widgets.Layout(width='210px')
_text_style41 = {'description_width': '95px'}

theta41_slider = widgets.FloatSlider(
    value=30.0, min=-120.0, max=120.0, step=2.0,
    description='',
    style={'description_width': '0px'},
    layout=_slider_layout41,
    continuous_update=True,
    readout=True,
    readout_format='.1f'
)
theta41_text = widgets.BoundedFloatText(
    value=30.0, min=-120.0, max=120.0, step=0.1,
    description='da tastiera:',
    style=_text_style41,
    layout=_text_layout41
)
theta41_link = widgets.jslink((theta41_slider, 'value'), (theta41_text, 'value'))

omegaA41_slider = widgets.FloatSlider(
    value=0.0, min=-6.0, max=6.0, step=0.1,
    description='',
    style={'description_width': '0px'},
    layout=_slider_layout41,
    continuous_update=True,
    readout=True,
    readout_format='.2f'
)
omegaA41_text = widgets.BoundedFloatText(
    value=0.0, min=-6.0, max=6.0, step=0.01,
    description='da tastiera:',
    style=_text_style41,
    layout=_text_layout41
)
omegaA41_link = widgets.jslink((omegaA41_slider, 'value'), (omegaA41_text, 'value'))

omegaB41_slider = widgets.FloatSlider(
    value=1.5, min=-6.0, max=6.0, step=0.1,
    description='',
    style={'description_width': '0px'},
    layout=_slider_layout41,
    continuous_update=True,
    readout=True,
    readout_format='.2f'
)
omegaB41_text = widgets.BoundedFloatText(
    value=1.5, min=-6.0, max=6.0, step=0.01,
    description='da tastiera:',
    style=_text_style41,
    layout=_text_layout41
)
omegaB41_link = widgets.jslink((omegaB41_slider, 'value'), (omegaB41_text, 'value'))

tmax41_slider = widgets.FloatSlider(
    value=3.0, min=1.0, max=8.0, step=0.25,
    description='',
    style={'description_width': '0px'},
    layout=_slider_layout41,
    continuous_update=True,
    readout=True,
    readout_format='.2f'
)
tmax41_text = widgets.BoundedFloatText(
    value=3.0, min=1.0, max=8.0, step=0.05,
    description='da tastiera:',
    style=_text_style41,
    layout=_text_layout41
)
tmax41_link = widgets.jslink((tmax41_slider, 'value'), (tmax41_text, 'value'))

btn_info41 = widgets.Button(
    description='Mostra informazione mancante',
    button_style='info',
    icon='eye',
    layout=widgets.Layout(width='290px', height='40px')
)

btn_run41 = widgets.Button(
    description='Avvia / aggiorna simulazione',
    button_style='success',
    icon='play',
    layout=widgets.Layout(width='290px', height='40px')
)

out_info41 = widgets.Output()
out_sim41 = widgets.Output()

instruction41 = widgets.HTML(
    """
    <div style="
        padding:12px 14px;margin:8px 0 10px 0;
        border-left:5px solid #7c3aed;background:#f5f3ff;color:#111827;
        border-radius:6px;line-height:1.45;">
      <b>Prima di premere «Avvia»:</b><br>
      scegli lo stesso angolo iniziale per i due pendoli e due valori, a tua scelta,
      della velocità angolare iniziale. <br><br><b>✍️ CARTA E PENNA</b> <br><br>Poi traccia <b>su carta</b>, nello stesso piano
      <b>(t, θ)</b>, uno schizzo qualitativo delle due curve
      <b>θ<sub>A</sub>(t)</b> e <b>θ<sub>B</sub>(t)</b> che prevedi.
    </div>
    """
)

controls41 = widgets.VBox([
    instruction41,
    _att41_riga_controllo(
        '<span style="font-size:15px;white-space:nowrap;font-family:Georgia,Times New Roman,serif;font-style:italic;">θ<sub>A</sub>(0) = θ<sub>B</sub>(0)</span> [°]:',
        theta41_slider, theta41_text, label_width='225px'
    ),
    _att41_riga_controllo(_att41_theta_dot_label('A'), omegaA41_slider, omegaA41_text, label_width='205px'),
    _att41_riga_controllo(_att41_theta_dot_label('B'), omegaB41_slider, omegaB41_text, label_width='205px'),
    _att41_riga_controllo(
        '<span style="font-size:15px;white-space:nowrap;font-family:Georgia,Times New Roman,serif;font-style:italic;">t<sub>finale</sub></span> [s]:',
        tmax41_slider, tmax41_text
    ),
    btn_run41
])
controls41.layout.display = 'none'


def _att41_mostra_info(_=None):
    controls41.layout.display = ''
    with out_info41:
        clear_output(wait=True)
        display(Markdown("### L'informazione mancante è la velocità angolare iniziale"))
        display(Markdown(
            r"Se fai un'istantanea del moto all'istante $t$ puoi riconoscere $\theta(t)$, "
            r"ma non puoi ricavare la **velocità angolare** $\dot\theta(t)$."
        ))
        # display(Math(r"\boxed{\dot{\theta}(t)}"))
        display(Markdown(
            r"Per specificare le condizioni in $t$, poniamo ora $t=0$: puoi assegnare "
            r"separatamente $\dot\theta_A(0)$ e $\dot\theta_B(0)$ e verificare se "
            r"la stessa configurazione iniziale produce necessariamente la stessa evoluzione."
        ))

def _att41_integra(accel_expr, theta0, omega0, m_val, l_val, g_val, tmax):
    accel_fun = sp.lambdify(
        (theta, theta_dot, t, m, l, g),
        accel_expr,
        modules='numpy'
    )

    def rhs(tt, z):
        th, om = z
        a = float(np.asarray(accel_fun(th, om, tt, m_val, l_val, g_val)))
        if not np.isfinite(a):
            raise FloatingPointError(
                f"Accelerazione non finita a t={tt:.4g} s."
            )
        return [om, a]

    sol = solve_ivp(
        rhs,
        (0.0, tmax),
        [theta0, omega0],
        method='DOP853',
        dense_output=True,
        rtol=1e-9,
        atol=1e-11,
    )
    if not sol.success:
        raise RuntimeError(sol.message)

    tt = np.linspace(0.0, tmax, 700)
    th, om = sol.sol(tt)
    return tt, np.asarray(th, float), np.asarray(om, float)


def _att41_canvas(tt, thA, omA, thB, omB, th0_deg, omA0, omB0):
    canvas_id = "att41_canvas_" + uuid.uuid4().hex
    slider_id = "att41_time_" + uuid.uuid4().hex
    play_id = "att41_play_" + uuid.uuid4().hex
    pause_id = "att41_pause_" + uuid.uuid4().hex
    speed_id = "att41_speed_" + uuid.uuid4().hex

    html = f"""
    <div style="max-width:1100px;margin:8px 0 14px 0;">
      <canvas id="{canvas_id}" style="width:100%;height:600px;border:1px solid #d1d5db;
              border-radius:10px;background:white;"></canvas>
      <div style="display:flex;align-items:center;gap:8px;flex-wrap:wrap;margin-top:8px;">
        <button id="{play_id}" style="padding:6px 14px;">▶ Avvia</button>
        <button id="{pause_id}" style="padding:6px 14px;">⏸ Pausa</button>
        <label style="font-size:13px;">velocità
          <select id="{speed_id}" style="margin-left:4px;">
            <option value="0.5">0.5×</option>
            <option value="1" selected>1×</option>
            <option value="2">2×</option>
          </select>
        </label>
        <input id="{slider_id}" type="range" min="0" max="{len(tt)-1}" value="0"
               step="1" style="flex:1;min-width:280px;">
      </div>
    </div>
    """
    display(HTML(html))

    payload = {
        "t": np.asarray(tt, float).tolist(),
        "a": np.asarray(thA, float).tolist(),
        "wa": np.asarray(omA, float).tolist(),
        "b": np.asarray(thB, float).tolist(),
        "wb": np.asarray(omB, float).tolist(),
    }

    js = f"""
    (() => {{
      const D = {json.dumps(payload)};
      const canvas = document.getElementById({json.dumps(canvas_id)});
      const slider = document.getElementById({json.dumps(slider_id)});
      const play = document.getElementById({json.dumps(play_id)});
      const pause = document.getElementById({json.dumps(pause_id)});
      const speedSel = document.getElementById({json.dumps(speed_id)});
      if (!canvas || !slider) return;

      const cssW = Math.max(780, Math.floor(canvas.getBoundingClientRect().width || 1050));
      const cssH = 600;
      const dpr = window.devicePixelRatio || 1;
      canvas.width = Math.floor(cssW*dpr);
      canvas.height = Math.floor(cssH*dpr);
      const ctx = canvas.getContext('2d');
      ctx.setTransform(dpr,0,0,dpr,0,0);
      canvas.style.height = cssH + 'px';

      const W = cssW, H = cssH;
      const n = D.t.length;
      let idx = 0;
      let timer = null;

      const allTheta = D.a.concat(D.b);
      let ymin = Math.min(...allTheta), ymax = Math.max(...allTheta);
      if (!(ymax > ymin)) {{ ymin -= 0.5; ymax += 0.5; }}
      const padY = Math.max(0.08, 0.12*(ymax-ymin));
      ymin -= padY; ymax += padY;

      const gx0 = 82, gx1 = W-40, gy0 = 365, gy1 = H-58;

      function line(x1,y1,x2,y2,stroke='#111827',lw=1.5,dash=[]) {{
        ctx.save();
        ctx.beginPath();
        ctx.setLineDash(dash);
        ctx.moveTo(x1,y1); ctx.lineTo(x2,y2);
        ctx.strokeStyle=stroke; ctx.lineWidth=lw; ctx.stroke();
        ctx.restore();
      }}

      function text(s,x,y,size=14,align='left',weight='normal',fill='#111827') {{
        ctx.save();
        ctx.fillStyle=fill;
        ctx.font=`${{weight}} ${{size}}px sans-serif`;
        ctx.textAlign=align;
        ctx.textBaseline='middle';
        ctx.fillText(s,x,y);
        ctx.restore();
      }}

      function drawPendulum(cx, cy, th, label, om, stroke) {{
        const R = 108;
        line(cx, cy-118, cx, cy+124, '#d1d5db', 1, [5,5]);
        const x = cx + R*Math.sin(th);
        const y = cy + R*Math.cos(th);
        line(cx, cy, x, y, stroke, 3);
        ctx.beginPath(); ctx.arc(cx,cy,5,0,2*Math.PI);
        ctx.fillStyle='#111827'; ctx.fill();
        ctx.beginPath(); ctx.arc(x,y,13,0,2*Math.PI);
        ctx.fillStyle=stroke; ctx.fill();
        text(label,cx,32,18,'center','700');
        text(`θ = ${{th.toFixed(3)}} rad`,cx,56,13,'center');
        text(`dθ/dt = ${{om.toFixed(3)}} rad/s`,cx,77,13,'center');
      }}

      function graphX(t) {{
        return gx0 + (gx1-gx0)*(t-D.t[0])/(D.t[n-1]-D.t[0]);
      }}

      function graphY(v) {{
        return gy1 - (gy1-gy0)*(v-ymin)/(ymax-ymin);
      }}

      function drawGrid() {{
        const nx = 5, ny = 5;

        for (let j=0;j<=nx;j++) {{
          const f = j/nx;
          const x = gx0 + f*(gx1-gx0);
          const tv = D.t[0] + f*(D.t[n-1]-D.t[0]);
          line(x,gy0,x,gy1,'#e5e7eb',1);
          text(tv.toFixed(2),x,gy1+18,11,'center','normal','#4b5563');
        }}

        for (let j=0;j<=ny;j++) {{
          const f = j/ny;
          const y = gy1 - f*(gy1-gy0);
          const vv = ymin + f*(ymax-ymin);
          line(gx0,y,gx1,y,'#e5e7eb',1);
          text(vv.toFixed(2),gx0-10,y,11,'right','normal','#4b5563');
        }}

        line(gx0,gy1,gx1,gy1,'#6b7280',1.2);
        line(gx0,gy0,gx0,gy1,'#6b7280',1.2);
        text('t [s]',gx1,gy1+38,12,'right','600');
        text('θ [rad]',gx0-54,gy0-10,12,'left','600');
      }}

      function drawCurve(arr, stroke, upto) {{
        ctx.save();
        ctx.beginPath();
        const imax = Math.max(1, Math.min(n-1, upto));
        for (let k=0;k<=imax;k++) {{
          const x=graphX(D.t[k]), y=graphY(arr[k]);
          if(k===0) ctx.moveTo(x,y); else ctx.lineTo(x,y);
        }}
        ctx.strokeStyle=stroke; ctx.lineWidth=2.2; ctx.stroke();
        ctx.restore();
      }}

      function draw(i) {{
        idx = Math.max(0,Math.min(n-1,i));
        ctx.clearRect(0,0,W,H);

        text('Stessa configurazione iniziale, condizioni dinamiche diverse',W/2,17,18,'center','700');

        drawPendulum(W*0.27, 185, D.a[idx], 'Pendolo A', D.wa[idx], '#2563eb');
        drawPendulum(W*0.73, 185, D.b[idx], 'Pendolo B', D.wb[idx], '#dc2626');

        text(`t = ${{D.t[idx].toFixed(3)}} s`,W/2,317,15,'center','700');
        text(
          `inizio: θA(0)=θB(0)={th0_deg:.1f}°   |   dθA/dt(0)={omA0:.2f} rad/s   |   dθB/dt(0)={omB0:.2f} rad/s`,
          W/2,340,12,'center','normal','#374151'
        );

        drawGrid();
        drawCurve(D.a,'#2563eb',idx);
        drawCurve(D.b,'#dc2626',idx);

        const xc=graphX(D.t[idx]);
        line(xc,gy0,xc,gy1,'#9ca3af',1,[4,4]);

        ctx.beginPath(); ctx.arc(xc,graphY(D.a[idx]),4.5,0,2*Math.PI);
        ctx.fillStyle='#2563eb'; ctx.fill();
        ctx.beginPath(); ctx.arc(xc,graphY(D.b[idx]),4.5,0,2*Math.PI);
        ctx.fillStyle='#dc2626'; ctx.fill();

        text('A',gx1-45,gy0+15,12,'left','700','#2563eb');
        text('B',gx1-22,gy0+15,12,'left','700','#dc2626');

        slider.value=String(idx);
      }}

      function stop() {{
        if(timer!==null) {{ clearInterval(timer); timer=null; }}
      }}

      function start() {{
        stop();
        if(idx>=n-1) idx=0;
        const speed = parseFloat(speedSel.value || '1');
        const dt = Math.max(12, 1000*(D.t[1]-D.t[0])/speed);
        timer=setInterval(() => {{
          idx += 2;
          if(idx>=n-1) {{ idx=n-1; draw(idx); stop(); return; }}
          draw(idx);
        }},dt);
      }}

      slider.addEventListener('input',() => {{ stop(); draw(parseInt(slider.value)); }});
      play.addEventListener('click',start);
      pause.addEventListener('click',stop);
      speedSel.addEventListener('change',() => {{ if(timer!==null) start(); }});

      draw(0);
    }})()
    """

    try:
        from google.colab import output as colab_output
        colab_output.eval_js(js)
    except Exception:
        display(Markdown(
            "_L'animazione interattiva è progettata per Google Colab. "
            "Il diagramma orario statico riportato sopra contiene comunque "
            "la soluzione numerica completa._"
        ))


def _att41_esegui(_=None):
    with out_sim41:
        clear_output(wait=True)

        try:
            accel_expr = _att41_accelerazione_da_L()
            massa_nella_dinamica = (m in accel_expr.free_symbols)

            theta0 = np.deg2rad(float(theta41_slider.value))
            omegaA0 = float(omegaA41_slider.value)
            omegaB0 = float(omegaB41_slider.value)
            tmax = float(tmax41_slider.value)

            l_val = _att41_valore_widget_precedente('l_widget', 1.0)
            g_val = _att41_valore_widget_precedente('g_widget', 9.81)
            # Se la ODE contiene m, usiamo il valore scelto nella sezione energetica precedente.
            m_val = _att41_valore_widget_precedente('m_energy_widget', 1.0)

            if l_val <= 0 or g_val <= 0:
                raise ValueError("ℓ e g devono essere positivi.")
            if massa_nella_dinamica and m_val <= 0:
                raise ValueError("Nel modello inserito m compare nella dinamica e deve essere positiva.")

            display(Markdown("### Modello effettivamente esplorato"))
            display(Markdown(
                "Le due evoluzioni sono calcolate dalla **stessa equazione ottenuta dalla tua Lagrangiana**:"
            ))
            display(Math(
                r"\boxed{\ddot{\theta}=" + sp.latex(accel_expr) + r"}"
            ))

            if massa_nella_dinamica:
                display(Math(
                    rf"m={m_val:.3g}\,\mathrm{{kg}},\qquad "
                    rf"\ell={l_val:.3g}\,\mathrm{{m}},\qquad "
                    rf"g={g_val:.3g}\,\mathrm{{m\,s^{{-2}}}}"
                ))
            else:
                display(Math(
                    rf"\ell={l_val:.3g}\,\mathrm{{m}},\qquad "
                    rf"g={g_val:.3g}\,\mathrm{{m\,s^{{-2}}}}"
                ))

            tt, thA, omA = _att41_integra(
                accel_expr, theta0, omegaA0, m_val, l_val, g_val, tmax
            )
            _, thB, omB = _att41_integra(
                accel_expr, theta0, omegaB0, m_val, l_val, g_val, tmax
            )

            display(Markdown("### Diagramma orario completo"))
            fig, ax = plt.subplots(figsize=(9.2, 5.0))
            ax.plot(tt, thA, linewidth=1.8, label='Pendolo A')
            ax.plot(tt, thB, linewidth=1.8, label='Pendolo B')
            ax.set_xlabel(r'$t$ [s]')
            ax.set_ylabel(r'$\theta$ [rad]')
            ax.set_title(r'Confronto delle evoluzioni $\theta_A(t)$ e $\theta_B(t)$')
            ax.grid(True, which='both', alpha=0.30)
            ax.legend()
            ax.tick_params(axis='both', labelsize=10)
            plt.tight_layout()
            plt.show()

            display(Markdown(
                "**Confronta le due curve con lo schizzo che avevi tracciato su carta.**"
            ))

            display(Markdown("### Simulazione sincronizzata"))
            _att41_canvas(
                tt, thA, omA, thB, omB,
                float(theta41_slider.value), omegaA0, omegaB0
            )

        except Exception as exc:
            display(Markdown("### L'esperimento non può essere avviato con il modello attuale"))
            display(Markdown(
                "Il notebook non sostituisce le tue espressioni con quelle attese. "
                "Il modello inserito, nella forma attuale, non consente questa esplorazione numerica."
            ))
            display(Markdown(f"`{type(exc).__name__}: {exc}`"))


btn_info41.on_click(_att41_mostra_info)
btn_run41.on_click(_att41_esegui)

display(Markdown(
    "Premi il pulsante per rendere esplicita la grandezza che completa le condizioni iniziali."
))
display(btn_info41)
display(out_info41)
display(controls41)
display(out_sim41)
````

## Teoria — Configurazione e Stato

La coordinata generalizzata

$$
q=\theta
$$

descrive la **configurazione geometrica** del pendolo in un determinato istante.

L'equazione del moto ottenuta mediante le equazioni di Eulero–Lagrange è però un'equazione differenziale del secondo ordine, che possiamo scrivere nella forma generale

$$
\ddot{\theta}=F(\theta,\dot{\theta},t).
$$

Per determinare univocamente l'evoluzione del sistema a partire da un istante iniziale $t_0$, non è quindi sufficiente conoscere soltanto la configurazione $\theta(t_0)$: occorre conoscere anche la velocità generalizzata $\dot{\theta}(t_0)$.

Introduciamo allora il vettore

$$
\boxed{
\mathbf z(t)=
\begin{pmatrix}
\theta(t)\\[4pt]
\dot{\theta}(t)
\end{pmatrix}
}
$$

che rappresenta lo **stato istantaneo** del pendolo.

---

### Dallo stato alla traiettoria nello spazio degli stati

Mentre il pendolo evolve nel tempo, cambiano sia $\theta(t)$ sia $\dot{\theta}(t)$.

Possiamo quindi rappresentare ogni stato del sistema come un punto del piano avente coordinate

$$
(\theta,\dot{\theta}).
$$

Questo piano costituisce lo **spazio degli stati** del pendolo; nel caso di un sistema a un solo grado di libertà viene anche comunemente indicato come **piano delle fasi**.

Durante il moto, il punto che rappresenta lo stato non rimane fermo: al variare del tempo descrive una curva

$$
\boxed{
\mathbf z(t)=
\bigl(\theta(t),\dot{\theta}(t)\bigr)
}
$$

nello spazio degli stati.

Questa curva è la **traiettoria del sistema nello spazio degli stati**: una curva parametrica il cui parametro è il tempo $t$.

È importante distinguere le due rappresentazioni:

- il grafico $\theta(t)$ descrive come cambia nel tempo una singola grandezza;
- la curva $(\theta(t),\dot{\theta}(t))$ descrive invece come evolve lo **stato complessivo** del sistema.

Il tempo non compare quindi come asse del piano $(\theta,\dot{\theta})$: esso è il parametro che determina la posizione del punto lungo la traiettoria.

Nella prossima attività proverai prima a **prevedere la traiettoria** e poi a osservare, in un suo punto, la direzione istantanea con cui lo stato sta evolvendo.

> **Nota terminologica**. In questa esplorazione usiamo il piano $(\theta,\dot\theta)$. In formulazione hamiltoniana lo spazio delle fasi canonico è invece parametrizzato da $(q,p)$, dove $p=\partial L/\partial\dot q$. Per il pendolo semplice le due rappresentazioni sono strettamente collegate, ma è utile mantenerle concettualmente distinte.

---

### La direzione istantanea dell'evoluzione

Poiché $\mathbf z(t)$ varia nel tempo, possiamo considerarne la derivata

$$
\boxed{
\dot{\mathbf z}(t)
=
\frac{d\mathbf z}{dt}
=
\begin{pmatrix}
\dot{\theta}(t)\\[4pt]
\ddot{\theta}(t)
\end{pmatrix}.
}
$$

Il vettore $\dot{\mathbf z}(t)$ rappresenta la **direzione istantanea dell'evoluzione nello spazio degli stati**.

Geometricamente, esso è un vettore **tangente alla traiettoria** nel punto

$$
\mathbf z(t)=
\begin{pmatrix}
\theta(t)\\
\dot{\theta}(t)
\end{pmatrix}.
$$

La sua prima componente indica con quale rapidità varia la coordinata $\theta$; la seconda indica con quale rapidità varia la velocità angolare $\dot{\theta}$.

Utilizzando l'equazione del moto

$$
\ddot{\theta}=F(\theta,\dot{\theta},t),
$$

possiamo quindi riscrivere l'equazione differenziale del secondo ordine come un sistema di due equazioni del primo ordine:

$$
\boxed{
\begin{cases}
\dot{\theta}=\omega,\\[4pt]
\dot{\omega}=F(\theta,\omega,t),
\end{cases}
}
\qquad
\omega\equiv\dot{\theta}.
$$

In forma vettoriale,

$$
\boxed{
\dot{\mathbf z}
=
\mathbf f(\mathbf z,t).
}
$$

L'equazione del moto può quindi essere interpretata geometricamente: a ogni stato $\mathbf z$ associa un vettore $\dot{\mathbf z}$ che indica **in quale direzione e con quale rapidità lo stato tende a evolvere**.

Per un sistema autonomo, cioè quando l'equazione del moto non dipende esplicitamente dal tempo,

$$
\dot{\mathbf z}=\mathbf f(\mathbf z),
$$

la direzione dell'evoluzione è determinata unicamente dal punto occupato nello spazio degli stati.



````python
#@title Attività 5 — Dal modello costruito al modello di riferimento { display-mode: "form" }

import sympy as sp
import numpy as np
from scipy.integrate import solve_ivp
from IPython.display import display, Markdown, Math, HTML


# ============================================================
# MODELLO COSTRUITO DALLO STUDENTE
# ============================================================

def _fase_accelerazione_da_TV():
    """
    Ricava l'accelerazione angolare dal modello costruito dallo studente
    attraverso la Lagrangiana L = T - V.
    """

    dL_dtheta = sp.diff(L, theta)
    dL_dtheta_dot = sp.diff(L, theta_dot)

    d_dt_dL_dtheta_dot = (
        sp.diff(dL_dtheta_dot, theta) * theta_dot
        + sp.diff(dL_dtheta_dot, theta_dot) * theta_ddot
        + sp.diff(dL_dtheta_dot, t)
    )

    eq_EL = sp.simplify(
        d_dt_dL_dtheta_dot - dL_dtheta
    )

    soluzioni = sp.solve(
        sp.Eq(eq_EL, 0),
        theta_ddot
    )

    if len(soluzioni) == 0:
        raise ValueError(
            "Non è stato possibile isolare l'accelerazione angolare "
            "a partire dalla Lagrangiana inserita."
        )

    return sp.simplify(soluzioni[0])


# ============================================================
# MODELLO DI RIFERIMENTO USATO DA QUI IN AVANTI
# ============================================================

MODELLO_PENDOLO_RIFERIMENTO_ATTIVO = False


def _verifica_pendolo_esatto():
    """
    Funzione di servizio mantenuta per compatibilità con eventuali
    attività successive.

    Da questa Attività 5 in avanti il modello di riferimento è quello
    corretto del pendolo semplice ideale, indipendentemente dall'esito
    del lavoro precedente dello studente.
    """

    if not globals().get(
        'MODELLO_PENDOLO_RIFERIMENTO_ATTIVO',
        False
    ):
        raise RuntimeError(
            "Prima di proseguire devi eseguire l'Attività 5."
        )

    # Se i simboli delle celle precedenti esistono, li riutilizziamo.
    # Altrimenti costruiamo simboli locali: il modello di riferimento
    # resta comunque ben definito.
    g_ref = globals().get('g', sp.Symbol('g', positive=True))
    l_ref = globals().get('l', sp.Symbol('l', positive=True))
    theta_ref = globals().get('theta', sp.Symbol('theta', real=True))

    return sp.simplify(
        -(g_ref / l_ref) * sp.sin(theta_ref)
    )


def _integra_pendolo(
    theta0,
    omega0,
    lval,
    gval,
    tmax,
    n=1600
):
    """
    Integra il modello di riferimento del pendolo semplice ideale:

        theta_dot = omega
        omega_dot = -(g/l) sin(theta)
    """

    def rhs(tt, z):

        th, om = z

        return [
            om,
            -(gval / lval) * np.sin(th)
        ]

    sol = solve_ivp(
        rhs,
        (0.0, tmax),
        [theta0, omega0],
        method='DOP853',
        dense_output=True,
        rtol=1e-10,
        atol=1e-12
    )

    if not sol.success:
        raise RuntimeError(sol.message)

    tt = np.linspace(
        0.0,
        tmax,
        n
    )

    th, om = sol.sol(tt)

    return tt, th, om


# ============================================================
# CONTROLLO DEL MODELLO INSERITO DALL'UTENTE
# ============================================================

display(HTML(
    """
    <div style="
        background:#e8f1fb;
        color:#17324d;
        border-left:5px solid #3977b7;
        padding:11px 14px;
        border-radius:6px;
        margin:10px 0 14px 0;
        font-size:18px;
        font-weight:700;">
        🔎 CONTROLLO DEL MODELLO INSERITO DALL'UTENTE
    </div>
    """
))


_required_5 = [
    'L',
    'theta',
    'theta_dot',
    'theta_ddot',
    't',
    'g',
    'l'
]

_missing_5 = [
    name
    for name in _required_5
    if name not in globals()
]


modello_studente_calcolabile = False
modello_studente_corretto = False
accel_studente = None
errore_modello_studente = None


# ============================================================
# TENTATIVO DI CONFRONTO, SE IL MODELLO È DISPONIBILE
# ============================================================

if not _missing_5:

    try:

        accel_studente = _fase_accelerazione_da_TV()

        accel_riferimento = sp.simplify(
            -(g / l) * sp.sin(theta)
        )

        modello_studente_calcolabile = True

        modello_studente_corretto = (
            sp.simplify(
                accel_studente - accel_riferimento
            ) == 0
        )

    except Exception as exc:

        errore_modello_studente = exc


# ============================================================
# ESITO 1 — MODELLO COSTRUITO E CORRETTO
# ============================================================

if (
    not _missing_5
    and modello_studente_calcolabile
    and modello_studente_corretto
):

    display(HTML(
        """
        <div style="
            background:#d9f2dc;
            color:#173b1d;
            border-left:5px solid #28a745;
            padding:12px 15px;
            border-radius:6px;
            margin:10px 0 14px 0;">
            <div style="font-size:19px; font-weight:800;">
                ESITO 1 — Modello costruito e corretto
            </div>
            <div style="margin-top:5px; font-weight:600;">
                Il modello ricavato dalla tua Lagrangiana coincide con
                quello del pendolo semplice ideale.
            </div>
        </div>
        """
    ))

    display(Markdown(r"""
**Accelerazione angolare ricavata dal tuo modello:**
"""))

    display(Math(
        r"\ddot{\theta}=" + sp.latex(accel_studente)
    ))

    display(Markdown(r"""
**Accelerazione angolare del modello di riferimento:**
"""))

    display(Math(
        r"\ddot{\theta}"
        r"="
        r"-\frac{g}{\ell}\sin\theta"
    ))

    display(HTML(
        """
        <div style="
            background:#edf8ef;
            color:#173b1d;
            padding:8px 12px;
            border-radius:6px;
            margin:10px 0 14px 0;
            font-weight:700;">
            ✓ Confronto: le due espressioni coincidono.
        </div>
        """
    ))


# ============================================================
# ESITO 2 — MODELLO COSTRUITO MA NON CORRETTO
# ============================================================

elif (
    not _missing_5
    and modello_studente_calcolabile
):

    display(HTML(
        """
        <div style="
            background:#fff0e0;
            color:#6b3c00;
            border-left:5px solid #e68a00;
            padding:12px 15px;
            border-radius:6px;
            margin:10px 0 14px 0;">
            <div style="font-size:19px; font-weight:800;">
                ESITO 2 — Modello costruito ma non coincidente
            </div>
            <div style="margin-top:5px; font-weight:600;">
                Dalla tua Lagrangiana è possibile ricavare
                un'equazione del moto, ma essa non coincide con
                quella del pendolo semplice ideale.
            </div>
        </div>
        """
    ))

    display(Markdown(r"""
**Accelerazione angolare ricavata dal tuo modello:**
"""))

    display(Math(
        r"\ddot{\theta}=" + sp.latex(accel_studente)
    ))

    display(Markdown(r"""
**Accelerazione angolare del modello di riferimento:**
"""))

    display(Math(
        r"\ddot{\theta}"
        r"="
        r"-\frac{g}{\ell}\sin\theta"
    ))

    display(HTML(
        """
        <div style="
            background:#fff7ed;
            color:#6b3c00;
            padding:8px 12px;
            border-radius:6px;
            margin:10px 0 14px 0;
            font-weight:700;">
            ⚠️ Confronto: le due espressioni non coincidono.
        </div>
        """
    ))


# ============================================================
# ESITO 3 — CONFRONTO NON ESEGUIBILE
# ============================================================

else:

    display(HTML(
        """
        <div style="
            background:#f1f1f1;
            color:#333333;
            border-left:5px solid #777777;
            padding:12px 15px;
            border-radius:6px;
            margin:10px 0 14px 0;">
            <div style="font-size:19px; font-weight:800;">
                ESITO 3 — Confronto non eseguibile
            </div>
            <div style="margin-top:5px; font-weight:600;">
                Non è possibile confrontare il modello dello studente
                con quello di riferimento.
            </div>
        </div>
        """
    ))

    if _missing_5:

        display(Markdown(r"""
Il motivo è che il modello precedente **non risulta completamente costruito
nel notebook**. In particolare mancano uno o più elementi necessari per
ricavare l'equazione di Eulero-Lagrange.
"""))

        display(HTML(
            "<div style='margin:6px 0 14px 20px;'>"
            "<b>Elementi non disponibili:</b> "
            + ", ".join(f"<code>{name}</code>" for name in _missing_5)
            + "</div>"
        ))

        display(Markdown(r"""
Questo non impedisce di proseguire: semplicemente, in questo caso
**non possiamo valutare il modello costruito dallo studente**.
"""))

    else:

        display(Markdown(r"""
La Lagrangiana o le variabili necessarie risultano presenti, ma da esse
**non è stato possibile ricavare in modo utilizzabile l'accelerazione
angolare**.
"""))

        if errore_modello_studente is not None:

            display(HTML(
                """
                <div style="
                    background:#fafafa;
                    color:#555555;
                    padding:8px 12px;
                    border-radius:6px;
                    margin:8px 0 14px 0;
                    font-size:14px;">
                    Il confronto simbolico non ha prodotto
                    un'equazione del moto utilizzabile.
                </div>
                """
            ))


# ============================================================
# PASSAGGIO ESPLICITO AL MODELLO CORRETTO
# ============================================================

display(HTML(
    """
    <div style="
        background:#d9f2dc;
        color:#173b1d;
        border-left:5px solid #28a745;
        padding:12px 15px;
        border-radius:6px;
        margin:18px 0 14px 0;
        font-size:18px;
        font-weight:700;">
        ✓ Da questo punto in avanti: il modello di riferimento
        sarà quello corretto del pendolo semplice ideale
    </div>
    """
))

display(Markdown(r"""
Nelle attività precedenti il notebook ha utilizzato il **modello costruito da te**
a partire dalle espressioni di $T$ e $V$, propagandone anche eventuali errori.

Il controllo appena eseguito ha quindi tre possibili esiti:

- **ESITO 1:** il modello costruito è disponibile e coincide con quello corretto;
- **ESITO 2:** il modello costruito è disponibile, ma non coincide con quello corretto;
- **ESITO 3:** il confronto non può essere eseguito perché il modello non è stato
  completato oppure non consente di ricavare un'equazione del moto utilizzabile.

**Qualunque sia l'esito del controllo**, da questo punto in avanti vogliamo
studiare tutti lo **stesso sistema dinamico**.

Il documento utilizzerà quindi esplicitamente il modello corretto
del **pendolo semplice ideale**:
"""))

display(HTML('<div style="height:12px;"></div>'))

display(Math(
    r"\boxed{"
    r"\ddot{\theta}"
    r"="
    r"-\frac{g}{\ell}\sin\theta"
    r"}"
))

display(HTML('<div style="height:12px;"></div>'))

display(Markdown(r"""
ovvero, introducendo
"""))

display(Math(
    r"\omega=\dot{\theta},"
))

display(Markdown(r"""
il sistema del primo ordine
"""))

display(HTML('<div style="height:12px;"></div>'))

display(Math(
    r"\boxed{"
    r"\dot{\theta}=\omega,"
    r"\qquad"
    r"\dot{\omega}=-\frac{g}{\ell}\sin\theta"
    r"}."
))

display(HTML('<div style="height:12px;"></div>'))

display(Markdown(r"""
Questa scelta **non modifica retroattivamente** ciò che hai osservato
nelle attività precedenti: serve a fissare da qui in avanti un modello comune
con cui studiare lo spazio degli stati, le traiettorie e le proprietà
dinamiche del pendolo.
"""))

# Il modello di riferimento viene attivato SEMPRE,
# indipendentemente dall'esito 1, 2 o 3.
MODELLO_PENDOLO_RIFERIMENTO_ATTIVO = True


# ============================================================
# PREPARAZIONE ALL'ATTIVITÀ 5.1
# ============================================================

display(HTML(
    """
    <div style="
        background:#fff3a3;
        color:#111111;
        padding:10px 14px;
        border-radius:6px;
        font-weight:700;
        margin:18px 0 14px 0;">
        ✍️ CARTA E PENNA
    </div>
    """
))

display(Markdown(r"""
Scegli una condizione iniziale, per esempio
"""))

display(Math(
    r"\theta(0)=30^\circ,"
    r"\qquad"
    r"\dot{\theta}(0)=\ldots\;[\mathrm{rad/s}]."
))

display(Markdown(r"""
**Prima di eseguire la cella successiva**, prova a prevedere che cosa disegnerà
nel piano $(\theta,\dot{\theta})$ il punto che rappresenta lo stato del pendolo:
fai un disegno che rappresenti la tua previsione.

Domandati in particolare:

- il punto tornerà allo stato iniziale?
- la curva sarà aperta o chiusa?
- può attraversare due volte lo stesso punto $(\theta,\dot{\theta})$?
- che cosa succederà nel diagramma quando il pendolo raggiunge un punto di inversione del moto?

Nella cella di simulazione successiva puoi cambiare liberamente
$\theta(0)$ e $\dot{\theta}(0)$ e confrontare la tua previsione
con il risultato numerico.
"""))
````

````python
#@title Attività 5.1 — Una traiettoria nello spazio degli stati { display-mode: "form" }

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
import ipywidgets as widgets
from IPython.display import display, clear_output, Math, Markdown


# ============================================================
# RESET DIDATTICO: MODELLO DI RIFERIMENTO AUTONOMO
# ============================================================
#
# IMPORTANTE
# ----------
# Da questa attività in avanti NON utilizziamo più le espressioni T, V,
# la Lagrangiana L o l'equazione del moto ricavata dallo studente nelle
# attività precedenti.
#
# Anche se in precedenza sono state inserite espressioni errate, questa
# cella riparte esplicitamente dal modello corretto del pendolo semplice
# ideale:
#
#     theta_dot = omega
#     omega_dot = -(g/l) sin(theta)
#
# La cella è quindi autosufficiente rispetto al lavoro simbolico svolto
# in precedenza. I valori numerici di l e g vengono scelti qui sotto.


MODELLO_PENDOLO_RIFERIMENTO_ATTIVO = True


def _verifica_pendolo_esatto():
    """
    Verifica di servizio usata anche dalle attività successive.

    Non consulta T, V, L né alcun risultato prodotto dallo studente:
    il modello di riferimento è definito direttamente in questa cella.
    """
    if not MODELLO_PENDOLO_RIFERIMENTO_ATTIVO:
        raise RuntimeError(
            "Il modello di riferimento del pendolo non è attivo."
        )

    return True


def _integra_pendolo(theta0, omega0, lval, gval, tmax, n=1600):
    """
    Integra esclusivamente il modello di riferimento del pendolo semplice:

        theta_dot = omega
        omega_dot = -(g/l) sin(theta)

    Tutti i parametri necessari sono passati esplicitamente alla funzione.
    Nessuna variabile simbolica costruita nelle attività precedenti viene usata.
    """

    lval = float(lval)
    gval = float(gval)
    tmax = float(tmax)

    if lval <= 0:
        raise ValueError("La lunghezza ℓ deve essere positiva.")

    if gval <= 0:
        raise ValueError("L'accelerazione di gravità g deve essere positiva.")

    if tmax <= 0:
        raise ValueError("Il tempo finale deve essere positivo.")

    def f(tt, z):
        th, om = z
        return [
            om,
            -(gval / lval) * np.sin(th)
        ]

    sol = solve_ivp(
        f,
        (0.0, tmax),
        [float(theta0), float(omega0)],
        method='DOP853',
        dense_output=True,
        rtol=1e-10,
        atol=1e-12
    )

    if not sol.success:
        raise RuntimeError(sol.message)

    tt = np.linspace(0.0, tmax, int(n))
    th, om = sol.sol(tt)

    return tt, th, om


# ============================================================
# PARAMETRI DEL MODELLO
# ============================================================

l_phase_w = widgets.FloatText(
    value=1.0,
    description='ℓ [m]:'
)

g_phase_w = widgets.FloatText(
    value=9.81,
    description='g [m/s²]:'
)


# ============================================================
# CONDIZIONE INIZIALE: theta(0)
# ============================================================

th0_w = widgets.FloatSlider(
    value=30.0,
    min=-175.0,
    max=175.0,
    step=5.0,
    description='θ(0) [°]:',
    continuous_update=True,
    readout=True,
    readout_format='.1f',
    layout=widgets.Layout(width='430px')
)

th0_manual_w = widgets.BoundedFloatText(
    value=30.0,
    min=-175.0,
    max=175.0,
    step=0.1,
    description='da tastiera:',
    style={'description_width': '95px'},
    layout=widgets.Layout(width='210px')
)

th0_link = widgets.jslink(
    (th0_w, 'value'),
    (th0_manual_w, 'value')
)


# ============================================================
# CONDIZIONE INIZIALE: theta_dot(0)
# ============================================================
#
# Il range esplorabile di theta_dot(0) viene adattato automaticamente
# ai valori correnti di ℓ e g.
#
# La regola usata internamente lascia un margine del 20% oltre il valore
# energetico necessario, partendo dalla configurazione inferiore, per
# raggiungere la configurazione superiore instabile.
#
# Questa informazione NON viene mostrata nell'interfaccia: lo studente
# può eventualmente ricavare autonomamente tale valore durante il
# percorso didattico.


def _omega_range_da_l_g(lval, gval):
    lval = float(lval)
    gval = float(gval)

    if (
        not np.isfinite(lval)
        or not np.isfinite(gval)
        or lval <= 0
        or gval <= 0
    ):
        return None

    return 1.20 * np.sqrt(4.0 * gval / lval)


_om_lim_init = _omega_range_da_l_g(
    l_phase_w.value,
    g_phase_w.value
)

om0_w = widgets.FloatSlider(
    value=0.0,
    min=-_om_lim_init,
    max=_om_lim_init,
    step=0.1,
    description='theta_punto(0) [rad/s]:',
    style={'description_width': '175px'},
    continuous_update=True,
    readout=True,
    readout_format='.2f',
    layout=widgets.Layout(width='560px')
)

om0_manual_w = widgets.BoundedFloatText(
    value=0.0,
    min=-_om_lim_init,
    max=_om_lim_init,
    step=0.01,
    description='da tastiera:',
    style={'description_width': '95px'},
    layout=widgets.Layout(width='210px')
)

om0_link = widgets.jslink(
    (om0_w, 'value'),
    (om0_manual_w, 'value')
)


def _aggiorna_limiti_omega(change=None):
    """
    Aggiorna silenziosamente i limiti ammessi per theta_dot(0)
    quando cambiano ℓ o g.

    Nessun valore critico viene mostrato allo studente.
    """

    nuovo_limite = _omega_range_da_l_g(
        l_phase_w.value,
        g_phase_w.value
    )

    # Se ℓ o g sono temporaneamente non validi mentre l'utente
    # sta scrivendo, lasciamo invariato il range precedente.
    if nuovo_limite is None:
        return

    valore_corrente = float(om0_w.value)

    valore_corretto = float(np.clip(
        valore_corrente,
        -nuovo_limite,
        nuovo_limite
    ))

    # Prima riportiamo l'eventuale valore corrente all'interno
    # del nuovo intervallo; poi aggiorniamo i limiti.
    om0_w.value = valore_corretto
    om0_manual_w.value = valore_corretto

    om0_w.min = -nuovo_limite
    om0_w.max = nuovo_limite

    om0_manual_w.min = -nuovo_limite
    om0_manual_w.max = nuovo_limite


l_phase_w.observe(
    _aggiorna_limiti_omega,
    names='value'
)

g_phase_w.observe(
    _aggiorna_limiti_omega,
    names='value'
)


# ============================================================
# TEMPO DI SIMULAZIONE
# ============================================================

t_phase_w = widgets.FloatSlider(
    value=10.0,
    min=2.0,
    max=25.0,
    step=1.0,
    description='t finale [s]:',
    continuous_update=True,
    readout=True,
    readout_format='.1f',
    layout=widgets.Layout(width='430px')
)

t_phase_manual_w = widgets.BoundedFloatText(
    value=10.0,
    min=2.0,
    max=25.0,
    step=0.1,
    description='da tastiera:',
    style={'description_width': '95px'},
    layout=widgets.Layout(width='210px')
)

t_phase_link = widgets.jslink(
    (t_phase_w, 'value'),
    (t_phase_manual_w, 'value')
)


# ============================================================
# PULSANTE E OUTPUT
# ============================================================

run_phase_btn = widgets.Button(
    description='Disegna traiettoria',
    button_style=''
)

out_phase = widgets.Output()


# Memorizza ESATTAMENTE l'ultima traiettoria visualizzata.
# Verrà utilizzata dall'Attività 5.2.
_att5_last_data = {}


display(widgets.VBox([
    widgets.HBox([th0_w, th0_manual_w]),
    widgets.HBox([om0_w, om0_manual_w]),
    widgets.HBox([l_phase_w, g_phase_w]),
    widgets.HBox([t_phase_w, t_phase_manual_w]),
    run_phase_btn,
    out_phase
]))


# ============================================================
# ESECUZIONE
# ============================================================

def _run_phase(_):

    with out_phase:

        clear_output(wait=True)

        try:

            _verifica_pendolo_esatto()

            lval = float(l_phase_w.value)
            gval = float(g_phase_w.value)

            if lval <= 0 or gval <= 0:
                raise ValueError(
                    'ℓ e g devono essere positivi.'
                )

            th0 = np.deg2rad(
                float(th0_w.value)
            )

            om0 = float(
                om0_w.value
            )

            tt, th, om = _integra_pendolo(
                th0,
                om0,
                lval,
                gval,
                float(t_phase_w.value)
            )

            _att5_last_data.clear()

            _att5_last_data.update({
                't': np.asarray(tt, float),
                'theta': np.asarray(th, float),
                'omega': np.asarray(om, float),
                'l': lval,
                'g': gval,
                'theta0': th0,
                'omega0': om0
            })

            display(Math(
                rf"(\theta_0,\dot{{\theta}}_0)="
                rf"({th0:.3f}\,\mathrm{{rad}},"
                rf"\,{om0:.3f}\,\mathrm{{rad\,s^{{-1}}}})"
            ))

            fig, ax = plt.subplots(
                figsize=(8.2, 5.2)
            )

            ax.plot(
                th,
                om,
                linewidth=1.6
            )

            ax.scatter(
                [th[0]],
                [om[0]],
                s=45,
                label='stato iniziale',
                zorder=4
            )

            ax.scatter(
                [th[-1]],
                [om[-1]],
                s=28,
                label='stato al tempo finale',
                zorder=4
            )

            ax.set_xlabel(
                r'$\theta$ [rad]'
            )

            ax.set_ylabel(
                r'$\dot\theta$ [rad/s]'
            )

            ax.set_title(
                r'Traiettoria nello spazio degli stati '
                r'$(\theta,\dot\theta)$'
            )

            ax.grid(
                True,
                alpha=0.25
            )

            ax.legend()

            plt.show()

        except Exception as e:

            print(
                'Impossibile eseguire l’attività:',
                e
            )


run_phase_btn.on_click(
    _run_phase
)
````

````python
#@title Simulazione 5.2 — Vettore tangente alla traiettoria { display-mode: "form" }

import secrets
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output, Math, Markdown


_att51_data = {}


btn_vector51 = widgets.Button(
    description='Verifica il vettore di evoluzione',
    button_style='info',
    icon='arrow-right',
    disabled=True,
    layout=widgets.Layout(
        width='290px',
        height='40px',
        display='none'
    )
)


paper_prompt51 = widgets.HTML(
    """
    <div style="
        padding:11px 14px;
        margin:12px 0;
        border-left:5px solid #e0a800;
        background:#fff8d6;
        color:#222;
        border-radius:6px;
        line-height:1.50;">

      <b>✍️ CARTA E PENNA</b><br><br>

      <b>1.</b> Osserva il punto arancione e la traiettoria nello spazio degli stati.
      Disegna nel punto scelto il <b>verso del vettore tangente</b>
      e prevedi il <b>segno delle sue due componenti</b>.<br><br>

      <b>2.</b> Disegna inoltre uno <b>schizzo del pendolo all'istante
      <i>t</i>*</b>: rappresenta approssimativamente l'angolo
      <b>θ(<i>t</i>*)</b> rispetto alla verticale e disegna sulla massa
      il <b>vettore velocità</b>,
      indicando il verso del moto che prevedi.<br><br>

      Solo dopo premi <b>«Verifica il vettore di evoluzione»</b>.
    </div>
    """,
    layout=widgets.Layout(display='none')
)


out_att51_plot = widgets.Output()
out_att51_state = widgets.Output()
out_att51_vector = widgets.Output()
out_att51_pendulum = widgets.Output()


display(widgets.VBox([
    out_att51_plot,
    out_att51_state,
    paper_prompt51,
    btn_vector51,
    out_att51_vector,
    out_att51_pendulum
]))


def _att51_sign_relation(value, tol=1e-12):

    if value > tol:
        return '>0', 'positiva'

    if value < -tol:
        return '<0', 'negativa'

    return '=0', 'nulla'


def _att51_draw_pendulum(thj, omj, lval):

    """
    Disegna la configurazione fisica del pendolo a t*.

    Convenzione:
        x = l sin(theta)
        y = -l cos(theta)

    Il vettore velocità ha direzione fisica corretta:
        v = l theta_dot (cos(theta), sin(theta))

    La lunghezza grafica della freccia è riscalata.
    """

    with out_att51_pendulum:

        clear_output(wait=True)

        xb = lval * np.sin(thj)
        yb = -lval * np.cos(thj)

        vx = lval * omj * np.cos(thj)
        vy = lval * omj * np.sin(thj)

        speed = float(np.hypot(vx, vy))

        fig, ax = plt.subplots(figsize=(6.4, 6.0))

        # Verticale di riferimento
        ax.plot(
            [0, 0],
            [0, -1.15*lval],
            '--',
            linewidth=1.0,
            alpha=0.55,
            label='verticale'
        )

        # Filo / asta
        ax.plot(
            [0, xb],
            [0, yb],
            linewidth=2.2
        )

        # Punto di sospensione
        ax.scatter(
            [0], [0],
            s=45,
            zorder=5
        )

        # Massa
        ax.scatter(
            [xb], [yb],
            s=150,
            zorder=6
        )

        # Arco che indica theta
        if abs(thj) > 1e-5:

            r_arc = 0.28 * lval

            phi0 = -np.pi/2
            phi1 = -np.pi/2 + thj

            phi = np.linspace(phi0, phi1, 100)

            ax.plot(
                r_arc*np.cos(phi),
                r_arc*np.sin(phi),
                linewidth=1.4
            )

            phi_mid = 0.5 * (phi0 + phi1)

            ax.text(
                1.18*r_arc*np.cos(phi_mid),
                1.18*r_arc*np.sin(phi_mid),
                r'$\theta(t^*)$',
                fontsize=12,
                ha='center',
                va='center'
            )

        # Vettore velocità: verso fisico corretto,
        # lunghezza grafica normalizzata
        if speed > 1e-12:

            ux = vx / speed
            uy = vy / speed

            arrow_len = 0.38 * lval

            ax.annotate(
                '',
                xy=(
                    xb + arrow_len*ux,
                    yb + arrow_len*uy
                ),
                xytext=(xb, yb),
                arrowprops=dict(
                    arrowstyle='->',
                    lw=2.5,
                    mutation_scale=20
                ),
                zorder=7
            )

            ax.text(
                xb + 1.08*arrow_len*ux,
                yb + 1.08*arrow_len*uy,
                r'$\mathbf{v}(t^*)$',
                fontsize=13,
                ha='center',
                va='center'
            )

        else:

            ax.text(
                xb + 0.08*lval,
                yb,
                r'$\mathbf{v}(t^*)=\mathbf{0}$',
                fontsize=12,
                va='center'
            )

# Limiti simmetrici: devono contenere il pendolo
# in qualunque configurazione e la freccia della velocità.
        lim = 1.45 * lval

        ax.set_xlim(-lim, lim)
        ax.set_ylim(-lim, lim)

        ax.set_aspect('equal', adjustable='box')

        ax.set_xlabel('$x$ [m]')
        ax.set_ylabel('$y$ [m]')

        ax.set_title(
            r'Configurazione del pendolo all’istante $t^*$'
        )

        ax.grid(True, alpha=0.20)

        plt.show()


        display(Math(
            rf"\theta(t^*)="
            rf"{thj:.4f}\,\mathrm{{rad}}"
            rf"\approx "
            rf"{np.rad2deg(thj):.1f}^\circ"
        ))


        display(Math(
            rf"\dot{{\theta}}(t^*)="
            rf"{omj:.4f}\,\mathrm{{rad\,s^{{-1}}}}"
        ))


        display(Math(
            rf"|\mathbf{{v}}(t^*)|="
            rf"\ell\,|\dot{{\theta}}(t^*)|="
            rf"{speed:.4f}\,\mathrm{{m\,s^{{-1}}}}"
        ))


        display(Markdown(
            "La freccia rappresenta il **verso fisico della velocità della massa**. "
            "La sua lunghezza nel disegno è riscalata per renderla leggibile."
        ))


def _att51_draw(show_vector=False):

    if not _att51_data:
        return

    tt = _att51_data['t']
    th = _att51_data['theta']
    om = _att51_data['omega']

    gval = _att51_data['g']
    lval = _att51_data['l']

    tstar = _att51_data['tstar']


    # Stato corrispondente all'istante casuale t*
    thj = float(np.interp(tstar, tt, th))
    omj = float(np.interp(tstar, tt, om))

    accj = float(
        -(gval / lval) * np.sin(thj)
    )


    with out_att51_plot:

        clear_output(wait=True)

        fig, ax = plt.subplots(figsize=(8.4, 5.4))


        ax.plot(
            th,
            om,
            linewidth=1.6,
            label='traiettoria dell’Attività 5'
        )


        ax.scatter(
            [th[0]],
            [om[0]],
            s=48,
            label='stato iniziale',
            zorder=4
        )


        ax.scatter(
            [thj],
            [omj],
            s=78,
            color='tab:orange',
            label=rf'punto casuale a $t^*={tstar:.3f}$ s',
            zorder=5
        )


        if show_vector:

            # La lunghezza della freccia viene scalata graficamente.
            # Il verso è invece determinato dalle componenti fisiche
            # (theta_dot, theta_ddot).

            xr = max(float(np.ptp(th)), 1e-6)
            yr = max(float(np.ptp(om)), 1e-6)

            norm_display = np.hypot(
                omj / xr,
                accj / yr
            )


            if norm_display > 1e-12:

                dt_display = 0.13 / norm_display

                dx = omj * dt_display
                dy = accj * dt_display


                ax.annotate(
                    '',
                    xy=(thj + dx, omj + dy),
                    xytext=(thj, omj),
                    arrowprops=dict(
                        arrowstyle='->',
                        lw=2.3,
                        mutation_scale=19,
                        color='tab:orange'
                    ),
                    zorder=6
                )


                ax.text(
                    thj + dx,
                    omj + dy,
                    r'  $\dot{\mathbf{z}}$',
                    fontsize=13,
                    va='center'
                )


            else:

                ax.text(
                    thj,
                    omj,
                    r'  $\dot{\mathbf{z}}=\mathbf{0}$',
                    fontsize=13,
                    va='bottom'
                )


        ax.set_xlabel(r'$\theta$ [rad]')
        ax.set_ylabel(r'$\dot\theta$ [rad/s]')

        ax.set_title(
            r'Traiettoria nello spazio degli stati '
            r'$(\theta,\dot\theta)$'
        )

        ax.grid(True, alpha=0.25)
        ax.legend()

        plt.show()


    with out_att51_state:

        clear_output(wait=True)

        display(Markdown(
            rf"Il codice ha scelto casualmente "
            rf"**$t^*={tstar:.3f}\,\mathrm{{s}}$**. "
            "Il punto arancione rappresenta lo stato corrispondente "
            "lungo la traiettoria dell'Attività 5."
        ))


    with out_att51_vector:

        clear_output(wait=True)

        if show_vector:

            sign_om, word_om = _att51_sign_relation(omj)
            sign_acc, word_acc = _att51_sign_relation(accj)


            display(
                Markdown("### Verifica della previsione")
            )


            display(Math(
                rf"\dot{{\mathbf{{z}}}}(t^*)="
                rf"\begin{{pmatrix}}"
                rf"\dot{{\theta}}(t^*)\\[2pt]"
                rf"\ddot{{\theta}}(t^*)"
                rf"\end{{pmatrix}}="
                rf"\begin{{pmatrix}}"
                rf"{omj:.4f}\,\mathrm{{rad\,s^{{-1}}}}\\[3pt]"
                rf"{accj:.4f}\,\mathrm{{rad\,s^{{-2}}}}"
                rf"\end{{pmatrix}}"
            ))


            display(Math(
                rf"\dot{{\theta}}(t^*)\;{sign_om},"
                rf"\qquad"
                rf"\ddot{{\theta}}(t^*)\;{sign_acc}"
            ))


            display(Markdown(
                f"- componente lungo l'asse $\\theta$: **{word_om}**;\n"
                f"- componente lungo l'asse $\\dot\\theta$: **{word_acc}**."
            ))


            display(Markdown(
                "Il grafico nello spazio degli stati (vedi sopra) è stato aggiornato! "
                "Il vettore arancione mostra il **verso del vettore tangente**. "
                "La sua lunghezza è riscalata soltanto per renderla leggibile."
            ))


    # Lo schizzo fisico compare SOLO dopo la verifica
    if show_vector:

        _att51_draw_pendulum(
            thj,
            omj,
            lval
        )

    else:

        with out_att51_pendulum:
            clear_output(wait=True)


def _att51_choose_random_tstar():

    tt = _att51_data['t']

    t0 = float(tt[0])
    tf = float(tt[-1])


    # Escludiamo soltanto il 10% iniziale e finale.
    lo = t0 + 0.10 * (tf - t0)
    hi = t0 + 0.90 * (tf - t0)


    if hi <= lo:
        lo, hi = t0, tf


    previous = globals().get(
        '_att51_previous_tstar',
        None
    )


    # secrets usa entropia del sistema:
    # runtime/studenti diversi producono scelte indipendenti.
    # Inoltre evitiamo di ottenere consecutivamente
    # lo stesso t* nello stesso runtime.

    for _ in range(100):

        u = secrets.randbits(53) / float(1 << 53)

        candidate = lo + (hi - lo) * u


        if (
            previous is None
            or abs(candidate - previous)
            > max(1e-6, 1e-4 * (tf - t0))
        ):

            globals()['_att51_previous_tstar'] = candidate

            return float(candidate)


    candidate = lo + (hi - lo) * 0.5

    if (
        previous is not None
        and abs(candidate - previous) <= 1e-12
    ):

        candidate = lo + (hi - lo) * 0.5001


    globals()['_att51_previous_tstar'] = candidate

    return float(candidate)


def _att51_initialize():

    with out_att51_plot:
        clear_output(wait=True)

    with out_att51_state:
        clear_output(wait=True)

    with out_att51_vector:
        clear_output(wait=True)

    with out_att51_pendulum:
        clear_output(wait=True)


    paper_prompt51.layout.display = 'none'

    btn_vector51.layout.display = 'none'
    btn_vector51.disabled = True


    try:

        if (
            '_att5_last_data' not in globals()
            or not _att5_last_data
        ):

            raise RuntimeError(
                "Prima esegui l'Attività 5 e premi "
                "«Disegna traiettoria»: "
                "l'Attività 5.1 deve lavorare sulla traiettoria "
                "numerica che hai appena ottenuto."
            )


        _att51_data.clear()


        _att51_data.update({

            't':
                np.asarray(
                    _att5_last_data['t'],
                    float
                ).copy(),

            'theta':
                np.asarray(
                    _att5_last_data['theta'],
                    float
                ).copy(),

            'omega':
                np.asarray(
                    _att5_last_data['omega'],
                    float
                ).copy(),

            'l':
                float(_att5_last_data['l']),

            'g':
                float(_att5_last_data['g'])

        })


        if _att51_data['t'].size < 3:

            raise RuntimeError(
                "La traiettoria dell'Attività 5 "
                "contiene troppo pochi punti."
            )


        _att51_data['tstar'] = (
            _att51_choose_random_tstar()
        )


        _att51_draw(show_vector=False)


        paper_prompt51.layout.display = ''

        btn_vector51.layout.display = ''
        btn_vector51.disabled = False


    except Exception as e:

        _att51_data.clear()

        with out_att51_plot:

            clear_output(wait=True)

            print(
                "Impossibile eseguire l'Attività 5.1:",
                e
            )


def _att51_show_vector(_=None):

    if _att51_data:

        _att51_draw(
            show_vector=True
        )


btn_vector51.on_click(
    _att51_show_vector
)


# Ogni nuova esecuzione della cella sceglie
# automaticamente un nuovo t*.
_att51_initialize()
````

## Teoria — Vettore tangente, campo dinamico e unicità

Nell'attività precedente hai osservato che lo stato

$$
\mathbf z(t)=
\begin{pmatrix}
\theta(t)\\[2pt]
\dot\theta(t)
\end{pmatrix}
$$

descrive una traiettoria nello spazio degli stati.

La sua derivata rispetto al tempo è

$$
\boxed{
\dot{\mathbf z}(t)
=
\frac{d\mathbf z}{dt}
=
\begin{pmatrix}
\dot\theta(t)\\[3pt]
\ddot\theta(t)
\end{pmatrix}.
}
$$

Il vettore $\dot{\mathbf z}$ rappresenta la **direzione istantanea dell'evoluzione dello stato** ed è tangente alla traiettoria nel punto $\mathbf z(t)$.

Le sue due componenti hanno significato e unità differenti:

$$
\dot\theta \quad [\mathrm{rad/s}],
\qquad
\ddot\theta \quad [\mathrm{rad/s^2}].
$$

Per questo, nel grafico dell'attività precedente, la lunghezza della freccia viene riscalata per ragioni di leggibilità: sono la sua **direzione** e le componenti numeriche riportate a lato a contenere l'informazione fisica.

---

Per il pendolo semplice ideale l'equazione esatta è

$$
\ddot\theta=-\frac{g}{\ell}\sin\theta.
$$

Ponendo

$$
\omega=\dot\theta,
$$

lo stato può essere scritto come

$$
\mathbf z=
\begin{pmatrix}
\theta\\
\omega
\end{pmatrix},
$$

e l'equazione del secondo ordine diventa il sistema autonomo del primo ordine

$$
\boxed{
\begin{cases}
\dot\theta=\omega,\\[2mm]
\dot\omega=-\dfrac{g}{\ell}\sin\theta.
\end{cases}}
$$

In forma vettoriale,

$$
\boxed{
\dot{\mathbf z}=\mathbf f(\mathbf z)
}
$$

con

$$
\boxed{
\mathbf f(\theta,\omega)
=
\begin{pmatrix}
\omega\\[1mm]
-\dfrac{g}{\ell}\sin\theta
\end{pmatrix}.
}
$$

Il campo dinamico assegna quindi a ogni punto dello spazio degli stati il vettore che indica la direzione istantanea dell'evoluzione. Una traiettoria è una curva la cui tangente, in ogni suo punto, è orientata come questo campo.

### Perché due traiettorie non possono intersecarsi?

Supponiamo che due traiettorie differenti passino per lo stesso punto

$$
(\theta_*,\omega_*).
$$

Quel punto rappresenterebbe lo **stesso stato iniziale** per entrambe. Ma, sotto le usuali ipotesi di regolarità soddisfatte dal campo del pendolo, il problema di Cauchy possiede soluzione locale unica. Dallo stesso stato non possono quindi partire due futuri differenti.

Ne segue una proprietà geometrica fondamentale:

$$
\boxed{\text{due traiettorie distinte dello spazio degli stati non si intersecano.}}
$$

Quando una traiettoria chiusa ritorna al punto iniziale, non sta intersecando una seconda traiettoria: sta ripercorrendo periodicamente **la stessa** orbita.

### L'angolo è periodico

Fisicamente le configurazioni $\theta$ e $\theta+2\pi$ sono equivalenti. Il piano $(\theta,\dot\theta)$ che disegniamo è quindi una rappresentazione ``srotolata'' di uno spazio che, rispetto alla coordinata angolare, ha struttura cilindrica. Questa osservazione diventerà importante quando incontreremo i moti di rotazione.

````python
#@title Simulazione 6 — Costruisci progressivamente il ritratto di fase { display-mode: "form" }

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display, clear_output, HTML, Markdown, Math


# ============================================================
# CONTROLLO DEI PARAMETRI DEL PENDOLO
# ============================================================

_sim6_ready = True

if 'l_phase_w' not in globals() or 'g_phase_w' not in globals():

    _sim6_ready = False

    display(HTML(
        """
        <div style="
            background:#fde2e1;
            color:#7a1f1f;
            border-left:5px solid #c62828;
            padding:10px 14px;
            border-radius:6px;
            margin:10px 0 14px 0;
            font-weight:700;">
            ⚠️ Prima di eseguire questa simulazione devi eseguire
            l'Attività 5.1, nella quale vengono definiti
            la lunghezza ℓ e l'accelerazione di gravità g.
        </div>
        """
    ))


if _sim6_ready:

    # Valori del pendolo scelti nell'Attività 5.1
    l_sim6 = float(l_phase_w.value)
    g_sim6 = float(g_phase_w.value)


    # ========================================================
    # INFORMAZIONI SUL SISTEMA UTILIZZATO
    # ========================================================

    display(HTML(
        f"""
        <div style="
            background:#e8f1ff;
            color:#102a43;
            border-left:5px solid #3b82f6;
            padding:10px 14px;
            border-radius:6px;
            margin:10px 0 8px 0;">
            <b>Stesso pendolo dell'Attività 5.1</b><br>
            ℓ = {l_sim6:.3g} m
            &nbsp;&nbsp;&nbsp;&nbsp;
            g = {g_sim6:.3g} m/s²
        </div>
        """
    ))

    display(Markdown(r"""
In tutte le traiettorie di questa attività la condizione iniziale sulla
velocità angolare è fissata:
"""))

    display(Markdown(r"""
<div style="text-align:center; margin:10px 0;">

$$
\boxed{\dot{\theta}(0)=0}
$$

</div>
"""))


    display(Markdown(r"""
Scegli quindi soltanto l'angolo iniziale $\theta_0$ e aggiungi
progressivamente le traiettorie al grafico.
"""))


    # ========================================================
    # WIDGET: ANGOLO INIZIALE
    # ========================================================

    multi_th_w = widgets.FloatSlider(
        value=20.0,
        min=-175.0,
        max=175.0,
        step=5.0,
        description='θ(0) [°]:',
        continuous_update=True,
        readout=True,
        readout_format='.1f',
        layout=widgets.Layout(width='430px')
    )

    multi_th_manual_w = widgets.BoundedFloatText(
        value=20.0,
        min=-175.0,
        max=175.0,
        step=0.1,
        description='da tastiera:',
        style={'description_width': '95px'},
        layout=widgets.Layout(width='210px')
    )

    multi_th_link = widgets.jslink(
        (multi_th_w, 'value'),
        (multi_th_manual_w, 'value')
    )


    # ========================================================
    # WIDGET: DURATA SIMULAZIONE
    # ========================================================

    multi_t_w = widgets.FloatSlider(
        value=12.0,
        min=4.0,
        max=25.0,
        step=1.0,
        description='t [s]:',
        continuous_update=True,
        readout=True,
        readout_format='.1f',
        layout=widgets.Layout(width='430px')
    )

    multi_t_manual_w = widgets.BoundedFloatText(
        value=12.0,
        min=4.0,
        max=25.0,
        step=0.1,
        description='da tastiera:',
        style={'description_width': '95px'},
        layout=widgets.Layout(width='210px')
    )

    multi_t_link = widgets.jslink(
        (multi_t_w, 'value'),
        (multi_t_manual_w, 'value')
    )


    # ========================================================
    # PULSANTI
    # ========================================================

    add_btn = widgets.Button(
        description='Aggiungi traiettoria',
        button_style='info'
    )

    clear_btn = widgets.Button(
        description='Cancella tutte'
    )

    out_multi = widgets.Output()


    # ========================================================
    # MEMORIA DELLE TRAIETTORIE
    # ========================================================

    _traiettorie_fase = []


    # ========================================================
    # INTERFACCIA
    # ========================================================

    display(
        widgets.VBox([
            widgets.HBox([
                multi_th_w,
                multi_th_manual_w
            ]),

            widgets.HBox([
                multi_t_w,
                multi_t_manual_w
            ]),

            widgets.HBox([
                add_btn,
                clear_btn
            ]),

            out_multi
        ])
    )


    # ========================================================
    # DISEGNO DEL RITRATTO DI FASE
    # ========================================================

    def _ridisegna_multi():

        with out_multi:

            clear_output(wait=True)

            if not _traiettorie_fase:

                print(
                    'Nessuna traiettoria ancora aggiunta.'
                )

                return


            fig, ax = plt.subplots(
                figsize=(8.4, 5.4)
            )


            for tr in _traiettorie_fase:

                ax.plot(
                    tr['th'],
                    tr['om'],
                    linewidth=1.35,
                    label=rf"$\theta_0={tr['deg']:.0f}^\circ$"
                )

                # Stato iniziale
                ax.scatter(
                    [tr['th'][0]],
                    [tr['om'][0]],
                    s=22,
                    zorder=4
                )


            ax.axhline(
                0.0,
                linewidth=0.8,
                alpha=0.35
            )

            ax.axvline(
                0.0,
                linewidth=0.8,
                alpha=0.35
            )

            ax.set_xlabel(
                r'$\theta$ [rad]'
            )

            ax.set_ylabel(
                r'$\dot{\theta}$ [rad/s]'
            )

            ax.set_title(
                'Famiglia di traiettorie nello spazio degli stati'
            )

            ax.grid(
                True,
                alpha=0.25
            )

            ax.legend(
                fontsize=8,
                ncol=2
            )

            plt.show()


    # ========================================================
    # AGGIUNTA DI UNA TRAIETTORIA
    # ========================================================

    def _add_multi(_):

        try:

            # Verifica che sia stato effettuato nell'Attività 5
            # il passaggio al modello di riferimento.
            _verifica_pendolo_esatto()


            deg = float(
                multi_th_w.value
            )

            # In questa attività il pendolo parte SEMPRE da fermo.
            om0 = 0.0


            # Leggiamo i valori correnti scelti nell'Attività 5.1
            lval = float(
                l_phase_w.value
            )

            gval = float(
                g_phase_w.value
            )


            tt, th, om = _integra_pendolo(
                np.deg2rad(deg),
                om0,
                lval,
                gval,
                float(multi_t_w.value),
                n=1800
            )


            _traiettorie_fase.append(
                {
                    'deg': deg,
                    'om0': om0,
                    'l': lval,
                    'g': gval,
                    'th': th,
                    'om': om
                }
            )


            # Manteniamo al massimo 10 traiettorie
            if len(_traiettorie_fase) > 10:

                _traiettorie_fase.pop(0)


            _ridisegna_multi()


        except Exception as e:

            with out_multi:

                clear_output(wait=True)

                print(
                    'Impossibile aggiungere la traiettoria:',
                    e
                )


    # ========================================================
    # CANCELLAZIONE
    # ========================================================

    def _clear_multi(_):

        _traiettorie_fase.clear()

        _ridisegna_multi()


    # ========================================================
    # COLLEGAMENTO PULSANTI
    # ========================================================

    add_btn.on_click(
        _add_multi
    )

    clear_btn.on_click(
        _clear_multi
    )
````

````python
#@title Attività 6.1 — Piccole oscillazioni e forma della traiettoria { display-mode: "form" }

import ipywidgets as widgets
from IPython.display import display, Markdown, HTML, clear_output


# ============================================================
# FORMULE AUTONOME CENTRATE
# ============================================================

def _center_math_61(latex):
    """
    Renderizza una formula autonoma centrata in Colab usando
    Markdown + HTML + delimitatori $$ ... $$.
    """
    return Markdown(
        '<div style="text-align:center; margin:10px 0;">\n\n'
        '$$\n'
        + latex +
        '\n$$\n\n'
        '</div>'
    )



# ============================================================
# CONSEGNA INIZIALE
# ============================================================

out_intro_61 = widgets.Output()

with out_intro_61:

    display(HTML(
        """
        <div style="
            display:inline-block;
            background:#fff3a3;
            color:#111;
            padding:5px 10px;
            border-radius:5px;
            font-weight:700;
            margin:6px 0 14px 0;">
            ✍️ CARTA E PENNA
        </div>
        """
    ))

    display(Markdown(r"""
Nell'**Attività 6** hai osservato che, per piccoli valori dell'ampiezza iniziale,
la traiettoria nel piano $(\theta,\dot{\theta})$ appare molto simile a un'ellisse.

Considera il pendolo lasciato da fermo dalla configurazione iniziale
"""))

    display(_center_math_61(
        r"\theta(0)=\theta_0,"
        r"\qquad"
        r"\dot{\theta}(0)=0."
    ))

    display(Markdown(r"""
**Prova a dimostrare analiticamente** che, nel limite delle piccole oscillazioni,
la traiettoria nello spazio degli stati è approssimativamente un'ellisse.

Lavora autonomamente su carta e prova ad arrivare a un'equazione della curva
nel piano $(\theta,\dot{\theta})$.

Se riesci, non utilizzare i suggerimenti.
"""))


# ============================================================
# PULSANTI
# ============================================================

btn_hint1_61 = widgets.Button(
    description='Ho bisogno di un primo suggerimento',
    button_style='info',
    layout=widgets.Layout(width='310px')
)

btn_hint2_61 = widgets.Button(
    description='Mostra il suggerimento successivo',
    button_style='info',
    layout=widgets.Layout(width='310px', display='none')
)

btn_hint3_61 = widgets.Button(
    description='Ancora un suggerimento',
    button_style='info',
    layout=widgets.Layout(width='310px', display='none')
)

btn_solution_61 = widgets.Button(
    description='Verifica la conclusione',
    button_style='success',
    layout=widgets.Layout(width='310px')
)


# ============================================================
# OUTPUT
# ============================================================

out_hint1_61 = widgets.Output()
out_hint2_61 = widgets.Output()
out_hint3_61 = widgets.Output()
out_solution_61 = widgets.Output()


display(
    widgets.VBox([
        out_intro_61,
        widgets.HBox([
            btn_hint1_61,
            btn_solution_61
        ]),
        out_hint1_61,
        btn_hint2_61,
        out_hint2_61,
        btn_hint3_61,
        out_hint3_61,
        out_solution_61
    ])
)


# ============================================================
# SUGGERIMENTO 1
# ============================================================

def show_hint1_61(_):

    with out_hint1_61:

        clear_output(wait=True)

        display(Markdown(r"""
### Suggerimento 1 — Cerca una relazione tra $\theta$ e $\dot{\theta}$

Il pendolo considerato è un sistema ideale e conservativo.

Quale grandezza rimane costante durante il moto?

Scrivila:

- nello stato iniziale, quando
"""))

        display(_center_math_61(
            r"\theta=\theta_0,"
            r"\qquad"
            r"\dot{\theta}=0;"
        ))

        display(Markdown(r"""
- in un generico stato della traiettoria, caratterizzato da
"""))

        display(_center_math_61(
            r"(\theta,\dot{\theta})."
        ))

        display(Markdown(r"""
**Uguaglia tu le due espressioni** e prova a ricavare una relazione
contenente soltanto $\theta$, $\dot{\theta}$ e $\theta_0$.

Fermati qui e svolgi il calcolo su carta prima di continuare.
"""))

    btn_hint2_61.layout.display = ''


# ============================================================
# SUGGERIMENTO 2
# ============================================================

def show_hint2_61(_):

    with out_hint2_61:

        clear_output(wait=True)

        display(Markdown(r"""
### Suggerimento 2 — Controlla la relazione ottenuta

Se hai svolto correttamente il passaggio precedente, dovresti arrivare a:
"""))

        display(_center_math_61(
            r"\boxed{"
            r"\dot{\theta}^{\,2}"
            r"="
            r"\frac{2g}{\ell}"
            r"\left("
            r"\cos\theta-\cos\theta_0"
            r"\right)"
            r"}"
        ))

        display(Markdown(r"""
Questa relazione lega direttamente $\theta$ e $\dot{\theta}$:
è quindi l'**equazione esatta della traiettoria nello spazio degli stati**.

Ora considera l'ipotesi di **piccole oscillazioni**.

Quale approssimazione al secondo ordine puoi utilizzare per il coseno
quando il suo argomento è piccolo?

Applicala sia a $\theta$ sia a $\theta_0$ e sostituiscila
nell'equazione precedente.

**Esegui tu i calcoli prima di chiedere il suggerimento successivo.**
"""))

    btn_hint3_61.layout.display = ''


# ============================================================
# SUGGERIMENTO 3
# ============================================================

def show_hint3_61(_):

    with out_hint3_61:

        clear_output(wait=True)

        display(Markdown(r"""
### Suggerimento 3 — Riconosci la curva

Per piccoli angoli puoi utilizzare
"""))

        display(_center_math_61(
            r"\cos\theta"
            r"\simeq"
            r"1-\frac{\theta^2}{2},"
            r"\qquad"
            r"\cos\theta_0"
            r"\simeq"
            r"1-\frac{\theta_0^2}{2}."
        ))

        display(Markdown(r"""
Ora **esegui tu la sostituzione e semplifica**.

Prova quindi a portare l'equazione ottenuta nella forma
"""))

        display(_center_math_61(
            r"\frac{\theta^2}{A^2}"
            r"+"
            r"\frac{\dot{\theta}^{\,2}}{B^2}"
            r"=1."
        ))

        display(Markdown(r"""
Determina autonomamente **A** e **B**.
"""))

        display(Markdown(r"""
e infine chiediti:

**quale curva geometrica ha questa equazione?**

Quando hai concluso il ragionamento, **torna in alto**, dove trovi il pulsante verde
**«Verifica la conclusione»**, e confronta il risultato con la tua deduzione.
"""))


# ============================================================
# VERIFICA FINALE
# ============================================================

def show_solution_61(_):

    with out_solution_61:

        clear_output(wait=True)

        display(HTML(
    """
    <div style="
        background:#d9f2dc;
        color:#173b1d;
        border-left:5px solid #28a745;
        padding:9px 13px;
        margin:8px 0 14px 0;
        border-radius:5px;
        font-size:18px;
        font-weight:700;">
        ✓ Verifica della conclusione
    </div>
    """
))

        display(Markdown(r"""
Sostituendo le approssimazioni dei coseni nell'equazione della traiettoria
si ottiene
"""))

        display(_center_math_61(
            r"\dot{\theta}^{\,2}"
            r"\simeq"
            r"\frac{g}{\ell}"
            r"\left("
            r"\theta_0^2-\theta^2"
            r"\right)."
        ))

        display(Markdown(r"""
Riordinando:
"""))

        display(_center_math_61(
            r"\boxed{"
            r"\frac{\theta^2}{\theta_0^2}"
            r"+"
            r"\frac{\dot{\theta}^{\,2}}"
            r"{(g/\ell)\theta_0^2}"
            r"=1"
            r"}"
        ))

        display(Markdown(r"""
Pertanto
"""))

        display(_center_math_61(
            r"A=\theta_0,"
            r"\qquad"
            r"B=\theta_0\sqrt{\frac{g}{\ell}}."
        ))

        display(Markdown(r"""
La curva è un'**ellisse** nel piano $(\theta,\dot{\theta})$.

Dunque ciò che hai osservato numericamente nell'Attività 6 ha una
spiegazione analitica:
"""))

        display(_center_math_61(
            r"\boxed{"
            r"\text{piccole oscillazioni}"
            r"\quad\Longrightarrow\quad"
            r"\text{traiettorie quasi ellittiche}"
            r"}"
        ))

        display(Markdown(r"""
L'ellisse non è la traiettoria esatta del pendolo per qualunque ampiezza:
emerge come conseguenza dell'approssimazione valida per piccoli angoli.
"""))


# ============================================================
# COLLEGAMENTO PULSANTI
# ============================================================

btn_hint1_61.on_click(show_hint1_61)
btn_hint2_61.on_click(show_hint2_61)
btn_hint3_61.on_click(show_hint3_61)
btn_solution_61.on_click(show_solution_61)
````

````python
#@title Attività 6.2 — Come si deformano le traiettorie alle grandi ampiezze? { display-mode: "form" }

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from scipy.integrate import solve_ivp
from IPython.display import display, Markdown, Math, HTML, clear_output
from matplotlib.figure import Figure

# ============================================================
# FORMULE AUTONOME CENTRATE
# ============================================================

def _center_math_62(latex):
    """
    Renderizza una formula autonoma centrata in Colab usando
    Markdown + HTML + delimitatori $$ ... $$.
    """
    return Markdown(
        '<div style="text-align:center; margin:10px 0;">\n\n'
        '$$\n'
        + latex +
        '\n$$\n\n'
        '</div>'
    )



# ============================================================
# CONSEGNA INIZIALE
# ============================================================

out_intro_62 = widgets.Output()

with out_intro_62:

    display(HTML(
        """
        <div style="
            display:inline-block;
            background:#fff3a3;
            color:#111;
            padding:5px 10px;
            border-radius:5px;
            font-weight:700;
            margin:6px 0 14px 0;">
            ✍️ CARTA E PENNA
        </div>
        """
    ))

    display(Markdown(r"""
Nell'**Attività 6** avrai notato che, aumentando l'ampiezza iniziale
$\theta_0$, sempre con **partenza da fermo**,
"""))

    display(Markdown(r"""
<div style="text-align:center; margin:10px 0;">

$$
\dot{\theta}(0)=0,
$$

</div>
"""))


    display(Markdown(r"""
le traiettorie nello spazio degli stati si allontanano progressivamente
dalla forma quasi ellittica osservata per le piccole oscillazioni.

In particolare, quando $\theta_0$ si avvicina a $180^\circ$, la traiettoria
appare sempre più **appuntita** nei punti di massima ampiezza angolare,
cioè nei punti nei quali il pendolo si ferma e inverte il moto.

Consideriamo il punto di inversione che si avvicina a
"""))

    display(Markdown(r"""
<div style="text-align:center; margin:10px 0;">

$$
\theta=\pi.
$$

</div>
"""))


    display(Markdown(r"""
**Prova a spiegare analiticamente perché la traiettoria assume
progressivamente questa forma quando $\theta_0$ si avvicina a $\pi$.**

Lavora autonomamente su carta.

Se riesci, non utilizzare i suggerimenti.
"""))


# ============================================================
# PULSANTI
# ============================================================

btn_hint1_62 = widgets.Button(
    description='Ho bisogno di un primo suggerimento',
    button_style='info',
    layout=widgets.Layout(width='310px')
)

btn_hint2_62 = widgets.Button(
    description='Mostra il suggerimento successivo',
    button_style='info',
    layout=widgets.Layout(width='310px', display='none')
)

btn_solution_62 = widgets.Button(
    description='Verifica la conclusione',
    button_style='success',
    layout=widgets.Layout(width='310px')
)


# ============================================================
# OUTPUT
# ============================================================

out_hint1_62 = widgets.Output()
out_hint2_62 = widgets.Output()
out_solution_62 = widgets.Output()

out_verify_controls_62 = widgets.Output()
out_verify_plot_62 = widgets.Output()


display(
    widgets.VBox([
        out_intro_62,

        widgets.HBox([
            btn_hint1_62,
            btn_solution_62
        ]),

        out_hint1_62,
        btn_hint2_62,
        out_hint2_62,

        out_solution_62,

        out_verify_controls_62,
        out_verify_plot_62
    ])
)


# ============================================================
# SUGGERIMENTO 1
# ============================================================

def show_hint1_62(_):

    with out_hint1_62:

        clear_output(wait=True)

        display(Markdown(r"""
### Suggerimento 1 — Riparti dall'equazione della traiettoria

Nell'attività precedente hai ottenuto l'equazione esatta della traiettoria
per un pendolo lasciato da fermo da $\theta_0$:
"""))

        display(_center_math_62(
            r"\dot{\theta}^{\,2}"
            r"="
            r"\frac{2g}{\ell}"
            r"\left("
            r"\cos\theta-\cos\theta_0"
            r"\right)."
        ))

        display(Markdown(r"""
Ora vogliamo studiare la forma della traiettoria quando l'ampiezza iniziale
è vicina a $180^\circ$ e, in particolare, quando $\theta$ assume valori
vicini a $\pi$.

Procedi così:

1. sviluppa $\cos\theta$ in **serie di Taylor attorno a $\theta=\pi$**,
   fermandoti al secondo ordine;

2. considera un angolo iniziale leggermente inferiore a $\pi$ e scrivilo come
"""))

        display(_center_math_62(
            r"\theta_0=\pi-\delta_0,"
        ))

        display(Markdown(r"""
dove $\delta_0$ rappresenta il piccolo scostamento iniziale da $\pi$;

3. usa la formula di sottrazione del coseno per esprimere
"""))

        display(_center_math_62(
            r"\cos(\pi-\delta_0)"
        ))

        display(Markdown(r"""
e sviluppa al secondo ordine ciò che ottieni;

4. sostituisci infine entrambe le approssimazioni
nell'equazione della traiettoria.

**Svolgi tutti i passaggi su carta prima di chiedere il suggerimento successivo.**
"""))

    btn_hint2_62.layout.display = ''


# ============================================================
# SUGGERIMENTO 2
# ============================================================

def show_hint2_62(_):

    with out_hint2_62:

        clear_output(wait=True)

        display(Markdown(r"""
### Suggerimento 2 — Studia direttamente lo scostamento da $\pi$

Nel lavoro innescato dal Suggerimento 1 dovresti essere arrivato a
"""))

        display(_center_math_62(
            r"\dot{\theta}^{\,2}"
            r"\simeq"
            r"\frac{2g}{\ell}"
            r"\left["
            r"-1+\frac{1}{2}(\theta-\pi)^2"
            r"+1-\frac{1}{2}\delta_0^2"
            r"\right]."
        ))

        display(Markdown(r"""
Ora concentriamoci sulla distanza angolare dalla configurazione $\theta=\pi$.

Definisci
"""))

        display(_center_math_62(
            r"\delta=\theta-\pi."
        ))

        display(Markdown(r"""
$\delta$ è quindi lo **scostamento, con segno, dell'angolo $\theta$ da $\pi$**.

Riscrivi l'equazione precedente utilizzando soltanto
$\delta$, $\delta_0$, $\dot{\theta}$, $g$ e $\ell$, e semplificala.

Poi considera il limite in cui il punto di inversione si avvicina sempre più
alla posizione superiore.

Poiché
"""))

        display(_center_math_62(
            r"\theta_0=\pi-\delta_0,"
        ))

        display(Markdown(r"""
questo corrisponde a far tendere $\delta_0$ a zero.

Applica quindi il limite
"""))

        display(_center_math_62(
            r"\delta_0\to0"
        ))

        display(Markdown(r"""
alla relazione che hai ottenuto.

A questo punto ricava le **due possibili relazioni tra $\dot{\theta}$ e $\delta$**
e chiediti:

**che tipo di curve rappresentano nel piano $(\theta,\dot{\theta})$?**

E soprattutto:

**che cosa ci dicono sulla forma assunta dalla traiettoria vicino al punto
di inversione quando $\theta_0$ si avvicina a $\pi$?**

Quando hai concluso il ragionamento, **torna in alto**, dove trovi il pulsante
verde **«Verifica la conclusione»**, e confronta il risultato con la tua deduzione.
"""))

# ============================================================
# PARAMETRI PER LA VERIFICA NUMERICA
# ============================================================

# Qui memorizziamo direttamente le traiettorie già calcolate.
# In questo modo, quando se ne aggiunge una nuova,
# NON vengono ricalcolate quelle precedenti.
_traiettorie_62 = []

# Parametri con cui sono state costruite le traiettorie in cache
_parametri_cache_62 = None


theta0_verify_62 = widgets.FloatSlider(
    value=30.0,
    min=5.0,
    max=175.0,
    step=5.0,
    description='θ₀ [°]:',
    continuous_update=False,
    readout=True,
    readout_format='.0f',
    layout=widgets.Layout(width='500px')
)

btn_add_62 = widgets.Button(
    description='Aggiungi traiettoria',
    button_style='info',
    layout=widgets.Layout(width='220px')
)

btn_clear_62 = widgets.Button(
    description='Pulisci grafico',
    layout=widgets.Layout(width='180px')
)




# Banner persistente:
# viene visualizzato una volta e poi il suo contenuto viene aggiornato
_banner_pendolo_62 = widgets.HTML()




# ============================================================
# PARAMETRI DEL PENDOLO
# ============================================================

def _get_pendulum_parameters_62():

    if (
        'l_phase_w' not in globals()
        or
        'g_phase_w' not in globals()
    ):
        raise RuntimeError(
            "Prima di eseguire questa verifica devi eseguire "
            "l'Attività 5.1, nella quale vengono definiti ℓ e g."
        )

    lval = float(l_phase_w.value)
    gval = float(g_phase_w.value)

    return lval, gval


# ============================================================
# BANNER ESPLICITO
# ============================================================

def _update_banner_62(lval, gval):

    _banner_pendolo_62.value = f"""
    <div style="
        background-color:#e8f1ff !important;
        color:#102a43 !important;
        border-left:5px solid #3b82f6;
        padding:11px 15px;
        border-radius:6px;
        margin:8px 0 14px 0;
        width:100%;
        box-sizing:border-box;">

        <b>Pendolo utilizzato nella verifica grafica</b><br><br>

        In questa attività stiamo lavorando con lo
        <b>stesso pendolo definito nell'Attività 5.1</b>:

        <br><br>

        <span style="font-size:17px;">
            ℓ = <b>{lval:.3g} m</b>
            &nbsp;&nbsp;&nbsp;&nbsp;
            g = <b>{gval:.3g} m/s²</b>
        </span>

        <br><br>

        Per tutte le traiettorie:
        <b>θ̇(0) = 0</b>.

        Viene modificato soltanto l'angolo iniziale θ₀.
    </div>
    """


# ============================================================
# INTEGRAZIONE DI UNA SINGOLA TRAIETTORIA
# ============================================================

def _integrate_phase_62(theta0_deg, lval, gval):

    # Controlliamo che sia stato eseguito il passaggio
    # al modello di riferimento nell'Attività 5.
    _verifica_pendolo_esatto()

    _, th, om = _integra_pendolo(
        np.deg2rad(theta0_deg),
        0.0,             # il pendolo parte sempre da fermo
        lval,
        gval,
        24.0,
        n=3500
    )

    return th, om


# ============================================================
# COSTRUZIONE DEL GRAFICO
# ============================================================

def _make_figure_62(lval, gval):

    slope = np.sqrt(gval / lval)

    fig = Figure(
    figsize=(9.0, 6.0)
    )

    ax = fig.add_subplot(111)



    # --------------------------------------------------------
    # DIREZIONI LIMITE VICINO A +pi
    # --------------------------------------------------------

    theta_right = np.linspace(
        np.pi - 0.95,
        np.pi,
        250
    )

    delta_right = theta_right - np.pi

    ax.plot(
        theta_right,
        +slope * delta_right,
        '--',
        linewidth=1.5,
        label=r'direzioni limite vicino a $+\pi$'
    )

    ax.plot(
        theta_right,
        -slope * delta_right,
        '--',
        linewidth=1.5
    )


    # --------------------------------------------------------
    # DIREZIONI LIMITE VICINO A -pi
    # --------------------------------------------------------

    theta_left = np.linspace(
        -np.pi,
        -np.pi + 0.95,
        250
    )

    delta_left = theta_left + np.pi

    ax.plot(
        theta_left,
        +slope * delta_left,
        '--',
        linewidth=1.5,
        alpha=0.8,
        label=r'direzioni limite vicino a $-\pi$'
    )

    ax.plot(
        theta_left,
        -slope * delta_left,
        '--',
        linewidth=1.5,
        alpha=0.8
    )


    # --------------------------------------------------------
    # TRAIETTORIE GIÀ CALCOLATE
    # --------------------------------------------------------

    for tr in _traiettorie_62:

        ax.plot(
            tr['theta'],
            tr['omega'],
            linewidth=1.5,
            label=rf"$\theta_0={tr['angle']:.0f}^\circ$"
        )

        # stato iniziale
        ax.scatter(
            [np.deg2rad(tr['angle'])],
            [0.0],
            s=24,
            zorder=4
        )


    # --------------------------------------------------------
    # ASPETTO DEL GRAFICO
    # --------------------------------------------------------

    ax.axhline(
        0.0,
        linewidth=0.8,
        alpha=0.30
    )

    ax.axvline(
        0.0,
        linewidth=0.8,
        alpha=0.30
    )

    ax.set_xlabel(
        r'$\theta$ [rad]'
    )

    ax.set_ylabel(
        r'$\dot{\theta}$ [rad/s]'
    )

    ax.set_title(
        r'Forma delle traiettorie per '
        r'$\theta_0\rightarrow180^\circ$'
    )

    ax.set_xlim(
        -3.35,
        3.35
    )

    ax.grid(
        True,
        alpha=0.25
    )

    ax.legend(
        loc='best'
    )

    fig.tight_layout()

    return fig


# ============================================================
# AGGIORNAMENTO DELL'UNICO GRAFICO
# ============================================================

def _draw_verification_62():

    try:

        # ----------------------------------------------------
        # PARAMETRI DELLO STESSO PENDOLO DELLA 5.1
        # ----------------------------------------------------

        lval, gval = _get_pendulum_parameters_62()

        _update_banner_62(
            lval,
            gval
        )


        # ----------------------------------------------------
        # COSTRUIAMO PRIMA COMPLETAMENTE IL NUOVO GRAFICO
        # ----------------------------------------------------

        fig = _make_figure_62(
            lval,
            gval
        )


        # ----------------------------------------------------
        # SOLO ORA SOSTITUIAMO IL GRAFICO PRECEDENTE
        # ----------------------------------------------------

        with out_verify_plot_62:

            clear_output(wait=True)

            display(
                _banner_pendolo_62
            )

            display(
                fig
            )


    except Exception as e:

        with out_verify_plot_62:

            clear_output(wait=True)

            display(HTML(
                f"""
                <div style="
                    background:#fde2e1;
                    color:#7a1f1f;
                    border-left:5px solid #c62828;
                    padding:10px 14px;
                    border-radius:6px;
                    margin:8px 0;">
                    ⚠️ Impossibile eseguire la verifica numerica:
                    {str(e)}
                </div>
                """
            ))


# ============================================================
# AGGIUNTA DI UNA TRAIETTORIA
# ============================================================

def _add_trajectory_62(_):

    global _parametri_cache_62

    try:

        angle = float(
            theta0_verify_62.value
        )

        lval, gval = _get_pendulum_parameters_62()

        parametri_correnti = (
            lval,
            gval
        )


        # ----------------------------------------------------
        # SE l O g SONO CAMBIATI NELLA 5.1
        # ----------------------------------------------------

        # Le vecchie traiettorie appartengono a un altro
        # pendolo e quindi vengono eliminate.
        if (
            _parametri_cache_62 is not None
            and
            parametri_correnti != _parametri_cache_62
        ):

            _traiettorie_62.clear()


        _parametri_cache_62 = (
            parametri_correnti
        )


        # ----------------------------------------------------
        # EVITIAMO DUPLICATI
        # ----------------------------------------------------

        for tr in _traiettorie_62:

            if tr['angle'] == angle:
                return


        # ----------------------------------------------------
        # CALCOLIAMO SOLTANTO LA NUOVA TRAIETTORIA
        # ----------------------------------------------------

        th, om = _integrate_phase_62(
            angle,
            lval,
            gval
        )


        # ----------------------------------------------------
        # LA MEMORIZZIAMO
        # ----------------------------------------------------

        _traiettorie_62.append(
            {
                'angle': angle,
                'theta': th,
                'omega': om
            }
        )

        _traiettorie_62.sort(
            key=lambda tr: tr['angle']
        )


        # ----------------------------------------------------
        # AGGIORNIAMO LO STESSO GRAFICO
        # ----------------------------------------------------

        _draw_verification_62()


    except Exception as e:

        _banner_pendolo_62.value = f"""
        <div style="
            background:#fde2e1;
            color:#7a1f1f;
            border-left:5px solid #c62828;
            padding:10px 14px;
            border-radius:6px;">
            ⚠️ Impossibile aggiungere la traiettoria:
            {str(e)}
        </div>
        """


# ============================================================
# PULIZIA
# ============================================================

def _clear_trajectories_62(_):

    _traiettorie_62.clear()

    _draw_verification_62()


# ============================================================
# COLLEGAMENTO PULSANTI
# ============================================================

btn_add_62.on_click(
    _add_trajectory_62
)

btn_clear_62.on_click(
    _clear_trajectories_62
)

# ============================================================
# VERIFICA FINALE
# ============================================================

def show_solution_62(_):

    with out_solution_62:

        clear_output(wait=True)

        display(HTML(
            """
            <div style="
                background:#d9f2dc;
                color:#173b1d;
                border-left:5px solid #28a745;
                padding:9px 13px;
                margin:8px 0 14px 0;
                border-radius:5px;
                font-size:18px;
                font-weight:700;">
                ✓ Verifica della conclusione
            </div>
            """
        ))

        display(Markdown(r"""
Nel primo passaggio hai sviluppato $\cos\theta$ attorno a $\theta=\pi$.
Al secondo ordine:
"""))

        display(_center_math_62(
            r"\cos\theta"
            r"\simeq"
            r"-1+\frac{1}{2}(\theta-\pi)^2."
        ))

        display(Markdown(r"""
Per l'angolo iniziale
"""))

        display(_center_math_62(
            r"\theta_0=\pi-\delta_0,"
        ))

        display(Markdown(r"""
si ha invece
"""))

        display(_center_math_62(
            r"\cos\theta_0"
            r"="
            r"\cos(\pi-\delta_0)"
            r"="
            r"-\cos\delta_0"
            r"\simeq"
            r"-1+\frac{\delta_0^2}{2}."
        ))

        display(Markdown(r"""
Sostituendo nell'equazione della traiettoria:
"""))

        display(_center_math_62(
            r"\dot{\theta}^{\,2}"
            r"\simeq"
            r"\frac{2g}{\ell}"
            r"\left["
            r"-1+\frac{1}{2}(\theta-\pi)^2"
            r"+1-\frac{1}{2}\delta_0^2"
            r"\right]."
        ))

        display(Markdown(r"""
Nel secondo passaggio hai posto
"""))

        display(_center_math_62(
            r"\delta=\theta-\pi."
        ))

        display(Markdown(r"""
La relazione diventa quindi
"""))

        display(_center_math_62(
            r"\boxed{"
            r"\dot{\theta}^{\,2}"
            r"\simeq"
            r"\frac{g}{\ell}"
            r"\left("
            r"\delta^2-\delta_0^2"
            r"\right)"
            r"}"
        ))

        display(Markdown(r"""
Ora facciamo avvicinare il punto di inversione alla posizione superiore:
"""))

        display(_center_math_62(
            r"\delta_0\to0."
        ))

        display(Markdown(r"""
Nel limite:
"""))

        display(_center_math_62(
            r"\dot{\theta}^{\,2}"
            r"\simeq"
            r"\frac{g}{\ell}\delta^2."
        ))

        display(Markdown(r"""
Prendendo entrambe le radici:
"""))

        display(_center_math_62(
            r"\boxed{"
            r"\dot{\theta}"
            r"\simeq"
            r"\pm\sqrt{\frac{g}{\ell}}\,\delta"
            r"}"
        ))

        display(Markdown(r"""
e quindi, poiché $\delta=\theta-\pi$,
"""))

        display(_center_math_62(
            r"\boxed{"
            r"\dot{\theta}"
            r"\simeq"
            r"\pm\sqrt{\frac{g}{\ell}}"
            r"(\theta-\pi)"
            r"}"
        ))

        display(Markdown(r"""
Nel piano $(\theta,\dot{\theta})$ queste sono **due rette che passano
per $(\pi,0)$**.

Quando $\theta_0$ si avvicina sempre più a $\pi$, i due rami della traiettoria
nelle vicinanze del punto di inversione si dispongono progressivamente
lungo queste due direzioni.

Questo spiega perché, aumentando l'ampiezza, le traiettorie osservate
nell'Attività 6 appaiono sempre più **appuntite** vicino ai punti
di inversione.

Le rette non costituiscono però un vero spigolo della traiettoria per
$\theta_0<\pi$: rappresentano le **direzioni limite** verso cui i rami
si dispongono quando il punto di inversione si avvicina a $(\pi,0)$.

Infine, se ponessimo esattamente
$\theta_0=\pi$ e $\dot{\theta}(0)=0$, il pendolo ideale rimarrebbe fermo
nella posizione superiore. Stiamo quindi studiando il limite
$\theta_0\to\pi$, non il moto con $\theta_0=\pi$ esatto.
"""))


    # ========================================================
    # VERIFICA NUMERICA
    # ========================================================

    with out_verify_controls_62:

        clear_output(wait=True)

        display(Markdown(r"""
### Verifica numerica

Ora confronta la previsione analitica con le traiettorie numeriche.

Nel grafico sono disegnate sullo sfondo le **due direzioni limite**
che hai appena ricavato.

Aggiungi una traiettoria alla volta mantenendo
"""))

        display(_center_math_62(
            r"\dot{\theta}(0)=0"
        ))

        display(Markdown(r"""
e aumenta progressivamente $\theta_0$ fino a $175^\circ$.

Osserva come i rami della traiettoria, vicino ai punti di inversione,
si avvicinino progressivamente alle direzioni previste analiticamente.
"""))

        display(
            widgets.VBox([
                theta0_verify_62,

                widgets.HBox([
                    btn_add_62,
                    btn_clear_62
                ])
            ])
        )

    _draw_verification_62()


# ============================================================
# COLLEGAMENTO PULSANTI
# ============================================================

btn_hint1_62.on_click(show_hint1_62)
btn_hint2_62.on_click(show_hint2_62)
btn_solution_62.on_click(show_solution_62)
````

## Teoria — Dal diagramma di una traiettoria al ritratto di fase

Una singola curva nel piano $(\theta,\dot\theta)$ rappresenta l'evoluzione associata a una particolare condizione iniziale. L'insieme qualitativo delle traiettorie generate da condizioni iniziali differenti prende il nome di **ritratto di fase**.

Il ritratto di fase non mostra direttamente il tempo sull'asse orizzontale. Mostra invece **come sono organizzati gli stati possibili e le loro evoluzioni**.

Per le condizioni iniziali che producono oscillazioni, le traiettorie osservate sono curve chiuse attorno allo stato

$$
(\theta,\dot\theta)=(0,0).
$$

Aumentando l'ampiezza, le curve si allontanano da questo punto e cambiano forma: il sistema non è più ben descritto dall'approssimazione armonica. La deformazione delle orbite è quindi una manifestazione geometrica della non linearità

$$
\sin\theta\neq\theta
$$

per angoli grandi.

Il ritratto di fase permette inoltre di porre una domanda che nel solo grafico $\theta(t)$ è meno immediata: **esistono stati che non evolvono affatto?**

````python
#@title Attività 7 — Quali stati non evolvono nel tempo? { display-mode: "form" }

from IPython.display import display, Markdown, Math, HTML


display(Markdown(r"""


Riassumiamo un momento ciò che abbiamo costruito.

Lo **spazio degli stati** del pendolo è costituito dai punti
"""))

display(HTML('<div style="height:12px;"></div>'))

display(Math(
    r"\mathbf z="
    r"\begin{pmatrix}"
    r"\theta\\"
    r"\omega"
    r"\end{pmatrix},"
    r"\qquad"
    r"\omega=\dot{\theta}."
))
display(HTML('<div style="height:12px;"></div>'))
display(Markdown(r"""
Durante il moto, lo stato cambia nel tempo e il punto rappresentativo
"""))
display(HTML('<div style="height:12px;"></div>'))

display(Math(
    r"\mathbf z(t),"
   ))

display(HTML('<div style="height:12px;"></div>'))

display(Markdown(r"""
descrive una **traiettoria nello spazio degli stati**.

Il vettore tangente alla traiettoria è la derivata dello stato rispetto al tempo:
"""))
display(HTML('<div style="height:12px;"></div>'))

display(Math(
    r"\dot{\mathbf z}(t)"
    r"="
    r"\frac{d\mathbf z}{dt}."
))
display(HTML('<div style="height:12px;"></div>'))
display(Markdown(r"""
Per il pendolo abbiamo scritto
"""))
display(HTML('<div style="height:12px;"></div>'))
display(Math(
    r"\dot{\mathbf z}"
    r"="
    r"\mathbf f(\theta,\omega)"
    r"="
    r"\begin{pmatrix}"
    r"\omega\\[2mm]"
    r"-\dfrac{g}{\ell}\sin\theta"
    r"\end{pmatrix}."
))


# ============================================================
# CARTA E PENNA
# ============================================================

display(HTML(
        """
       <br><br> <div style="
            display:inline-block;
            background:#fff3a3;
            color:#111;
            padding:5px 10px;
            border-radius:5px;
            font-weight:700;
            margin:6px 0 14px 0;">
            ✍️ CARTA E PENNA
        </div> <br>
        """
    ))



display(Markdown(r"""
Immagina ora uno stato nel quale il sistema **non evolva nel tempo**:
una volta posto il pendolo in quella condizione, il punto che lo rappresenta
nello spazio degli stati rimane fermo.

**Che cosa deve significare questa condizione dal punto di vista matematico?**

Partendo dalle equazioni precedenti, prova a determinare tutte le condizioni iniziali
"""))

display(HTML('<div style="height:12px;"></div>'))
display(Math(
    r"(\theta_0,\omega_0)"
))
display(HTML('<div style="height:12px;"></div>'))

display(Markdown(r"""
per le quali il sistema, se posto inizialmente in quello stato,
rimarrebbe indefinitamente nello stesso punto dello spazio degli stati.
"""))
````

````python
#@title Soluzione Attività 7  { display-mode: "form" }

from IPython.display import display, HTML, Math
import ipywidgets as widgets

# Pulsante
btn = widgets.Button(
    description="Mostra soluzione",
    button_style="",
    tooltip="Visualizza la soluzione dell'Attività 7"
)

# Area in cui comparirà la soluzione
out = widgets.Output()


def mostra_soluzione(_):

    with out:

        out.clear_output()

        display(HTML("""
        <h3>Soluzione — Stati di equilibrio</h3>

        <p>
        Se il punto che rappresenta il sistema deve rimanere fermo
        nello spazio degli stati, il suo vettore tangente deve essere nullo.
        </p><br>
        """))


        display(HTML('<div style="height:12px;"></div>'))
        display(Math(
            r"\boxed{"
            r"\dot{\mathbf z}"
            r"="
            r"\mathbf f(\mathbf z)"
            r"="
            r"\begin{pmatrix}"
            r"0\\[2mm]"
            r"0"
            r"\end{pmatrix}"
            r"}"
        ))

        display(HTML("<br>"))

        display(HTML("""
        <p>Per il pendolo abbiamo:</p>
        """))

        display(HTML('<div style="height:12px;"></div>'))
        display(Math(
            r"\mathbf f(\theta,\omega)"
            r"="
            r"\begin{pmatrix}"
            r"\omega\\[3mm]"
            r"-\dfrac{g}{\ell}\sin\theta"
            r"\end{pmatrix}"
        ))

        display(HTML("<br>"))

        display(HTML("""
        <p>
        Affinché il vettore sia nullo, le sue due componenti
        devono quindi annullarsi simultaneamente:
        </p>
        """))


        display(HTML('<div style="height:12px;"></div>'))
        display(Math(
            r"\begin{cases}"
            r"\omega=0,\\[3mm]"
            r"-\dfrac{g}{\ell}\sin\theta=0."
            r"\end{cases}"
        ))

        display(HTML("<br>"))

        display(HTML("""
        <p>
        Poiché <b>g</b> e <b>ℓ</b> sono diversi da zero,
        la seconda condizione richiede:
        </p>
        """))

        display(HTML('<div style="height:12px;"></div>'))
        display(Math(
            r"\sin\theta=0"
        ))

        display(HTML("<br>"))

        display(HTML("""
<p>e quindi:</p>
"""))

        display(HTML('<div style="height:12px;"></div>'))
        display(Math(
    r"\boxed{\theta=k\pi,\qquad \omega=0,\qquad k\in\mathbb{Z}}"
))

        display(HTML("<br><br>"))

        display(HTML("""
        <p>
        Gli stati di equilibrio sono dunque:
        </p>
        """))

        display(HTML('<div style="height:12px;"></div>'))
        display(Math(
            r"\boxed{"
            r"\mathbf z_k"
            r"="
            r"\begin{pmatrix}"
            r"k\pi\\[2mm]"
            r"0"
            r"\end{pmatrix}"
            r"}"
        ))

        display(HTML("<br>"))

        display(HTML("""
        <p>
        Considerando l'angolo modulo <b>2π</b>,
        emergono due configurazioni di equilibrio
        fisicamente distinte:
        </p>
        """))

        display(HTML('<div style="height:12px;"></div>'))
        display(Math(
            r"(\theta,\omega)=(0,0)"
            r"\qquad\text{e}\qquad"
            r"(\theta,\omega)=(\pi,0)"
        ))

        display(HTML("<br><br>"))

        display(HTML("""
        <p>
        In entrambi i casi, se il sistema viene posto
        <strong>esattamente</strong> nello stato di equilibrio,
        il punto rappresentativo rimane fermo nello spazio degli stati.
        </p>

        <p>
        Immagina ora di <strong>perturbare leggermente</strong>
        ciascuno dei due equilibri, cioè di spostare di poco
        il sistema dalla configurazione esatta di equilibrio.
        </p>

        <p>
        <strong>Prevedi prima della simulazione:</strong>
        la perturbazione rimarrà piccola in entrambi i casi
        oppure i due equilibri avranno un comportamento
        qualitativamente diverso?
        </p>
        """))


btn.on_click(mostra_soluzione)

display(btn, out)
````

````python
#@title Attività 7.1 — Perturba i due equilibri: moto reale e spazio degli stati { display-mode: "form" }

import json
import uuid
import numpy as np
import ipywidgets as widgets

from IPython.display import display, clear_output, HTML


# ============================================================
# CONTROLLI
# ============================================================

# Slider perturbazione.
# continuous_update=True:
# il valore numerico viene aggiornato mentre trascini il cursore.
pert_w = widgets.FloatSlider(
    value=2.0,
    min=-15.0,
    max=15.0,
    step=0.1,
    description='perturb. ε [°]:',
    continuous_update=True,
    readout=True,
    readout_format='.1f',
    style={'description_width': '120px'},
    layout=widgets.Layout(width='430px')
)

# Campo numerico per inserire un valore esatto da tastiera:
# ad esempio 0, -0.01, 0.025, 3.27...
pert_exact_w = widgets.BoundedFloatText(
    value=2.0,
    min=-15.0,
    max=15.0,
    step=0.01,
    description='valore esatto [°]:',
    style={'description_width': '120px'},
    layout=widgets.Layout(width='250px')
)

# Collegamento bidirezionale lato browser:
# slider <--> campo numerico
pert_link = widgets.jslink(
    (pert_w, 'value'),
    (pert_exact_w, 'value')
)


pert_t_w = widgets.FloatSlider(
    value=8.0,
    min=3.0,
    max=15.0,
    step=1.0,
    description='t finale [s]:',
    continuous_update=True,
    readout=True,
    readout_format='.1f',
    style={'description_width': '100px'},
    layout=widgets.Layout(width='350px')
)

pert_t_exact_w = widgets.BoundedFloatText(
    value=8.0,
    min=3.0,
    max=15.0,
    step=0.1,
    description='da tastiera:',
    style={'description_width': '95px'},
    layout=widgets.Layout(width='210px')
)

pert_t_link = widgets.jslink(
    (pert_t_w, 'value'),
    (pert_t_exact_w, 'value')
)


pert_btn = widgets.Button(
    description='Avvia simulazione',
    icon='play'
)

out_pert = widgets.Output()


display(
    widgets.VBox([
        widgets.HBox([
            pert_w,
            pert_exact_w
        ]),
        widgets.HBox([
            pert_t_w,
            pert_t_exact_w,
            pert_btn
        ]),
        out_pert
    ])
)


# ============================================================
# LIMITI DELLO SPAZIO DEGLI STATI
# ============================================================

def _limiti_fase_inferiore(th, om):
    """
    Per l'equilibrio inferiore il grafico si adatta
    automaticamente alla piccola orbita prodotta.
    """

    th = np.asarray(th, dtype=float)
    om = np.asarray(om, dtype=float)

    xmin = min(float(np.min(th)), 0.0)
    xmax = max(float(np.max(th)), 0.0)

    ymin = min(float(np.min(om)), 0.0)
    ymax = max(float(np.max(om)), 0.0)

    # Limiti minimi per evitare un grafico degenerato
    # nel caso epsilon = 0.
    dx = max(xmax - xmin, 0.18)
    dy = max(ymax - ymin, 0.30)

    xm = 0.5 * (xmin + xmax)
    ym = 0.5 * (ymin + ymax)

    return [
        xm - 0.60 * dx,
        xm + 0.60 * dx,
        ym - 0.60 * dy,
        ym + 0.60 * dy
    ]


def _limiti_fase_superiore(th, om):
    """
    Per l'equilibrio superiore teniamo fisso l'intervallo:

        circa -pi  -->  3pi

    così perturbazioni positive e negative sono mostrate
    nello stesso sistema di riferimento.
    """

    om = np.asarray(om, dtype=float)

    margine_theta = 0.35

    xmin = -np.pi - margine_theta
    xmax =  3.0 * np.pi + margine_theta

    # Asse omega simmetrico rispetto a zero
    om_max = max(
        float(np.max(np.abs(om))),
        0.50
    )

    return [
        xmin,
        xmax,
        -1.10 * om_max,
         1.10 * om_max
    ]


# ============================================================
# SIMULAZIONE
# ============================================================

def _perturba_sincrona(_):

    with out_pert:

        clear_output(wait=True)

        try:

            # Verifica che T e V inserite precedentemente
            # conducano effettivamente al pendolo semplice.
            _verifica_pendolo_esatto()


            # --------------------------------------------------------
            # PARAMETRI
            # --------------------------------------------------------

            eps_deg = float(pert_exact_w.value)

            eps = np.deg2rad(eps_deg)

            tmax = float(pert_t_w.value)


            # Stesso pendolo definito nell'Attività 5.1
            if (
                'l_phase_w' not in globals()
                or
                'g_phase_w' not in globals()
            ):
                raise RuntimeError(
                    "Prima di eseguire l'Attività 7.1 devi eseguire "
                    "l'Attività 5.1, nella quale vengono definiti ℓ e g."
                )

            lval = float(l_phase_w.value)
            gval = float(g_phase_w.value)

            n_points = 600


            # ========================================================
            # DEFINIZIONE DELLA PERTURBAZIONE
            #
            #           theta(0) = theta_e + epsilon
            #
            # equilibrio inferiore:
            #           theta_e = 0
            #
            # equilibrio superiore:
            #           theta_e = pi
            # ========================================================

            theta0_inf = eps

            theta0_sup = np.pi + eps

            omega0 = 0.0


            # ========================================================
            # CASO epsilon = 0
            #
            # Impostiamo esattamente i due equilibri.
            #
            # In particolare evitiamo che il fatto numerico
            #
            #        sin(pi) != 0
            #
            # a precisione macchina produca una falsa perturbazione
            # dell'equilibrio superiore.
            # ========================================================

            if abs(eps_deg) < 1e-14:

                tt = np.linspace(
                    0.0,
                    tmax,
                    n_points
                )

                th_inf = np.zeros(
                    n_points
                )

                om_inf = np.zeros(
                    n_points
                )

                th_sup = np.full(
                    n_points,
                    np.pi
                )

                om_sup = np.zeros(
                    n_points
                )


            else:

                # ----------------------------------------------------
                # EQUILIBRIO INFERIORE
                # ----------------------------------------------------

                tt, th_inf, om_inf = _integra_pendolo(
                    theta0_inf,
                    omega0,
                    lval,
                    gval,
                    tmax,
                    n=n_points
                )


                # ----------------------------------------------------
                # EQUILIBRIO SUPERIORE
                # ----------------------------------------------------

                _, th_sup, om_sup = _integra_pendolo(
                    theta0_sup,
                    omega0,
                    lval,
                    gval,
                    tmax,
                    n=n_points
                )


            # ========================================================
            # DATI PER IL BROWSER
            # ========================================================

            dati = {

                "t":
                    np.asarray(tt).tolist(),

                "tmax":
                    tmax,

                "epsDeg":
                    eps_deg,


                "lower": {

                    "theta":
                        np.asarray(th_inf).tolist(),

                    "omega":
                        np.asarray(om_inf).tolist(),

                    "theta0":
                        float(theta0_inf),

                    "thetaEq":
                        0.0,

                    "lim":
                        _limiti_fase_inferiore(
                            th_inf,
                            om_inf
                        ),

                    "upperPhase":
                        False
                },


                "upper": {

                    "theta":
                        np.asarray(th_sup).tolist(),

                    "omega":
                        np.asarray(om_sup).tolist(),

                    "theta0":
                        float(theta0_sup),

                    "thetaEq":
                        float(np.pi),

                    "lim":
                        _limiti_fase_superiore(
                            th_sup,
                            om_sup
                        ),

                    "upperPhase":
                        True
                }
            }


            uid = (
                "eq61_"
                +
                uuid.uuid4().hex[:10]
            )


            # ========================================================
            # HTML
            # ========================================================

            html = f"""
            <div id="{uid}_root"
                 style="
                 font-family:-apple-system,BlinkMacSystemFont,
                 'Segoe UI',Roboto,Arial,sans-serif;
                 max-width:1150px;
                 margin:8px auto;
                 ">


              <div style="
                   margin-bottom:10px;
                   font-size:13px;
                   line-height:1.5;
                   ">

                                Perturbazione utilizzata:

                <b>
                    ε = {eps_deg:.3f}°
                </b>

                &nbsp;&nbsp;&nbsp;

                con

                <b>
                    θ(0) = θ<sub>e</sub> + ε
                </b>

                <br><br>

                <div style="
                     display:inline-block;
                     background:#e8f1ff;
                     color:#102a43;
                     border-left:4px solid #3b82f6;
                     padding:6px 10px;
                     border-radius:4px;
                     ">

                    Stesso pendolo definito nell'Attività 5.1:

                    &nbsp;&nbsp;

                    <b>
                        ℓ = {lval:.3g} m
                    </b>

                    &nbsp;&nbsp;

                    <b>
                        g = {gval:.3g} m/s²
                    </b>

                </div>

              </div>


              <!-- CONTROLLI ANIMAZIONE -->

              <div style="
                   display:flex;
                   gap:7px;
                   align-items:center;
                   flex-wrap:wrap;
                   margin-bottom:8px;
                   ">

                <button id="{uid}_reset">
                    ↺ Ricomincia
                </button>

                <button id="{uid}_play">
                    ▶ Avvia
                </button>

                <button id="{uid}_pause">
                    ⏸ Pausa
                </button>

                <button id="{uid}_slow">
                    − Rallenta
                </button>

                <button id="{uid}_fast">
                    + Accelera
                </button>


                <span style="margin-left:5px;">
                    velocità:
                </span>

                <b id="{uid}_speed">
                    1×
                </b>


                <span style="margin-left:14px;">
                    tempo:
                </span>

                <b id="{uid}_time"
                   style="
                   min-width:90px;
                   font-variant-numeric:tabular-nums;
                   ">
                    t = 0.00 s
                </b>

              </div>


              <!-- TIMELINE -->

              <input
                  id="{uid}_slider"
                  type="range"
                  min="0"
                  max="{n_points - 1}"
                  value="0"
                  step="1"
                  style="
                  width:100%;
                  margin-bottom:14px;
                  "
              >


              <!-- ================================================= -->
              <!-- EQUILIBRIO INFERIORE                             -->
              <!-- ================================================= -->

              <div style="
                   font-weight:650;
                   margin:5px 0 5px 0;
                   ">
                  Equilibrio inferiore
              </div>


              <div style="
                   display:grid;
                   grid-template-columns:
                   repeat(auto-fit,minmax(320px,1fr));
                   gap:10px;
                   ">


                <canvas
                    id="{uid}_pL"
                    style="
                    width:100%;
                    height:300px;
                    border:1px solid #e5e7eb;
                    border-radius:6px;
                    background:white;
                    ">
                </canvas>


                <canvas
                    id="{uid}_fL"
                    style="
                    width:100%;
                    height:300px;
                    border:1px solid #e5e7eb;
                    border-radius:6px;
                    background:white;
                    ">
                </canvas>

              </div>


              <!-- ================================================= -->
              <!-- EQUILIBRIO SUPERIORE                             -->
              <!-- ================================================= -->

              <div style="
                   font-weight:650;
                   margin:20px 0 5px 0;
                   ">
                  Equilibrio superiore
              </div>


              <div style="
                   display:grid;
                   grid-template-columns:
                   repeat(auto-fit,minmax(320px,1fr));
                   gap:10px;
                   ">


                <canvas
                    id="{uid}_pU"
                    style="
                    width:100%;
                    height:300px;
                    border:1px solid #e5e7eb;
                    border-radius:6px;
                    background:white;
                    ">
                </canvas>


                <canvas
                    id="{uid}_fU"
                    style="
                    width:100%;
                    height:300px;
                    border:1px solid #e5e7eb;
                    border-radius:6px;
                    background:white;
                    ">
                </canvas>

              </div>

            </div>
            """


            display(
                HTML(html)
            )


            # ========================================================
            # JAVASCRIPT
            # ========================================================

            js = r"""
            (() => {

              const D = __DATA__;
              const uid = __UID__;

              const byId =
                  s =>
                  document.getElementById(
                      uid + s
                  );


              const pL = byId('_pL');
              const fL = byId('_fL');

              const pU = byId('_pU');
              const fU = byId('_fU');

              const slider =
                  byId('_slider');

              const timeLabel =
                  byId('_time');

              const speedLabel =
                  byId('_speed');


              let k = 0;

              let playIndex = 0;

              let timer = null;

              let running = false;


              const speeds = [
                  0.25,
                  0.5,
                  1,
                  2,
                  4,
                  8
              ];


              let speedIndex = 2;

              let speed =
                  speeds[speedIndex];


              const FONT =
                  '-apple-system,BlinkMacSystemFont,' +
                  '"Segoe UI",Roboto,Arial,sans-serif';


              // ====================================================
              // FORMATO NUMERICO
              // ====================================================

              function fmt(x,n=2) {

                  if(
                      !Number.isFinite(x)
                  )
                      return '—';


                  let s =
                      x.toFixed(n);


                  s =
                      s
                      .replace(
                          /[.]00$/,
                          ''
                      )
                      .replace(
                          /([.][0-9])0$/,
                          '$1'
                      );


                  if(
                      s === '-0'
                  )
                      s = '0';


                  return s;
              }


              // ====================================================
              // PREPARAZIONE CANVAS
              // ====================================================

              function prepare(canvas) {

                  const rect =
                      canvas.getBoundingClientRect();


                  const W =
                      Math.max(
                          280,
                          rect.width
                      );


                  const H =
                      Math.max(
                          250,
                          rect.height
                      );


                  const dpr =
                      Math.min(
                          window.devicePixelRatio || 1,
                          2.2
                      );


                  const pw =
                      Math.round(
                          W*dpr
                      );


                  const ph =
                      Math.round(
                          H*dpr
                      );


                  if(
                      canvas.width !== pw
                      ||
                      canvas.height !== ph
                  ) {

                      canvas.width =
                          pw;

                      canvas.height =
                          ph;
                  }


                  const ctx =
                      canvas.getContext(
                          '2d'
                      );


                  ctx.setTransform(
                      dpr,0,0,dpr,0,0
                  );


                  ctx.clearRect(
                      0,0,W,H
                  );


                  ctx.fillStyle =
                      '#ffffff';


                  ctx.fillRect(
                      0,0,W,H
                  );


                  ctx.lineCap =
                      'round';


                  ctx.lineJoin =
                      'round';


                  return {
                      ctx,
                      W,
                      H
                  };
              }


              function mapX(
                  x,
                  xmin,
                  xmax,
                  W,
                  L,
                  R
              ) {

                  return (
                      L
                      +
                      (x-xmin)
                      /
                      (xmax-xmin)
                      *
                      (W-L-R)
                  );
              }


              function mapY(
                  y,
                  ymin,
                  ymax,
                  H,
                  T,
                  B
              ) {

                  return (
                      H-B
                      -
                      (y-ymin)
                      /
                      (ymax-ymin)
                      *
                      (H-T-B)
                  );
              }


              // ====================================================
              // PENDOLO REALE
              // ====================================================

              function drawPendulum(
                  canvas,
                  data,
                  title,
                  i
              ) {

                  const P =
                      prepare(canvas);


                  const ctx =
                      P.ctx;

                  const W =
                      P.W;

                  const H =
                      P.H;


                  const th =
                      data.theta[i];


                  const om =
                      data.omega[i];


                  ctx.fillStyle =
                      '#111827';


                  ctx.font =
                      '600 14px '
                      +
                      FONT;


                  ctx.textAlign =
                      'center';


                  ctx.fillText(
                      title,
                      W/2,
                      21
                  );


                  const cx =
                      W/2;


                  const cy =
                      H*0.46;


                  const R =
                      Math.min(
                          W*0.28,
                          H*0.34
                      );


                  function point(thv) {

                      return {

                          x:
                              cx
                              +
                              R*Math.sin(thv),

                          y:
                              cy
                              +
                              R*Math.cos(thv)

                      };
                  }


                  const peq =
                      point(
                          data.thetaEq
                      );


                  const p0 =
                      point(
                          data.theta0
                      );


                  const p =
                      point(th);


                  // ----------------------------------------------
                  // circonferenza del vincolo
                  // ----------------------------------------------

                  ctx.save();

                  ctx.setLineDash(
                      [3,4]
                  );


                  ctx.strokeStyle =
                      '#d1d5db';


                  ctx.lineWidth =
                      1;


                  ctx.beginPath();


                  ctx.arc(
                      cx,
                      cy,
                      R,
                      0,
                      2*Math.PI
                  );


                  ctx.stroke();

                  ctx.restore();


                  // ----------------------------------------------
                  // posizione di equilibrio
                  // ----------------------------------------------

                  ctx.save();

                  ctx.setLineDash(
                      [6,4]
                  );


                  ctx.strokeStyle =
                      '#9ca3af';


                  ctx.lineWidth =
                      1.2;


                  ctx.beginPath();


                  ctx.moveTo(
                      cx,
                      cy
                  );


                  ctx.lineTo(
                      peq.x,
                      peq.y
                  );


                  ctx.stroke();

                  ctx.restore();


                  // ----------------------------------------------
                  // posizione iniziale perturbata
                  // ----------------------------------------------

                  ctx.save();

                  ctx.setLineDash(
                      [2,4]
                  );


                  ctx.strokeStyle =
                      '#94a3b8';


                  ctx.lineWidth =
                      1;


                  ctx.beginPath();


                  ctx.moveTo(
                      cx,
                      cy
                  );


                  ctx.lineTo(
                      p0.x,
                      p0.y
                  );


                  ctx.stroke();

                  ctx.restore();


                  // ----------------------------------------------
                  // posizione corrente
                  // ----------------------------------------------

                  ctx.strokeStyle =
                      '#263238';


                  ctx.lineWidth =
                      2.2;


                  ctx.beginPath();


                  ctx.moveTo(
                      cx,
                      cy
                  );


                  ctx.lineTo(
                      p.x,
                      p.y
                  );


                  ctx.stroke();


                  // perno

                  ctx.fillStyle =
                      '#111827';


                  ctx.beginPath();


                  ctx.arc(
                      cx,
                      cy,
                      3.5,
                      0,
                      2*Math.PI
                  );


                  ctx.fill();


                  // massa

                  ctx.fillStyle =
                      '#c62828';


                  ctx.beginPath();


                  ctx.arc(
                      p.x,
                      p.y,
                      8,
                      0,
                      2*Math.PI
                  );


                  ctx.fill();


                  // ----------------------------------------------
                  // informazioni numeriche
                  // ----------------------------------------------

                  ctx.font =
                      '12px '
                      +
                      FONT;


                  ctx.textAlign =
                      'left';


                  ctx.fillStyle =
                      '#111827';


                  ctx.fillText(
                      'θ = '
                      +
                      fmt(
                          th,
                          3
                      )
                      +
                      ' rad',
                      12,
                      H-36
                  );


                  ctx.fillText(
                      'θ̇ = '
                      +
                      fmt(
                          om,
                          3
                      )
                      +
                      ' rad/s',
                      12,
                      H-18
                  );


                  ctx.textAlign =
                      'right';


                  ctx.fillStyle =
                      '#4b5563';


                  ctx.fillText(
                      'ε = '
                      +
                      fmt(
                          D.epsDeg,
                          3
                      )
                      +
                      '°',
                      W-12,
                      H-18
                  );
              }


              // ====================================================
              // DIAGRAMMA DELLO SPAZIO DEGLI STATI
              // ====================================================

              function drawPhase(
                  canvas,
                  data,
                  title,
                  i
              ) {

                  const P =
                      prepare(canvas);


                  const ctx =
                      P.ctx;

                  const W =
                      P.W;

                  const H =
                      P.H;


                  const xmin =
                      data.lim[0];

                  const xmax =
                      data.lim[1];

                  const ymin =
                      data.lim[2];

                  const ymax =
                      data.lim[3];


                  const L = 65;
                  const R = 18;
                  const T = 38;
                  const B = 46;


                  // titolo

                  ctx.fillStyle =
                      '#111827';


                  ctx.font =
                      '600 14px '
                      +
                      FONT;


                  ctx.textAlign =
                      'center';


                  ctx.fillText(
                      title,
                      W/2,
                      21
                  );


                  ctx.font =
                      '11px '
                      +
                      FONT;


                  // =================================================
                  // ASSE THETA
                  // =================================================

                  if(
                      data.upperPhase
                  ) {

                      // ---------------------------------------------
                      // EQUILIBRIO SUPERIORE:
                      //
                      //        -pi   0   pi   2pi   3pi
                      //
                      // ---------------------------------------------

                      const xticks = [
                          -Math.PI,
                          0,
                          Math.PI,
                          2*Math.PI,
                          3*Math.PI
                      ];


                      const labels = [
                          '−π',
                          '0',
                          'π',
                          '2π',
                          '3π'
                      ];


                      for(
                          let j=0;
                          j<xticks.length;
                          j++
                      ) {

                          const xx =
                              mapX(
                                  xticks[j],
                                  xmin,
                                  xmax,
                                  W,
                                  L,
                                  R
                              );


                          // Le tre posizioni superiori equivalenti
                          // (-pi, pi, 3pi) sono evidenziate.
                          const unstable =
                              (
                                  j===0
                                  ||
                                  j===2
                                  ||
                                  j===4
                              );


                          ctx.strokeStyle =
                              unstable
                              ?
                              '#cbd5e1'
                              :
                              '#e5e7eb';


                          ctx.lineWidth =
                              unstable
                              ?
                              1.2
                              :
                              1;


                          if(
                              unstable
                          ) {

                              ctx.save();

                              ctx.setLineDash(
                                  [5,4]
                              );
                          }


                          ctx.beginPath();


                          ctx.moveTo(
                              xx,
                              T
                          );


                          ctx.lineTo(
                              xx,
                              H-B
                          );


                          ctx.stroke();


                          if(
                              unstable
                          )
                              ctx.restore();


                          ctx.fillStyle =
                              '#374151';


                          ctx.textAlign =
                              'center';


                          ctx.fillText(
                              labels[j],
                              xx,
                              H-B+16
                          );
                      }

                  }

                  else {

                      // ---------------------------------------------
                      // EQUILIBRIO INFERIORE:
                      // scala adattiva normale
                      // ---------------------------------------------

                      for(
                          let j=0;
                          j<=4;
                          j++
                      ) {

                          const xx =
                              L
                              +
                              j
                              *
                              (W-L-R)
                              /4;


                          ctx.strokeStyle =
                              '#e5e7eb';


                          ctx.lineWidth =
                              1;


                          ctx.beginPath();


                          ctx.moveTo(
                              xx,
                              T
                          );


                          ctx.lineTo(
                              xx,
                              H-B
                          );


                          ctx.stroke();


                          const xv =
                              xmin
                              +
                              j
                              *
                              (xmax-xmin)
                              /4;


                          ctx.fillStyle =
                              '#374151';


                          ctx.textAlign =
                              'center';


                          ctx.fillText(
                              fmt(
                                  xv,
                                  2
                              ),
                              xx,
                              H-B+16
                          );
                      }
                  }


                  // =================================================
                  // ASSE OMEGA
                  // =================================================

                  for(
                      let j=0;
                      j<=4;
                      j++
                  ) {

                      const yy =
                          T
                          +
                          j
                          *
                          (H-T-B)
                          /4;


                      ctx.strokeStyle =
                          '#e5e7eb';


                      ctx.lineWidth =
                          1;


                      ctx.beginPath();


                      ctx.moveTo(
                          L,
                          yy
                      );


                      ctx.lineTo(
                          W-R,
                          yy
                      );


                      ctx.stroke();


                      const yv =
                          ymax
                          -
                          j
                          *
                          (ymax-ymin)
                          /4;


                      ctx.fillStyle =
                          '#374151';


                      ctx.textAlign =
                          'right';


                      ctx.fillText(
                          fmt(
                              yv,
                              2
                          ),
                          L-7,
                          yy+4
                      );
                  }


                  // bordo grafico

                  ctx.strokeStyle =
                      '#374151';


                  ctx.lineWidth =
                      1;


                  ctx.strokeRect(
                      L,
                      T,
                      W-L-R,
                      H-T-B
                  );


                  // =================================================
                  // NOMI DEGLI ASSI
                  // =================================================

                  ctx.fillStyle =
                      '#111827';


                  ctx.font =
                      '12px '
                      +
                      FONT;


                  ctx.textAlign =
                      'center';


                  ctx.fillText(
                      'θ [rad]',
                      (L+W-R)/2,
                      H-7
                  );


                  ctx.save();


                  ctx.translate(
                      15,
                      (T+H-B)/2
                  );


                  ctx.rotate(
                      -Math.PI/2
                  );


                  ctx.fillText(
                      'θ̇ [rad/s]',
                      0,
                      0
                  );


                  ctx.restore();


                  // =================================================
                  // PUNTI DI EQUILIBRIO
                  // =================================================

                  if(
                      data.upperPhase
                  ) {

                      // copie equivalenti dell'equilibrio superiore

                      const equilibria = [
                          -Math.PI,
                           Math.PI,
                           3*Math.PI
                      ];


                      for(
                          const eq of equilibria
                      ) {

                          const ex =
                              mapX(
                                  eq,
                                  xmin,
                                  xmax,
                                  W,
                                  L,
                                  R
                              );


                          const ey =
                              mapY(
                                  0,
                                  ymin,
                                  ymax,
                                  H,
                                  T,
                                  B
                              );


                          ctx.fillStyle =
                              '#111827';


                          ctx.beginPath();


                          ctx.arc(
                              ex,
                              ey,
                              4.2,
                              0,
                              2*Math.PI
                          );


                          ctx.fill();
                      }

                  }

                  else {

                      // equilibrio inferiore centrale

                      const ex =
                          mapX(
                              0,
                              xmin,
                              xmax,
                              W,
                              L,
                              R
                          );


                      const ey =
                          mapY(
                              0,
                              ymin,
                              ymax,
                              H,
                              T,
                              B
                          );


                      ctx.fillStyle =
                          '#111827';


                      ctx.beginPath();


                      ctx.arc(
                          ex,
                          ey,
                          4.2,
                          0,
                          2*Math.PI
                      );


                      ctx.fill();
                  }


                  // =================================================
                  // STATO INIZIALE
                  // =================================================

                  const ix =
                      mapX(
                          data.theta0,
                          xmin,
                          xmax,
                          W,
                          L,
                          R
                      );


                  const iy =
                      mapY(
                          0,
                          ymin,
                          ymax,
                          H,
                          T,
                          B
                      );


                  ctx.strokeStyle =
                      '#64748b';


                  ctx.lineWidth =
                      1.3;


                  ctx.beginPath();


                  ctx.arc(
                      ix,
                      iy,
                      5.5,
                      0,
                      2*Math.PI
                  );


                  ctx.stroke();


                  // =================================================
                  // TRAIETTORIA GIÀ PERCORSA
                  // =================================================

                  ctx.strokeStyle =
                      '#1565c0';


                  ctx.lineWidth =
                      1.9;


                  ctx.beginPath();


                  let started =
                      false;


                  for(
                      let j=0;
                      j<=i;
                      j++
                  ) {

                      const thj =
                          data.theta[j];


                      const omj =
                          data.omega[j];


                      if(
                          thj < xmin
                          ||
                          thj > xmax
                      ) {

                          started =
                              false;

                          continue;
                      }


                      const xx =
                          mapX(
                              thj,
                              xmin,
                              xmax,
                              W,
                              L,
                              R
                          );


                      const yy =
                          mapY(
                              omj,
                              ymin,
                              ymax,
                              H,
                              T,
                              B
                          );


                      if(
                          !started
                      ) {

                          ctx.moveTo(
                              xx,
                              yy
                          );


                          started =
                              true;

                      }

                      else {

                          ctx.lineTo(
                              xx,
                              yy
                          );
                      }
                  }


                  ctx.stroke();


                  // =================================================
                  // PUNTO CORRENTE z(t)
                  // =================================================

                  const thetaNow =
                      data.theta[i];


                  const omegaNow =
                      data.omega[i];


                  if(
                      thetaNow >= xmin
                      &&
                      thetaNow <= xmax
                  ) {

                      const px =
                          mapX(
                              thetaNow,
                              xmin,
                              xmax,
                              W,
                              L,
                              R
                          );


                      const py =
                          mapY(
                              omegaNow,
                              ymin,
                              ymax,
                              H,
                              T,
                              B
                          );


                      ctx.fillStyle =
                          '#c62828';


                      ctx.beginPath();


                      ctx.arc(
                          px,
                          py,
                          5.5,
                          0,
                          2*Math.PI
                      );


                      ctx.fill();
                  }


                  // valore numerico dello stato

                  ctx.fillStyle =
                      '#111827';


                  ctx.font =
                      '11px '
                      +
                      FONT;


                  ctx.textAlign =
                      'left';


                  ctx.fillText(
                      'z(t) = ('
                      +
                      fmt(
                          thetaNow,
                          2
                      )
                      +
                      ', '
                      +
                      fmt(
                          omegaNow,
                          2
                      )
                      +
                      ')',
                      L+8,
                      T+16
                  );
              }


              // ====================================================
              // DISEGNO SINCRONO DEI QUATTRO PANNELLI
              // ====================================================

              function draw() {

                  drawPendulum(
                      pL,
                      D.lower,
                      'Moto reale — equilibrio inferiore',
                      k
                  );


                  drawPhase(
                      fL,
                      D.lower,
                      'Spazio degli stati — equilibrio inferiore',
                      k
                  );


                  drawPendulum(
                      pU,
                      D.upper,
                      'Moto reale — equilibrio superiore',
                      k
                  );


                  drawPhase(
                      fU,
                      D.upper,
                      'Spazio degli stati — equilibrio superiore',
                      k
                  );


                  slider.value =
                      k;


                  timeLabel.textContent =
                      't = '
                      +
                      fmt(
                          D.t[k],
                          2
                      )
                      +
                      ' s';


                  speedLabel.textContent =
                      speed
                      +
                      '×';
              }


              // ====================================================
              // ANIMAZIONE
              // ====================================================

              const baseInterval =
                  1000
                  *
                  D.tmax
                  /
                  (D.t.length-1);


              const refresh =
                  16;


              function stop() {

                  running =
                      false;


                  if(
                      timer !== null
                  ) {

                      clearTimeout(
                          timer
                      );


                      timer =
                          null;
                  }
              }


              function tick() {

                  if(
                      !running
                  )
                      return;


                  playIndex +=
                      speed
                      *
                      refresh
                      /
                      baseInterval;


                  if(
                      playIndex
                      >=
                      D.t.length-1
                  ) {

                      playIndex =
                          D.t.length-1;


                      k =
                          D.t.length-1;


                      draw();


                      stop();


                      return;
                  }


                  k =
                      Math.floor(
                          playIndex
                      );


                  draw();


                  timer =
                      setTimeout(
                          tick,
                          refresh
                      );
              }


              function play() {

                  if(
                      k
                      >=
                      D.t.length-1
                  ) {

                      k = 0;

                      playIndex = 0;
                  }


                  if(
                      running
                  )
                      return;


                  running =
                      true;


                  tick();
              }


              // ====================================================
              // COMANDI
              // ====================================================

              byId('_play').onclick =
                  () => {

                      play();

                  };


              byId('_pause').onclick =
                  () => {

                      stop();

                  };


              byId('_reset').onclick =
                  () => {

                      stop();


                      k = 0;


                      playIndex = 0;


                      draw();

                  };


              byId('_slow').onclick =
                  () => {

                      speedIndex =
                          Math.max(
                              0,
                              speedIndex-1
                          );


                      speed =
                          speeds[
                              speedIndex
                          ];


                      draw();

                  };


              byId('_fast').onclick =
                  () => {

                      speedIndex =
                          Math.min(
                              speeds.length-1,
                              speedIndex+1
                          );


                      speed =
                          speeds[
                              speedIndex
                          ];


                      draw();

                  };


              slider.oninput =
                  e => {

                      stop();


                      k =
                          parseInt(
                              e.target.value
                          );


                      playIndex =
                          k;


                      draw();

                  };


              window.addEventListener(
                  'resize',
                  draw,
                  {
                      passive:true
                  }
              );


              // ====================================================
              // PARTENZA
              // ====================================================

              draw();

              play();

              return 'ok';

            })()
            """


            js = (
                js
                .replace(
                    "__DATA__",
                    json.dumps(
                        dati,
                        separators=(",", ":")
                    )
                )
                .replace(
                    "__UID__",
                    json.dumps(uid)
                )
            )


            # ========================================================
            # ESECUZIONE JAVASCRIPT IN GOOGLE COLAB
            # ========================================================

            from google.colab import output as colab_output

            colab_output.eval_js(
                js
            )


        except Exception as e:

            print(
                "Impossibile eseguire l’attività:"
            )

            print(
                type(e).__name__
                +
                ":",
                e
            )


pert_btn.on_click(
    _perturba_sincrona
)
````

````python
#@title Simulazione Attività 8 — Esplora la soglia e segui il moto { display-mode: "form" }

import json
import uuid
import numpy as np
import ipywidgets as widgets

from scipy.integrate import solve_ivp
from IPython.display import display, HTML, clear_output
from google.colab import output as colab_output


# ============================================================
# PENDOLO DI RIFERIMENTO DELL'ATTIVITÀ 8
# ============================================================

L_ATT8 = 1.0       # m
G_ATT8 = 9.81      # m/s^2


# ============================================================
# CONTROLLI DELLA PROVA
# ============================================================

crit_om_w = widgets.FloatSlider(
    value=5.8,
    min=0.0,
    max=8.0,
    step=0.02,
    description='theta_punto(0) [rad/s]:',
    style={'description_width': '175px'},
    continuous_update=False,
    readout=True,
    readout_format='.2f',
    layout=widgets.Layout(width='560px')
)

crit_om_manual_w = widgets.BoundedFloatText(
    value=5.8,
    min=0.0,
    max=8.0,
    step=0.001,
    description='da tastiera:',
    style={'description_width': '95px'},
    layout=widgets.Layout(width='210px')
)

crit_om_link = widgets.jslink(
    (crit_om_w, 'value'),
    (crit_om_manual_w, 'value')
)


crit_t_w = widgets.FloatSlider(
    value=20.0,
    min=5.0,
    max=35.0,
    step=1.0,
    description='t simulato [s]:',
    continuous_update=False,
    readout=True,
    readout_format='.1f',
    layout=widgets.Layout(width='430px')
)

crit_t_manual_w = widgets.BoundedFloatText(
    value=20.0,
    min=5.0,
    max=35.0,
    step=0.1,
    description='da tastiera:',
    style={'description_width': '95px'},
    layout=widgets.Layout(width='210px')
)

crit_t_link = widgets.jslink(
    (crit_t_w, 'value'),
    (crit_t_manual_w, 'value')
)


crit_btn = widgets.Button(
    description='Esegui questa prova',
    button_style='info'
)

out_crit = widgets.Output()


display(widgets.VBox([
    widgets.HBox([
        crit_om_w,
        crit_om_manual_w
    ]),
    widgets.HBox([
        crit_t_w,
        crit_t_manual_w
    ]),
    crit_btn,
    out_crit
]))


# ============================================================
# INTEGRAZIONE LOCALE
#
# Questa cella è indipendente dalle attività precedenti.
# ============================================================

def _integra_att8(omega0, tmax):

    def rhs(t, z):

        th, om = z

        return [
            om,
            -(G_ATT8 / L_ATT8) * np.sin(th)
        ]

    sol = solve_ivp(
        rhs,
        (0.0, tmax),
        [0.0, omega0],
        method='DOP853',
        dense_output=True,
        rtol=1e-10,
        atol=1e-12
    )

    if not sol.success:
        raise RuntimeError(sol.message)

    n = max(
        1400,
        int(np.ceil(tmax * 120))
    )

    tt = np.linspace(
        0.0,
        tmax,
        n
    )

    th, om = sol.sol(tt)

    return tt, th, om


# UID dell'ultima animazione
_att8_uid = None


# ============================================================
# ESECUZIONE DELLA PROVA
# ============================================================

def _prova_critica(_):

    global _att8_uid


    # --------------------------------------------------------
    # FERMIAMO UN'EVENTUALE ANIMAZIONE PRECEDENTE
    # --------------------------------------------------------

    if _att8_uid is not None:

        try:

            colab_output.eval_js(
                f"if (window['{_att8_uid}_timer']) {{"
                f"clearInterval(window['{_att8_uid}_timer']);"
                f"window['{_att8_uid}_timer']=null;"
                f"}}"
            )

        except Exception:
            pass


    with out_crit:

        clear_output(wait=True)

        try:

            om0 = float(
                crit_om_w.value
            )

            tmax = float(
                crit_t_w.value
            )


            # ------------------------------------------------
            # INTEGRAZIONE
            # ------------------------------------------------

            tt, th, om = _integra_att8(
                om0,
                tmax
            )


            # ------------------------------------------------
            # CONTROLLO DEL SUPERAMENTO DI ±pi
            # ------------------------------------------------

            supera = bool(
                np.any(th > np.pi)
                or
                np.any(th < -np.pi)
            )


            if supera:

                messaggio = (
                    "Nell'intervallo simulato la traiettoria "
                    "supera θ = ±π."
                )

            else:

                distanza = float(
                    np.min(
                        np.abs(
                            np.abs(th) - np.pi
                        )
                    )
                )

                messaggio = (
                    "Nell'intervallo simulato non è stato osservato "
                    "il superamento di θ = ±π. "
                    "Minima distanza dalla configurazione superiore: "
                    f"{distanza:.4f} rad."
                )


            # ------------------------------------------------
            # INFORMAZIONI SULLA PROVA
            # ------------------------------------------------

            display(HTML(
                f"""
                <div style="
                    background:#f3f4f6;
                    color:#111827;
                    border-left:5px solid #6b7280;
                    padding:10px 14px;
                    border-radius:6px;
                    margin:6px 0 12px 0;">

                    <b>Esito della prova:</b>
                    {messaggio}

                    <br><br>

                    <span style="font-size:13px;">

                        Pendolo:

                        ℓ = {L_ATT8:.2f} m,

                        g = {G_ATT8:.2f} m/s²,

                        θ(0) = 0,

                        θ̇(0) = {om0:.3f} rad/s.

                    </span>

                </div>
                """
            ))


            # ------------------------------------------------
            # DATI PER L'ANIMAZIONE NEL BROWSER
            # ------------------------------------------------

            uid = (
                "att8_"
                + uuid.uuid4().hex[:10]
            )

            _att8_uid = uid


            dati = {

                't':
                    np.round(
                        tt,
                        8
                    ).tolist(),

                'theta':
                    np.round(
                        th,
                        8
                    ).tolist(),

                'omega':
                    np.round(
                        om,
                        8
                    ).tolist(),

                'ell':
                    L_ATT8,

                'g':
                    G_ATT8
            }


            # ------------------------------------------------
            # DASHBOARD
            # ------------------------------------------------

            html = f"""
            <div id="{uid}_box"
                 style="
                 margin:8px 0 14px 0;
                 padding:12px;
                 border:1px solid #d1d5db;
                 border-radius:9px;
                 background:#ffffff;
                 color:#111827;
                 overflow-x:auto;">


                <!-- =========================================
                     CONTROLLI DELL'ANIMAZIONE
                     ========================================= -->

                <div style="
                    display:flex;
                    align-items:center;
                    gap:8px;
                    flex-wrap:wrap;
                    margin-bottom:8px;">

                    <button id="{uid}_play">
                        ▶ Avvia
                    </button>

                    <button id="{uid}_pause">
                        ⏸ Pausa
                    </button>

                    <button id="{uid}_reset">
                        ↺ t = 0
                    </button>

                    <label style="margin-left:8px;">

                        velocità:

                        <select id="{uid}_speed">

                            <option value="0.5">
                                0.5×
                            </option>

                            <option value="1" selected>
                                1×
                            </option>

                            <option value="2">
                                2×
                            </option>

                            <option value="4">
                                4×
                            </option>

                        </select>

                    </label>

                </div>


                <!-- =========================================
                     CURSORE ROSSO: TEMPO DELL'ANIMAZIONE
                     ========================================= -->

                <div style="
                    display:flex;
                    align-items:center;
                    gap:10px;
                    margin:8px 0 12px 0;
                    min-width:820px;">

                    <b style="color:#b91c1c;">
                        Stato dell'animazione:
                    </b>

                    <input
                        id="{uid}_slider"
                        type="range"
                        min="0"
                        max="{len(tt)-1}"
                        value="0"
                        step="1"
                        style="
                        flex:1;
                        min-width:420px;
                        accent-color:#dc2626;">


                    <span
                        id="{uid}_readout"
                        style="
                        min-width:310px;
                        font-family:monospace;">
                    </span>

                </div>


                <!-- =========================================
                     CORPO PRINCIPALE
                     SINISTRA: GRAFICI
                     DESTRA: PENDOLO
                     ========================================= -->

                <div style="
                    min-width:820px;
                    display:grid;
                    grid-template-columns:
                        minmax(560px, 3fr)
                        minmax(220px, 1fr);
                    gap:10px;
                    align-items:start;">


                    <!-- COLONNA SINISTRA -->

                    <div style="
                        display:grid;
                        grid-template-rows:245px 245px;
                        gap:10px;">

                        <canvas
                            id="{uid}_phase"
                            style="
                            width:100%;
                            height:245px;
                            border:1px solid #e5e7eb;
                            border-radius:6px;
                            background:#fff;">
                        </canvas>


                        <canvas
                            id="{uid}_theta"
                            style="
                            width:100%;
                            height:245px;
                            border:1px solid #e5e7eb;
                            border-radius:6px;
                            background:#fff;">
                        </canvas>

                    </div>


                    <!-- COLONNA DESTRA -->

                    <div>

                        <canvas
                            id="{uid}_pend"
                            style="
                            width:100%;
                            height:300px;
                            border:1px solid #e5e7eb;
                            border-radius:6px;
                            background:#fff;">
                        </canvas>


                        <!-- MODALITÀ DI ESPLORAZIONE MANUALE -->

                        <div
                            id="{uid}_manual_box"
                            style="
                            display:none;
                            margin-top:12px;
                            padding:11px 12px;
                            background:#ecfdf5;
                            color:#14532d;
                            border-left:5px solid #22c55e;
                            border-radius:6px;">

                            <div style="
                                font-weight:700;
                                margin-bottom:7px;">

                                ● Esplorazione manuale

                            </div>

                            <div style="
                                font-size:13px;
                                line-height:1.4;
                                margin-bottom:8px;">

                                Muovi il cursore verde oppure clicca
                                direttamente su un punto della traiettoria
                                nel piano degli stati.

                            </div>


                            <input
                                id="{uid}_manual"
                                type="range"
                                min="0"
                                max="{len(tt)-1}"
                                value="0"
                                step="1"
                                style="
                                width:100%;
                                accent-color:#16a34a;">


                            <div
                                id="{uid}_manual_readout"
                                style="
                                margin-top:8px;
                                font-family:monospace;
                                font-size:12px;">
                            </div>

                        </div>

                    </div>

                </div>


                <div style="
                    font-size:12px;
                    color:#4b5563;
                    margin-top:9px;">

                    <span style="color:#dc2626;font-weight:700;">
                        Rosso
                    </span>
                    = stato seguito durante l'animazione.

                    &nbsp;&nbsp;

                    <span style="color:#16a34a;font-weight:700;">
                        Verde
                    </span>
                    = stato scelto dallo studente durante
                    l'esplorazione manuale.

                    <br>

                    Nel grafico θ(t), θ è rappresentato senza
                    riavvolgimento modulo 2π.

                </div>

            </div>
            """


            display(
                HTML(html)
            )


            # ------------------------------------------------
            # JAVASCRIPT / CANVAS
            # ------------------------------------------------

            data_js = json.dumps(
                dati,
                separators=(',', ':')
            )


            js_template = r"""
            (async function() {

                await new Promise(
                    r => setTimeout(r, 80)
                );


                const uid = "__UID__";

                const D = __DATA__;

                const N = D.t.length;


                const el = (name) =>
                    document.getElementById(
                        uid + "_" + name
                    );


                const slider =
                    el("slider");

                const manualSlider =
                    el("manual");

                const manualBox =
                    el("manual_box");

                const readout =
                    el("readout");

                const manualReadout =
                    el("manual_readout");

                const speedSel =
                    el("speed");


                let redIndex = 0;

                let greenIndex = null;

                let manualMode = false;


                let phaseGeometry = null;

                let thetaGeometry = null;


                // ==========================================
                // PREPARAZIONE CANVAS
                // ==========================================

                function canvasInfo(canvas) {

                    const rect =
                        canvas.getBoundingClientRect();

                    const dpr =
                        window.devicePixelRatio || 1;

                    const w =
                        Math.max(
                            20,
                            rect.width
                        );

                    const h =
                        Math.max(
                            20,
                            rect.height
                        );


                    const W =
                        Math.round(
                            w * dpr
                        );

                    const H =
                        Math.round(
                            h * dpr
                        );


                    if (
                        canvas.width !== W
                        ||
                        canvas.height !== H
                    ) {

                        canvas.width = W;

                        canvas.height = H;

                    }


                    const ctx =
                        canvas.getContext("2d");


                    ctx.setTransform(
                        dpr,
                        0,
                        0,
                        dpr,
                        0,
                        0
                    );


                    return {
                        ctx,
                        w,
                        h
                    };
                }


                // ==========================================
                // LIMITI
                // ==========================================

                function paddedBounds(
                    arr,
                    extras,
                    frac
                ) {

                    const vals =
                        arr
                        .concat(
                            extras || []
                        )
                        .filter(
                            Number.isFinite
                        );


                    let mn =
                        Math.min(...vals);

                    let mx =
                        Math.max(...vals);


                    if (
                        !Number.isFinite(mn)
                        ||
                        !Number.isFinite(mx)
                    ) {

                        mn = -1;

                        mx = 1;

                    }


                    if (
                        Math.abs(mx - mn)
                        <
                        1e-10
                    ) {

                        const a =
                            Math.max(
                                1,
                                Math.abs(mn)
                            );

                        mn -= 0.2 * a;

                        mx += 0.2 * a;

                    }

                    else {

                        const d =
                            mx - mn;

                        mn -= frac * d;

                        mx += frac * d;

                    }


                    return [
                        mn,
                        mx
                    ];
                }


                function fmt(v) {

                    if (
                        Math.abs(v) >= 10
                    ) {

                        return v.toFixed(1);

                    }

                    return v.toFixed(2);
                }


                // ==========================================
                // ASSI CARTESIANI
                // ==========================================

                function axes(
                    ctx,
                    w,
                    h,
                    xmin,
                    xmax,
                    ymin,
                    ymax,
                    xlabel,
                    ylabel,
                    title
                ) {

                    const m = {
                        l:58,
                        r:18,
                        t:36,
                        b:46
                    };


                    const pw =
                        w - m.l - m.r;

                    const ph =
                        h - m.t - m.b;


                    const X =
                        x =>
                        m.l
                        +
                        (x - xmin)
                        /
                        (xmax - xmin)
                        *
                        pw;


                    const Y =
                        y =>
                        m.t
                        +
                        (ymax - y)
                        /
                        (ymax - ymin)
                        *
                        ph;


                    ctx.clearRect(
                        0,
                        0,
                        w,
                        h
                    );


                    ctx.fillStyle =
                        "#ffffff";

                    ctx.fillRect(
                        0,
                        0,
                        w,
                        h
                    );


                    ctx.font =
                        "12px Arial";

                    ctx.textAlign =
                        "center";

                    ctx.textBaseline =
                        "middle";


                    // griglia verticale

                    for (
                        let k = 0;
                        k <= 5;
                        k++
                    ) {

                        const xv =
                            xmin
                            +
                            (xmax - xmin)
                            *
                            k / 5;

                        const xp =
                            X(xv);


                        ctx.strokeStyle =
                            "#e5e7eb";

                        ctx.lineWidth = 1;


                        ctx.beginPath();

                        ctx.moveTo(
                            xp,
                            m.t
                        );

                        ctx.lineTo(
                            xp,
                            h - m.b
                        );

                        ctx.stroke();


                        ctx.fillStyle =
                            "#374151";

                        ctx.fillText(
                            fmt(xv),
                            xp,
                            h - m.b + 17
                        );
                    }


                    // griglia orizzontale

                    ctx.textAlign =
                        "right";


                    for (
                        let k = 0;
                        k <= 5;
                        k++
                    ) {

                        const yv =
                            ymin
                            +
                            (ymax - ymin)
                            *
                            k / 5;

                        const yp =
                            Y(yv);


                        ctx.strokeStyle =
                            "#e5e7eb";


                        ctx.beginPath();

                        ctx.moveTo(
                            m.l,
                            yp
                        );

                        ctx.lineTo(
                            w - m.r,
                            yp
                        );

                        ctx.stroke();


                        ctx.fillStyle =
                            "#374151";

                        ctx.fillText(
                            fmt(yv),
                            m.l - 7,
                            yp
                        );
                    }


                    ctx.strokeStyle =
                        "#374151";

                    ctx.lineWidth =
                        1.2;


                    ctx.strokeRect(
                        m.l,
                        m.t,
                        pw,
                        ph
                    );


                    ctx.fillStyle =
                        "#111827";

                    ctx.font =
                        "bold 15px Arial";

                    ctx.textAlign =
                        "center";


                    ctx.fillText(
                        title,
                        m.l + pw/2,
                        17
                    );


                    ctx.font =
                        "13px Arial";


                    ctx.fillText(
                        xlabel,
                        m.l + pw/2,
                        h - 12
                    );


                    ctx.save();


                    ctx.translate(
                        15,
                        m.t + ph/2
                    );


                    ctx.rotate(
                        -Math.PI/2
                    );


                    ctx.fillText(
                        ylabel,
                        0,
                        0
                    );


                    ctx.restore();


                    return {
                        X,
                        Y,
                        m,
                        pw,
                        ph,
                        xmin,
                        xmax,
                        ymin,
                        ymax
                    };
                }


                // ==========================================
                // CURVA
                // ==========================================

                function polyline(
                    ctx,
                    xs,
                    ys,
                    X,
                    Y,
                    color,
                    width
                ) {

                    ctx.strokeStyle =
                        color;

                    ctx.lineWidth =
                        width;

                    ctx.setLineDash([]);


                    ctx.beginPath();


                    let started =
                        false;


                    for (
                        let i = 0;
                        i < xs.length;
                        i++
                    ) {

                        if (
                            !Number.isFinite(xs[i])
                            ||
                            !Number.isFinite(ys[i])
                        ) {

                            continue;
                        }


                        const xp =
                            X(xs[i]);

                        const yp =
                            Y(ys[i]);


                        if (!started) {

                            ctx.moveTo(
                                xp,
                                yp
                            );

                            started = true;

                        }

                        else {

                            ctx.lineTo(
                                xp,
                                yp
                            );
                        }
                    }


                    ctx.stroke();
                }


                // ==========================================
                // LIMITI GLOBALI
                // ==========================================

                const phaseXB =
                    paddedBounds(
                        D.theta,
                        [
                            -Math.PI,
                            Math.PI,
                            0
                        ],
                        0.06
                    );


                const phaseYB =
                    paddedBounds(
                        D.omega,
                        [0],
                        0.10
                    );


                const thetaYB =
                    paddedBounds(
                        D.theta,
                        [
                            -Math.PI,
                            Math.PI,
                            0
                        ],
                        0.08
                    );


                // ==========================================
                // PUNTO
                // ==========================================

                function point(
                    ctx,
                    x,
                    y,
                    radius,
                    color
                ) {

                    ctx.fillStyle =
                        color;


                    ctx.beginPath();


                    ctx.arc(
                        x,
                        y,
                        radius,
                        0,
                        2*Math.PI
                    );


                    ctx.fill();
                }


                // ==========================================
                // SPAZIO DEGLI STATI
                // ==========================================

                function drawPhase(
                    iRed,
                    iGreen
                ) {

                    const c =
                        canvasInfo(
                            el("phase")
                        );


                    const A =
                        axes(
                            c.ctx,
                            c.w,
                            c.h,
                            phaseXB[0],
                            phaseXB[1],
                            phaseYB[0],
                            phaseYB[1],
                            "θ [rad]",
                            "θ̇ [rad/s]",
                            "Traiettoria nel piano (θ, θ̇)"
                        );


                    phaseGeometry = A;


                    polyline(
                        c.ctx,
                        D.theta,
                        D.omega,
                        A.X,
                        A.Y,
                        "#1f77b4",
                        1.8
                    );


                    // θ = ±π

                    c.ctx.save();


                    c.ctx.setLineDash(
                        [5,4]
                    );


                    c.ctx.strokeStyle =
                        "#60a5fa";

                    c.ctx.lineWidth =
                        1.2;


                    [
                        -Math.PI,
                        Math.PI
                    ].forEach(
                        v => {

                            c.ctx.beginPath();

                            c.ctx.moveTo(
                                A.X(v),
                                A.m.t
                            );

                            c.ctx.lineTo(
                                A.X(v),
                                c.h - A.m.b
                            );

                            c.ctx.stroke();

                        }
                    );


                    // cursore rosso

                    c.ctx.strokeStyle =
                        "#dc2626";

                    c.ctx.lineWidth =
                        1.3;


                    const xr =
                        A.X(
                            D.theta[iRed]
                        );


                    c.ctx.beginPath();

                    c.ctx.moveTo(
                        xr,
                        A.m.t
                    );

                    c.ctx.lineTo(
                        xr,
                        c.h - A.m.b
                    );

                    c.ctx.stroke();


                    c.ctx.restore();


                    point(
                        c.ctx,
                        A.X(
                            D.theta[iRed]
                        ),
                        A.Y(
                            D.omega[iRed]
                        ),
                        5.3,
                        "#dc2626"
                    );


                    // cursore verde manuale

                    if (
                        iGreen !== null
                    ) {

                        c.ctx.save();


                        c.ctx.setLineDash(
                            [5,4]
                        );


                        c.ctx.strokeStyle =
                            "#16a34a";

                        c.ctx.lineWidth =
                            1.6;


                        const xg =
                            A.X(
                                D.theta[iGreen]
                            );


                        c.ctx.beginPath();

                        c.ctx.moveTo(
                            xg,
                            A.m.t
                        );

                        c.ctx.lineTo(
                            xg,
                            c.h - A.m.b
                        );

                        c.ctx.stroke();


                        c.ctx.restore();


                        point(
                            c.ctx,
                            A.X(
                                D.theta[iGreen]
                            ),
                            A.Y(
                                D.omega[iGreen]
                            ),
                            6.0,
                            "#16a34a"
                        );
                    }
                }


                // ==========================================
                // θ(t)
                // ==========================================

                function drawTheta(
                    iRed,
                    iGreen
                ) {

                    const c =
                        canvasInfo(
                            el("theta")
                        );


                    const A =
                        axes(
                            c.ctx,
                            c.w,
                            c.h,
                            D.t[0],
                            D.t[N-1],
                            thetaYB[0],
                            thetaYB[1],
                            "t [s]",
                            "θ [rad]",
                            "Evoluzione θ(t)"
                        );


                    thetaGeometry = A;


                    polyline(
                        c.ctx,
                        D.t,
                        D.theta,
                        A.X,
                        A.Y,
                        "#1f77b4",
                        1.8
                    );


                    // θ = ±π

                    c.ctx.save();


                    c.ctx.setLineDash(
                        [5,4]
                    );


                    c.ctx.strokeStyle =
                        "#60a5fa";

                    c.ctx.lineWidth =
                        1.2;


                    [
                        -Math.PI,
                        Math.PI
                    ].forEach(
                        v => {

                            c.ctx.beginPath();

                            c.ctx.moveTo(
                                A.m.l,
                                A.Y(v)
                            );

                            c.ctx.lineTo(
                                c.w - A.m.r,
                                A.Y(v)
                            );

                            c.ctx.stroke();

                        }
                    );


                    // cursore temporale rosso

                    c.ctx.strokeStyle =
                        "#dc2626";

                    c.ctx.lineWidth =
                        1.3;


                    const xr =
                        A.X(
                            D.t[iRed]
                        );


                    c.ctx.beginPath();

                    c.ctx.moveTo(
                        xr,
                        A.m.t
                    );

                    c.ctx.lineTo(
                        xr,
                        c.h - A.m.b
                    );

                    c.ctx.stroke();


                    c.ctx.restore();


                    point(
                        c.ctx,
                        A.X(
                            D.t[iRed]
                        ),
                        A.Y(
                            D.theta[iRed]
                        ),
                        5.3,
                        "#dc2626"
                    );


                    // esplorazione manuale verde

                    if (
                        iGreen !== null
                    ) {

                        c.ctx.save();


                        c.ctx.setLineDash(
                            [5,4]
                        );


                        c.ctx.strokeStyle =
                            "#16a34a";

                        c.ctx.lineWidth =
                            1.6;


                        const xg =
                            A.X(
                                D.t[iGreen]
                            );


                        c.ctx.beginPath();

                        c.ctx.moveTo(
                            xg,
                            A.m.t
                        );

                        c.ctx.lineTo(
                            xg,
                            c.h - A.m.b
                        );

                        c.ctx.stroke();


                        c.ctx.restore();


                        point(
                            c.ctx,
                            A.X(
                                D.t[iGreen]
                            ),
                            A.Y(
                                D.theta[iGreen]
                            ),
                            6.0,
                            "#16a34a"
                        );
                    }
                }


                // ==========================================
                // PENDOLO
                // ==========================================

                function drawPendulum(i) {

                    const c =
                        canvasInfo(
                            el("pend")
                        );


                    const ctx =
                        c.ctx;

                    const w =
                        c.w;

                    const h =
                        c.h;


                    ctx.clearRect(
                        0,
                        0,
                        w,
                        h
                    );


                    ctx.fillStyle =
                        "#ffffff";

                    ctx.fillRect(
                        0,
                        0,
                        w,
                        h
                    );


                    ctx.fillStyle =
                        "#111827";

                    ctx.font =
                        "bold 15px Arial";

                    ctx.textAlign =
                        "center";


                    ctx.fillText(
                        manualMode
                        ?
                        "Pendolo — esplorazione"
                        :
                        "Pendolo — animazione",
                        w/2,
                        22
                    );


                    // valori istantanei

                    ctx.font =
                        "13px monospace";

                    ctx.textAlign =
                        "left";


                    ctx.fillText(
                        "t  = "
                        +
                        D.t[i].toFixed(3)
                        +
                        " s",
                        14,
                        49
                    );


                    ctx.fillText(
                        "θ  = "
                        +
                        D.theta[i].toFixed(4)
                        +
                        " rad",
                        14,
                        70
                    );


                    ctx.fillText(
                        "θ̇ = "
                        +
                        D.omega[i].toFixed(4)
                        +
                        " rad/s",
                        14,
                        91
                    );


                    const px =
                        w / 2;

                    const py =
                        130;


                    const R =
                        Math.min(
                            w * 0.34,
                            h * 0.38
                        );


                    const th =
                        D.theta[i];


                    const bx =
                        px
                        +
                        R
                        *
                        Math.sin(th);


                    const by =
                        py
                        +
                        R
                        *
                        Math.cos(th);


                    // verticale

                    ctx.save();


                    ctx.setLineDash(
                        [5,5]
                    );


                    ctx.strokeStyle =
                        "#9ca3af";

                    ctx.lineWidth =
                        1;


                    ctx.beginPath();

                    ctx.moveTo(
                        px,
                        py
                    );

                    ctx.lineTo(
                        px,
                        py + R + 18
                    );

                    ctx.stroke();


                    ctx.restore();


                    // asta

                    ctx.strokeStyle =
                        "#374151";

                    ctx.lineWidth =
                        3.5;


                    ctx.beginPath();

                    ctx.moveTo(
                        px,
                        py
                    );

                    ctx.lineTo(
                        bx,
                        by
                    );

                    ctx.stroke();


                    // perno

                    ctx.fillStyle =
                        "#111827";


                    ctx.beginPath();

                    ctx.arc(
                        px,
                        py,
                        4.5,
                        0,
                        2*Math.PI
                    );

                    ctx.fill();


                    // massa

                    ctx.fillStyle =
                        manualMode
                        ?
                        "#16a34a"
                        :
                        "#dc2626";


                    ctx.beginPath();

                    ctx.arc(
                        bx,
                        by,
                        11,
                        0,
                        2*Math.PI
                    );

                    ctx.fill();


                    ctx.fillStyle =
                        "#4b5563";

                    ctx.font =
                        "12px Arial";

                    ctx.textAlign =
                        "center";


                    ctx.fillText(
                        "ℓ = "
                        +
                        D.ell.toFixed(2)
                        +
                        " m     g = "
                        +
                        D.g.toFixed(2)
                        +
                        " m/s²",
                        w/2,
                        h - 15
                    );
                }


                // ==========================================
                // AGGIORNAMENTO COMPLETO
                // ==========================================

                function drawAll() {

                    redIndex =
                        Math.max(
                            0,
                            Math.min(
                                N - 1,
                                Math.round(redIndex)
                            )
                        );


                    if (
                        greenIndex !== null
                    ) {

                        greenIndex =
                            Math.max(
                                0,
                                Math.min(
                                    N - 1,
                                    Math.round(greenIndex)
                                )
                            );
                    }


                    slider.value =
                        redIndex;


                    readout.textContent =
                        "t = "
                        +
                        D.t[redIndex].toFixed(3)
                        +
                        " s   |   θ = "
                        +
                        D.theta[redIndex].toFixed(4)
                        +
                        " rad";


                    drawPhase(
                        redIndex,
                        greenIndex
                    );


                    drawTheta(
                        redIndex,
                        greenIndex
                    );


                    const pendulumIndex =
                        (
                            manualMode
                            &&
                            greenIndex !== null
                        )
                        ?
                        greenIndex
                        :
                        redIndex;


                    drawPendulum(
                        pendulumIndex
                    );


                    if (
                        greenIndex !== null
                    ) {

                        manualSlider.value =
                            greenIndex;


                        manualReadout.textContent =
                            "t = "
                            +
                            D.t[greenIndex].toFixed(3)
                            +
                            " s"
                            +
                            "   |   θ = "
                            +
                            D.theta[greenIndex].toFixed(4)
                            +
                            " rad"
                            +
                            "   |   θ̇ = "
                            +
                            D.omega[greenIndex].toFixed(4)
                            +
                            " rad/s";
                    }
                }


                // ==========================================
                // MODALITÀ MANUALE
                // ==========================================

                function activateManualMode() {

                    manualMode =
                        true;


                    manualBox.style.display =
                        "block";


                    if (
                        greenIndex === null
                    ) {

                        greenIndex =
                            redIndex;
                    }


                    drawAll();
                }


                function deactivateManualMode() {

                    manualMode =
                        false;

                    greenIndex =
                        null;

                    manualBox.style.display =
                        "none";

                    drawAll();
                }


                // ==========================================
                // TIMER
                // ==========================================

                function stop() {

                    if (
                        window[
                            uid + "_timer"
                        ]
                    ) {

                        clearInterval(
                            window[
                                uid + "_timer"
                            ]
                        );


                        window[
                            uid + "_timer"
                        ] =
                            null;
                    }
                }


                function play() {

                    stop();


                    deactivateManualMode();


                    if (
                        redIndex >= N - 1
                    ) {

                        redIndex =
                            0;
                    }


                    window[
                        uid + "_timer"
                    ] =
                        setInterval(
                            () => {

                                const speed =
                                    Number(
                                        speedSel.value
                                    );


                                const dt =
                                    (
                                        D.t[N-1]
                                        -
                                        D.t[0]
                                    )
                                    /
                                    Math.max(
                                        1,
                                        N - 1
                                    );


                                const step =
                                    Math.max(
                                        1,
                                        Math.round(
                                            0.04
                                            *
                                            speed
                                            /
                                            dt
                                        )
                                    );


                                redIndex +=
                                    step;


                                if (
                                    redIndex >= N - 1
                                ) {

                                    redIndex =
                                        N - 1;


                                    drawAll();


                                    stop();


                                    activateManualMode();


                                    return;
                                }


                                drawAll();

                            },
                            40
                        );
                }


                el("play").onclick =
                    play;


                el("pause").onclick =
                    stop;


                el("reset").onclick =
                    () => {

                        stop();

                        manualMode =
                            false;

                        greenIndex =
                            null;

                        manualBox.style.display =
                            "none";

                        redIndex =
                            0;

                        drawAll();
                    };


                // ==========================================
                // CURSORE ROSSO
                // ==========================================

                slider.addEventListener(
                    "input",
                    () => {

                        stop();


                        manualMode =
                            false;

                        greenIndex =
                            null;

                        manualBox.style.display =
                            "none";


                        redIndex =
                            Number(
                                slider.value
                            );


                        drawAll();
                    }
                );


                // ==========================================
                // CURSORE VERDE
                // ==========================================

                manualSlider.addEventListener(
                    "input",
                    () => {

                        stop();


                        manualMode =
                            true;


                        greenIndex =
                            Number(
                                manualSlider.value
                            );


                        drawAll();
                    }
                );


                // ==========================================
                // CLICK SUL PIANO DEGLI STATI
                //
                // Troviamo il punto della traiettoria più
                // vicino al punto cliccato.
                // ==========================================

                el("phase").addEventListener(
                    "click",
                    event => {

                        if (
                            !manualMode
                            ||
                            phaseGeometry === null
                        ) {

                            return;
                        }


                        const canvas =
                            el("phase");


                        const rect =
                            canvas.getBoundingClientRect();


                        const px =
                            event.clientX
                            -
                            rect.left;


                        const py =
                            event.clientY
                            -
                            rect.top;


                        const A =
                            phaseGeometry;


                        if (
                            px < A.m.l
                            ||
                            px > rect.width - A.m.r
                            ||
                            py < A.m.t
                            ||
                            py > rect.height - A.m.b
                        ) {

                            return;
                        }


                        const thClick =
                            A.xmin
                            +
                            (
                                px - A.m.l
                            )
                            /
                            A.pw
                            *
                            (
                                A.xmax
                                -
                                A.xmin
                            );


                        const omClick =
                            A.ymax
                            -
                            (
                                py - A.m.t
                            )
                            /
                            A.ph
                            *
                            (
                                A.ymax
                                -
                                A.ymin
                            );


                        const xr =
                            Math.max(
                                1e-12,
                                A.xmax - A.xmin
                            );


                        const yr =
                            Math.max(
                                1e-12,
                                A.ymax - A.ymin
                            );


                        let best =
                            0;


                        let bestD =
                            Infinity;


                        for (
                            let i = 0;
                            i < N;
                            i++
                        ) {

                            const dx =
                                (
                                    D.theta[i]
                                    -
                                    thClick
                                )
                                /
                                xr;


                            const dy =
                                (
                                    D.omega[i]
                                    -
                                    omClick
                                )
                                /
                                yr;


                            const d =
                                dx*dx
                                +
                                dy*dy;


                            if (
                                d < bestD
                            ) {

                                bestD =
                                    d;

                                best =
                                    i;
                            }
                        }


                        greenIndex =
                            best;


                        drawAll();
                    }
                );


                // ==========================================
                // CLICK SU θ(t)
                //
                // Seleziona direttamente l'istante cliccato.
                // ==========================================

                el("theta").addEventListener(
                    "click",
                    event => {

                        if (
                            !manualMode
                            ||
                            thetaGeometry === null
                        ) {

                            return;
                        }


                        const canvas =
                            el("theta");


                        const rect =
                            canvas.getBoundingClientRect();


                        const px =
                            event.clientX
                            -
                            rect.left;


                        const A =
                            thetaGeometry;


                        if (
                            px < A.m.l
                            ||
                            px > rect.width - A.m.r
                        ) {

                            return;
                        }


                        const tClick =
                            A.xmin
                            +
                            (
                                px - A.m.l
                            )
                            /
                            A.pw
                            *
                            (
                                A.xmax
                                -
                                A.xmin
                            );


                        let best =
                            0;


                        let bestDt =
                            Infinity;


                        for (
                            let i = 0;
                            i < N;
                            i++
                        ) {

                            const d =
                                Math.abs(
                                    D.t[i]
                                    -
                                    tClick
                                );


                            if (
                                d < bestDt
                            ) {

                                bestDt =
                                    d;

                                best =
                                    i;
                            }
                        }


                        greenIndex =
                            best;


                        drawAll();
                    }
                );


                // ==========================================
                // RESIZE
                // ==========================================

                window.addEventListener(
                    "resize",
                    () => {

                        drawAll();

                    }
                );


                drawAll();

            })();
            """


            js = (
                js_template
                .replace(
                    "__UID__",
                    uid
                )
                .replace(
                    "__DATA__",
                    data_js
                )
            )


            colab_output.eval_js(
                js
            )


        except Exception as e:

            display(HTML(
                f"""
                <div style="
                    background:#fde2e1;
                    color:#7a1f1f;
                    border-left:5px solid #c62828;
                    padding:10px 14px;
                    border-radius:6px;">

                    Impossibile eseguire l'attività:
                    {str(e)}

                </div>
                """
            ))


crit_btn.on_click(
    _prova_critica
)
````

## Teoria — Equilibri, linearizzazione, centro e sella

Nelle Attività 7.1 e 8 hai osservato numericamente due proprietà importanti del pendolo:

- una piccola perturbazione della posizione inferiore produce un moto che rimane nelle sue vicinanze;
- una piccola perturbazione della posizione superiore porta invece il sistema ad allontanarsi rapidamente da quella configurazione.

Nell'Attività 8 hai inoltre osservato l'esistenza di un comportamento limite tra oscillazione e rotazione.

Ora vogliamo interpretare matematicamente queste osservazioni.

> **Nota.** Nell'Attività 8 abbiamo utilizzato il pendolo di riferimento
> $\ell=1\,\mathrm m$, $g=9.81\,\mathrm{m/s^2}$.
> Nella trattazione teorica torniamo invece a mantenere $\ell$ e $g$
> simbolici, così da ottenere risultati validi per qualunque pendolo semplice ideale.

Il sistema dinamico è

$$
\boxed{
\begin{cases}
\dot\theta=\omega,\\[1mm]
\dot\omega=-\dfrac{g}{\ell}\sin\theta.
\end{cases}}
$$

Uno stato di equilibrio

$$
\mathbf z_e=(\theta_e,\omega_e)
$$

deve soddisfare

$$
\dot{\mathbf z}=0.
$$

Pertanto

$$
\omega_e=0,
\qquad
\sin\theta_e=0,
$$

e quindi

$$
\boxed{
(\theta_e,\omega_e)=(k\pi,0),
\qquad k\in\mathbb Z.
}
$$

Poiché la configurazione angolare è periodica, esistono due tipi fisicamente distinti:

- $\theta=2k\pi$: pendolo rivolto verso il basso;
- $\theta=(2k+1)\pi$: pendolo rivolto verso l'alto.


### Linearizzazione del sistema

Per studiare ciò che accade vicino a un equilibrio introduciamo una piccola perturbazione

$$
\theta=\theta_e+\eta,
\qquad
|\eta|\ll1.
$$

Poiché

$$
\sin(\theta_e+\eta)
\simeq
\sin\theta_e+\cos\theta_e\,\eta
$$

e negli equilibri $\sin\theta_e=0$, otteniamo

$$
\ddot\eta
\simeq
-\frac{g}{\ell}\cos\theta_e\,\eta.
$$

Nel piano

$$
(\eta,\dot\eta)
$$

il sistema linearizzato può essere scritto

$$
\frac{d}{dt}
\begin{pmatrix}
\eta\\[1mm]
\dot\eta
\end{pmatrix}
=
A(\theta_e)
\begin{pmatrix}
\eta\\[1mm]
\dot\eta
\end{pmatrix},
$$

con

$$
\boxed{
A(\theta_e)=
\begin{pmatrix}
0&1\\[2mm]
-\dfrac{g}{\ell}\cos\theta_e&0
\end{pmatrix}.
}
$$


### Equilibrio inferiore

Per

$$
\theta_e=0
$$

si ha

$$
A_0=
\begin{pmatrix}
0&1\\[2mm]
-\dfrac{g}{\ell}&0
\end{pmatrix}.
$$

Gli autovalori sono

$$
\boxed{
\lambda_{1,2}
=
\pm i\sqrt{\frac{g}{\ell}}.
}
$$

Sono puramente immaginari: il sistema linearizzato non presenta né crescita
né decadimento esponenziale, ma un moto oscillatorio.

Nel piano $(\eta,\dot\eta)$ le traiettorie del sistema linearizzato sono
curve chiuse attorno all'origine.

Questo tipo di equilibrio del sistema linearizzato prende il nome di

$$
\boxed{\text{centro}.}
$$

È precisamente il comportamento locale che hai osservato numericamente
perturbando la posizione inferiore.

Gli autovalori puramente immaginari della linearizzazione, da soli, non
costituiscono in generale una prova di stabilità per qualunque sistema non
lineare. Nel caso del pendolo completo la stabilità dell'equilibrio inferiore
verrà successivamente collegata anche alla struttura energetica del sistema.


### Equilibrio superiore

Consideriamo ora

$$
\theta_e=\pi.
$$

Poiché

$$
\cos\pi=-1,
$$

la matrice linearizzata diventa

$$
A_\pi=
\begin{pmatrix}
0&1\\[2mm]
\dfrac{g}{\ell}&0
\end{pmatrix}.
$$

Gli autovalori sono

$$
\boxed{
\lambda_+
=
+\sqrt{\frac{g}{\ell}},
\qquad
\lambda_-
=
-\sqrt{\frac{g}{\ell}}.
}
$$

Questa volta gli autovalori sono reali e di segno opposto.

Esiste quindi un modo che cresce esponenzialmente nel tempo e un modo che
decade esponenzialmente.

L'equilibrio superiore ha perciò la struttura locale di una

$$
\boxed{\text{sella}.}
$$

Ponendo

$$
\omega_0=\sqrt{\frac{g}{\ell}},
$$

gli autovettori possono essere scelti proporzionali a

$$
\mathbf v_+
=
\begin{pmatrix}
1\\
+\omega_0
\end{pmatrix},
\qquad
\mathbf v_-
=
\begin{pmatrix}
1\\
-\omega_0
\end{pmatrix}.
$$

Nel piano $(\eta,\dot\eta)$ individuano quindi le due direzioni locali

$$
\boxed{
\dot\eta=+\sqrt{\frac{g}{\ell}}\,\eta,
\qquad
\dot\eta=-\sqrt{\frac{g}{\ell}}\,\eta.
}
$$

Una è associata al moto che si allontana dalla sella andando avanti nel
tempo; l'altra al moto che vi si avvicina.

Queste rette descrivono soltanto le **direzioni locali** vicino
all'equilibrio: non sono, in generale, l'intera traiettoria del sistema
non lineare.

Ed è proprio questa struttura di sella che ci permetterà di interpretare
la traiettoria limite osservata nell'Attività 8.

<br>

Questa sezione è un **ripasso facoltativo**.

Nella sezione precedente abbiamo utilizzato gli autovalori della matrice linearizzata per riconoscere un **centro** e una **sella**.

Qui ricordiamo brevemente **perché autovalori e autovettori contengono informazioni sull'evoluzione temporale del sistema**.

Se questo passaggio ti è già familiare, puoi proseguire direttamente alla teoria della separatrice.

---

### 1 — Un sistema lineare vicino all'equilibrio

Dopo aver linearizzato un sistema dinamico attorno a un equilibrio, si ottiene un sistema della forma

$$\boxed{\dot{\mathbf{x}}=A\mathbf{x}}$$

dove $\mathbf{x}$ rappresenta una piccola perturbazione dello stato rispetto all'equilibrio.

Nel caso del pendolo possiamo, per esempio, descrivere la perturbazione mediante

$$\boxed{\mathbf{x}=(\eta,\dot{\eta})^{\mathsf T}}$$

dove $\eta$ misura il piccolo scostamento angolare dall'equilibrio considerato.

---

### 2 — Perché compaiono autovalori e autovettori?

Cerchiamo un moto elementare della forma

$$\boxed{\mathbf{x}(t)=\mathbf{v}\,e^{\lambda t}}$$

dove $\mathbf{v}$ è un vettore costante.

Derivando rispetto al tempo otteniamo

$$\dot{\mathbf{x}}(t)=\lambda\,\mathbf{v}\,e^{\lambda t}.$$

D'altra parte il sistema deve soddisfare

$$\dot{\mathbf{x}}=A\mathbf{x},$$

quindi deve essere anche

$$\dot{\mathbf{x}}(t)=A\mathbf{v}\,e^{\lambda t}.$$

Confrontando le due espressioni,

$$\lambda\,\mathbf{v}\,e^{\lambda t}=A\mathbf{v}\,e^{\lambda t}.$$

Poiché $e^{\lambda t}\neq 0$, possiamo eliminarlo e otteniamo

$$\boxed{A\mathbf{v}=\lambda\mathbf{v}}$$

che è precisamente l'**equazione agli autovalori**.

**Idea chiave:** autovalori e autovettori non vengono introdotti artificialmente. Emergono quando cerchiamo i modi elementari di evoluzione del sistema linearizzato.

---

### 3 — Significato dinamico

Se lo stato iniziale è diretto lungo un autovettore,

$$\mathbf{x}(0)=C\mathbf{v},$$

la corrispondente evoluzione è

$$\boxed{\mathbf{x}(t)=C\,e^{\lambda t}\mathbf{v}}$$

I due oggetti hanno quindi ruoli differenti:

- l'**autovettore** $\mathbf{v}$ individua una direzione caratteristica nello spazio degli stati;
- l'**autovalore** $\lambda$ determina come il moto evolve nel tempo lungo quella direzione.

Il fattore fondamentale è

$$\boxed{e^{\lambda t}}$$

---

### 4 — Autovalori reali

Se

$$\lambda<0,$$

allora

$$e^{\lambda t}\longrightarrow 0 \qquad (t\rightarrow+\infty).$$

Il moto lungo quella direzione si avvicina all'equilibrio.

Se invece

$$\lambda>0,$$

il fattore $e^{\lambda t}$ cresce esponenzialmente e il moto si allontana dall'equilibrio.

Possiamo quindi ricordare:

$$\boxed{\lambda<0\quad\Longrightarrow\quad\text{direzione localmente attrattiva}}$$

$$\boxed{\lambda>0\quad\Longrightarrow\quad\text{direzione localmente repulsiva}}$$

Questo è precisamente il tipo di struttura che abbiamo trovato vicino alla **sella** del pendolo.

---

### 5 — Autovalori complessi

Se

$$\lambda=\alpha+i\beta,$$

allora

$$e^{\lambda t}=e^{\alpha t}e^{i\beta t}=e^{\alpha t}\left[\cos(\beta t)+i\sin(\beta t)\right].$$

La parte reale

$$\boxed{\operatorname{Re}(\lambda)=\alpha}$$

controlla la crescita o il decadimento dell'ampiezza.

La parte immaginaria

$$\boxed{\operatorname{Im}(\lambda)=\beta}$$

determina invece la frequenza dell'oscillazione.

Nel caso particolare

$$\lambda=\pm i\beta,$$

la parte reale è nulla e il sistema linearizzato oscilla senza crescita né decadimento.

È il comportamento trovato vicino al **centro** del pendolo.

---

### 6 — Una condizione iniziale generica

Se la matrice $A$ possiede una base di autovettori, una perturbazione iniziale può essere scritta come

$$\mathbf{x}(0)=c_1\mathbf{v}_1+c_2\mathbf{v}_2+\cdots$$

e la soluzione assume la forma

$$\boxed{\mathbf{x}(t)=c_1e^{\lambda_1 t}\mathbf{v}_1+c_2e^{\lambda_2 t}\mathbf{v}_2+\cdots}$$

Ogni componente dello stato evolve quindi secondo il proprio **modo caratteristico**.

---

### 7 — Il punto importante per la sella del pendolo

Vicino all'equilibrio superiore

$$(\theta,\dot{\theta})=(\pi,0),$$

abbiamo trovato due autovalori reali di segno opposto:

$$\boxed{\lambda_+=+\sqrt{\frac{g}{\ell}},\qquad \lambda_-=-\sqrt{\frac{g}{\ell}}}$$

Le corrispondenti direzioni locali nello spazio degli stati sono

$$\boxed{\dot{\eta}=+\sqrt{\frac{g}{\ell}}\,\eta}$$

e

$$\boxed{\dot{\eta}=-\sqrt{\frac{g}{\ell}}\,\eta}$$

Una è associata al modo che cresce andando avanti nel tempo; l'altra al modo che decade.

In particolare, lungo il modo associato a

$$\lambda_-<0,$$

la distanza dall'equilibrio diminuisce proporzionalmente a

$$e^{-\sqrt{g/\ell}\,t}.$$

Il sistema si avvicina quindi sempre di più alla sella, ma raggiunge esattamente l'equilibrio soltanto nel limite

$$\boxed{t\rightarrow+\infty}$$

Questa osservazione sarà importante per comprendere geometricamente la traiettoria limite osservata nell'Attività 8.

---

### Da ricordare prima di proseguire

**Per il sistema linearizzato:**

$$\boxed{\lambda<0\quad\Longrightarrow\quad\text{avvicinamento all'equilibrio}}$$

$$\boxed{\lambda>0\quad\Longrightarrow\quad\text{allontanamento dall'equilibrio}}$$

$$\boxed{\lambda=\pm i\beta\quad\Longrightarrow\quad\text{oscillazione}}$$

Gli **autovettori** individuano le corrispondenti **direzioni locali** nello spazio degli stati.

Nel sistema non lineare completo queste direzioni descrivono il comportamento nelle immediate vicinanze dell'equilibrio.

Il passo successivo sarà capire come alcune di queste direzioni locali si prolungano nel **ritratto di fase globale**.

</details>


````python
#@title Ripasso facoltativo — Perché gli autovalori sono collegati alle soluzioni del moto { display-mode: "form" }

import ipywidgets as widgets
from IPython.display import display, Markdown, Math, HTML, clear_output

# Elimina eventuali output/widget rimasti da esecuzioni precedenti
clear_output(wait=True)

# ============================================================
# CONTENITORE COLLASSABILE
# ============================================================

out_ripasso_autovalori = widgets.Output()

accordion_ripasso = widgets.Accordion(
    children=[out_ripasso_autovalori],
    selected_index=None
)

accordion_ripasso.set_title(
    0,
    "Ripasso facoltativo — Perché gli autovalori sono collegati alle soluzioni del moto"
)

display(accordion_ripasso)


# ============================================================
# CONTENUTO DEL RIPASSO
# ============================================================

with out_ripasso_autovalori:

    display(Markdown(r"""
Questa sezione è un **ripasso facoltativo**.

Nella sezione precedente abbiamo utilizzato gli autovalori della matrice
linearizzata per riconoscere un **centro** e una **sella**.

Qui ricordiamo brevemente **perché gli autovalori e gli autovettori
contengono informazioni sull'evoluzione temporale del sistema**.

Se questo passaggio ti è già familiare, puoi proseguire direttamente
alla teoria della separatrice.

---

### 1 — Un sistema lineare vicino all'equilibrio

Dopo aver linearizzato un sistema dinamico attorno a un equilibrio,
si ottiene un sistema della forma
"""))

    display(Math(
        r"\boxed{\dot{\mathbf x}=A\mathbf x}"
    ))

    display(Markdown(r"""
dove $\mathbf x$ rappresenta una piccola perturbazione dello stato
rispetto all'equilibrio.

Nel caso del pendolo possiamo, per esempio, scrivere
"""))

    display(Math(
        r"\mathbf x="
        r"\begin{pmatrix}"
        r"\eta\\[2mm]"
        r"\dot{\eta}"
        r"\end{pmatrix},"
    ))

    display(Markdown(r"""
dove $\eta$ misura il piccolo scostamento angolare
dall'equilibrio considerato.

---

### 2 — Perché compaiono autovalori e autovettori?

Cerchiamo un moto elementare della forma
"""))

    display(Math(
        r"\boxed{\mathbf x(t)=\mathbf v\,e^{\lambda t}}"
    ))

    display(Markdown(r"""
dove $\mathbf v$ è un vettore costante.

Derivando rispetto al tempo:
"""))

    display(Math(
        r"\dot{\mathbf x}(t)"
        r"="
        r"\lambda\,\mathbf v\,e^{\lambda t}."
    ))

    display(Markdown(r"""
Ma, poiché il sistema deve soddisfare
"""))

    display(Math(
        r"\dot{\mathbf x}=A\mathbf x,"
    ))

    display(Markdown(r"""
deve essere anche
"""))

    display(Math(
        r"\dot{\mathbf x}(t)"
        r"="
        r"A\mathbf v\,e^{\lambda t}."
    ))

    display(Markdown(r"""
Confrontando le due espressioni:
"""))

    display(Math(
        r"\lambda\mathbf v\,e^{\lambda t}"
        r"="
        r"A\mathbf v\,e^{\lambda t}."
    ))

    display(Markdown(r"""
Poiché $e^{\lambda t}\neq0$, possiamo eliminarlo e otteniamo
"""))

    display(Math(
        r"\boxed{A\mathbf v=\lambda\mathbf v}"
    ))

    display(Markdown(r"""
che è precisamente l'**equazione agli autovalori**.

Gli autovalori e gli autovettori non vengono quindi introdotti
artificialmente: emergono quando cerchiamo i **modi elementari di evoluzione**
del sistema linearizzato.

---

### 3 — Significato dinamico

Se lo stato iniziale è diretto lungo un autovettore,
"""))

    display(Math(
        r"\mathbf x(0)=C\mathbf v,"
    ))

    display(Markdown(r"""
la corrispondente evoluzione è
"""))

    display(Math(
        r"\boxed{\mathbf x(t)=C\,e^{\lambda t}\mathbf v}"
    ))

    display(Markdown(r"""
I due oggetti hanno quindi ruoli differenti:

- l'**autovettore** $\mathbf v$ individua una direzione caratteristica
  nello spazio degli stati;
- l'**autovalore** $\lambda$ determina come il moto evolve nel tempo
  lungo quella direzione.

Il fattore fondamentale è
"""))

    display(Math(
        r"e^{\lambda t}."
    ))


    display(Markdown(r"""
---

### 4 — Autovalori reali

Se
"""))

    display(Math(
        r"\lambda<0,"
    ))

    display(Markdown(r"""
allora
"""))

    display(Math(
        r"e^{\lambda t}\longrightarrow0"
        r"\qquad"
        r"(t\rightarrow+\infty)."
    ))

    display(Markdown(r"""
Il moto lungo quella direzione si avvicina all'equilibrio.

Se invece
"""))

    display(Math(
        r"\lambda>0,"
    ))

    display(Markdown(r"""
il fattore $e^{\lambda t}$ cresce esponenzialmente
e il moto si allontana dall'equilibrio.

Possiamo quindi ricordare:
"""))

    display(Math(
        r"\boxed{"
        r"\lambda<0"
        r"\quad\Longrightarrow\quad"
        r"\text{direzione localmente attrattiva}"
        r"}"
    ))

    display(Math(
        r"\boxed{"
        r"\lambda>0"
        r"\quad\Longrightarrow\quad"
        r"\text{direzione localmente repulsiva}"
        r"}"
    ))

    display(Markdown(r"""
Questo è precisamente il tipo di struttura che abbiamo trovato
vicino alla **sella** del pendolo.

---

### 5 — Autovalori complessi

Se
"""))

    display(Math(
        r"\lambda=\alpha+i\beta,"
    ))

    display(Markdown(r"""
allora
"""))

    display(Math(
        r"e^{\lambda t}"
        r"="
        r"e^{\alpha t}e^{i\beta t}"
        r"="
        r"e^{\alpha t}"
        r"\left["
        r"\cos(\beta t)+i\sin(\beta t)"
        r"\right]."
    ))

    display(Markdown(r"""
La parte reale
"""))

    display(Math(
        r"\boxed{\operatorname{Re}(\lambda)=\alpha}"
    ))

    display(Markdown(r"""
controlla la crescita o il decadimento dell'ampiezza, mentre la parte
immaginaria
"""))

    display(Math(
        r"\boxed{\operatorname{Im}(\lambda)=\beta}"
    ))

    display(Markdown(r"""
determina la frequenza dell'oscillazione.

Nel caso particolare
"""))

    display(Math(
        r"\lambda=\pm i\beta,"
    ))

    display(Markdown(r"""
la parte reale è nulla e il sistema linearizzato oscilla
senza crescita né decadimento.

È il comportamento trovato vicino al **centro** del pendolo.

---

### 6 — Una condizione iniziale generica

Se la matrice $A$ possiede una base di autovettori, una perturbazione
iniziale può essere scritta come
"""))

    display(Math(
        r"\mathbf x(0)"
        r"="
        r"c_1\mathbf v_1"
        r"+"
        r"c_2\mathbf v_2"
        r"+\cdots"
    ))

    display(Markdown(r"""
e la soluzione assume la forma
"""))

    display(Math(
        r"\boxed{"
        r"\mathbf x(t)"
        r"="
        r"c_1e^{\lambda_1t}\mathbf v_1"
        r"+"
        r"c_2e^{\lambda_2t}\mathbf v_2"
        r"+\cdots"
        r"}"
    ))

    display(Markdown(r"""
Ogni componente dello stato evolve quindi secondo il proprio
**modo caratteristico**.

---

### 7 — Il punto importante per la sella del pendolo

Vicino all'equilibrio superiore
"""))

    display(Math(
        r"(\theta,\dot{\theta})=(\pi,0),"
    ))

    display(Markdown(r"""
abbiamo trovato due autovalori reali di segno opposto:
"""))

    display(Math(
        r"\lambda_+"
        r"="
        r"+\sqrt{\frac{g}{\ell}},"
        r"\qquad"
        r"\lambda_-"
        r"="
        r"-\sqrt{\frac{g}{\ell}}."
    ))

    display(Markdown(r"""
Le corrispondenti direzioni locali nello spazio degli stati sono
"""))

    display(Math(
        r"\dot{\eta}"
        r"="
        r"+\sqrt{\frac{g}{\ell}}\,\eta"
    ))

    display(Markdown(r"""
e
"""))

    display(Math(
        r"\dot{\eta}"
        r"="
        r"-\sqrt{\frac{g}{\ell}}\,\eta."
    ))

    display(Markdown(r"""
Una è associata al modo che cresce andando avanti nel tempo;
l'altra al modo che decade.

In particolare, lungo il modo associato a
"""))

    display(Math(
        r"\lambda_-<0,"
    ))

    display(Markdown(r"""
la distanza dall'equilibrio diminuisce proporzionalmente a
"""))

    display(Math(
        r"e^{-\sqrt{g/\ell}\,t}."
    ))

    display(Markdown(r"""
Il sistema si avvicina quindi sempre di più alla sella, ma raggiunge
esattamente l'equilibrio soltanto nel limite
"""))

    display(Math(
        r"\boxed{t\rightarrow+\infty}"
    ))

    display(Markdown(r"""
Questa osservazione sarà importante per comprendere geometricamente
la traiettoria limite osservata nell'Attività 8.

---

### Da ricordare prima di proseguire
"""))

    display(HTML(
        """
        <div style="
            background:#eef6ff;
            color:#16324f;
            border-left:5px solid #4a90e2;
            padding:11px 14px;
            margin:10px 0 14px 0;
            border-radius:6px;">
            <b>Per il sistema linearizzato:</b>
        </div>
        """
    ))

    display(Math(
        r"\boxed{"
        r"\lambda<0"
        r"\quad\Longrightarrow\quad"
        r"\text{avvicinamento all'equilibrio}"
        r"}"
    ))

    display(Math(
        r"\boxed{"
        r"\lambda>0"
        r"\quad\Longrightarrow\quad"
        r"\text{allontanamento dall'equilibrio}"
        r"}"
    ))

    display(Math(
        r"\boxed{"
        r"\lambda=\pm i\beta"
        r"\quad\Longrightarrow\quad"
        r"\text{oscillazione}"
        r"}"
    ))

    display(Markdown(r"""
Gli **autovettori** individuano le corrispondenti **direzioni locali**
nello spazio degli stati.

Nel sistema non lineare completo queste direzioni descrivono il comportamento
nelle immediate vicinanze dell'equilibrio.

Il passo successivo sarà capire come alcune di queste direzioni locali
si prolungano nel **ritratto di fase globale**.
"""))
````

## Teoria — La separatrice

Nell'Attività 8 hai esplorato il moto del pendolo partendo dalla configurazione inferiore,

$$
\theta(0)=0,
$$

e modificando progressivamente la velocità angolare iniziale

$$
\dot\theta(0)>0.
$$

Hai osservato due comportamenti qualitativamente differenti.

Per valori sufficientemente piccoli di $\dot\theta(0)$, il pendolo sale,
raggiunge un angolo massimo e poi inverte il moto:

$$
\boxed{\text{oscillazione}}.
$$

Per valori sufficientemente grandi, invece, il pendolo supera la configurazione superiore

$$
\theta=\pi
$$

e continua il moto:

$$
\boxed{\text{rotazione}}.
$$

Aumentando progressivamente $\dot\theta(0)$, hai quindi incontrato un
**confine fra due comportamenti dinamici differenti**.

La domanda ora è:

> **che cosa rappresenta geometricamente questo confine nello spazio degli stati?**

---

### 1 — Torniamo all'equilibrio superiore

Nella sezione precedente abbiamo visto che

$$
(\theta,\dot\theta)=(\pi,0)
$$

è un equilibrio instabile di tipo **sella**.

Introducendo

$$
\eta=\theta-\pi,
$$

le due direzioni caratteristiche locali della sella sono

$$
\boxed{
\dot\eta=
+\sqrt{\frac{g}{\ell}}\,\eta
}
$$

e

$$
\boxed{
\dot\eta=
-\sqrt{\frac{g}{\ell}}\,\eta.
}
$$

Queste rette descrivono il comportamento soltanto **nelle immediate
vicinanze dell'equilibrio**.

Una delle due direzioni è associata, andando avanti nel tempo, a moti
che si allontanano dalla sella; l'altra a moti che vi si avvicinano.

---

### 2 — Dalle direzioni locali alle traiettorie globali

Le direzioni proprie ottenute con la linearizzazione non terminano
quando ci allontaniamo dall'equilibrio.

Se seguiamo il sistema non lineare completo, esse si prolungano in
particolari traiettorie dello spazio degli stati.

Possiamo quindi distinguere:

- traiettorie che **si avvicinano** alla sella per $t\to+\infty$;
- traiettorie che **si allontanano** dalla sella per $t\to+\infty$.

Le loro estensioni globali prendono rispettivamente il nome di
**rami stabili** e **rami instabili** della sella.

Vicino all'equilibrio essi sono tangenti alle direzioni individuate
dalla linearizzazione; più lontano dalla sella, invece, la loro forma
è determinata dall'equazione non lineare completa del pendolo.

---

### 3 — Che cosa accade alla traiettoria critica?

Consideriamo ora ciò che hai cercato numericamente nell'Attività 8.

Partendo da

$$
\theta(0)=0
$$

e aumentando $\dot\theta(0)$, esiste una condizione iniziale limite
fra oscillazione e rotazione.

Per questa condizione critica il pendolo sale verso la posizione superiore,
ma il suo moto rallenta sempre di più.

Nel limite,

$$
\boxed{
\theta(t)\longrightarrow\pi
}
$$

e contemporaneamente

$$
\boxed{
\dot\theta(t)\longrightarrow0
}
$$

per

$$
\boxed{
t\longrightarrow+\infty.
}
$$

Quindi il pendolo **non attraversa la configurazione superiore in un
tempo finito**.

La sua traiettoria nello spazio degli stati si avvicina sempre di più
al punto

$$
(\pi,0),
$$

seguendo uno dei rami stabili della sella.

Questo chiarisce anche perché, vicino alla soglia, nella simulazione il
pendolo sembra trascorrere un tempo sempre più lungo nelle vicinanze
della configurazione superiore.

---

### 4 — Tre condizioni iniziali quasi uguali, tre destini differenti

Immaginiamo ora tre valori iniziali molto vicini:

$$
\dot\theta_0^-,
\qquad
\dot\theta_0^*,
\qquad
\dot\theta_0^+,
$$

con

$$
\dot\theta_0^-<
\dot\theta_0^*<
\dot\theta_0^+.
$$

Il valore centrale rappresenta la condizione critica.

Se

$$
\dot\theta(0)=\dot\theta_0^-,
$$

il pendolo non raggiunge la posizione superiore: si arresta prima,
inverte il moto e rimane nel regime di oscillazione.

Se

$$
\dot\theta(0)=\dot\theta_0^*,
$$

il pendolo si avvicina asintoticamente alla configurazione superiore:

$$
(\theta,\dot\theta)\longrightarrow(\pi,0).
$$

Se invece

$$
\dot\theta(0)=\dot\theta_0^+,
$$

il pendolo supera la posizione superiore ed entra nel regime di rotazione.

Quindi una singola traiettoria divide nello spazio degli stati due
famiglie di moti qualitativamente differenti.

---

### 5 — La separatrice

Questa traiettoria limite prende il nome di **separatrice**.

In generale, una separatrice è una traiettoria — o un insieme di
traiettorie — dello spazio degli stati che costituisce il confine fra
regioni caratterizzate da comportamenti dinamici qualitativamente differenti.

Nel pendolo semplice:

$$
\boxed{
\text{oscillazioni}
\quad\Big|\quad
\text{separatrice}
\quad\Big|\quad
\text{rotazioni}.
}
$$

La separatrice è quindi il confine geometrico che organizza il ritratto
di fase.

Nel caso del pendolo essa è costruita dai rami stabili e instabili
associati agli equilibri superiori

$$
\boxed{
\big((2k+1)\pi,0\big),
\qquad
k\in\mathbb Z.
}
$$

È importante interpretare correttamente questa affermazione:

il punto di sella appartiene alla struttura della separatrice, ma una
traiettoria non banale della separatrice **non attraversa la sella in
un tempo finito**.

La raggiunge soltanto asintoticamente.

---

### 6 — Piano srotolato e natura periodica dell'angolo

Nel piano $(\theta,\dot\theta)$ abbiamo finora rappresentato $\theta$
come una coordinata reale non limitata.

In questo **piano srotolato** compaiono quindi copie dell'equilibrio
superiore:

$$
\ldots,\;
(-3\pi,0),\;
(-\pi,0),\;
(\pi,0),\;
(3\pi,0),\;
\ldots
$$

I rami della separatrice collegano asintoticamente copie adiacenti
di queste selle.

Per esempio, un ramo con $\dot\theta>0$ può essere interpretato come
una traiettoria che, andando avanti nel tempo, si allontana da una
copia della sella e si avvicina alla successiva.

Ma fisicamente

$$
\theta
\qquad\text{e}\qquad
\theta+2\pi
$$

rappresentano la stessa configurazione del pendolo.

Lo spazio fisico degli stati non è quindi realmente un piano infinito,
ma può essere pensato come un **cilindro**:

$$
S^1\times\mathbb R.
$$

Sul cilindro, le diverse copie

$$
\ldots,-\pi,\pi,3\pi,\ldots
$$

corrispondono allo stesso equilibrio fisico superiore.

---

### 7 — Che cosa sappiamo, e che cosa ancora non sappiamo?

A questo punto abbiamo identificato la separatrice utilizzando soltanto:

- l'equazione differenziale del pendolo;
- lo spazio degli stati;
- l'esplorazione numerica della condizione critica che separa oscillazioni da rotazioni (Attività 8);
- gli equilibri;
- la linearizzazione;
- la struttura locale della sella.

Abbiamo quindi capito **che cosa è** la separatrice e perché deve
costituire il confine fra oscillazioni e rotazioni.

Non abbiamo però ancora determinato analiticamente la sua forma.

Non sappiamo ancora, per esempio, ricavare direttamente una relazione

$$
\dot\theta=\dot\theta(\theta)
$$

che descriva esattamente quella curva.

E non abbiamo ancora calcolato analiticamente quale sia il valore preciso
della velocità angolare iniziale critica osservata nell'Attività 8.

La domanda successiva diventa quindi naturale:

> **esiste una quantità che rimane costante durante il moto e che permetta
> di riconoscere immediatamente a quale traiettoria appartiene uno stato
> $(\theta,\dot\theta)$?**

La risposta ci porterà alla conservazione dell'energia e permetterà di
ricavare analiticamente la separatrice.

## Teoria — Energia meccanica e forma analitica della separatrice

Nella sezione precedente abbiamo identificato la separatrice a partire
dalla dinamica nello spazio degli stati, dalla struttura della sella e
dall'esplorazione numerica.

Possiamo ora rispondere alla domanda rimasta aperta:

> **esiste una quantità costante lungo il moto che permetta di riconoscere
> analiticamente le diverse traiettorie del pendolo?**

Per il pendolo semplice ideale la risposta è sì.

---

### 1 — Una quantità conservata

La Lagrangiana del pendolo ideale è

$$
L(\theta,\dot\theta)
=
\frac12 m\ell^2\dot\theta^2
-
mg\ell(1-\cos\theta).
$$

Essa non dipende esplicitamente dal tempo.

Per un sistema che soddisfa le equazioni di Eulero–Lagrange si può
considerare la quantità

$$
\mathcal E
=
\dot\theta
\frac{\partial L}{\partial\dot\theta}
-
L.
$$

Derivandola rispetto al tempo si ottiene

$$
\frac{d\mathcal E}{dt}
=
\dot\theta
\left[
\frac{d}{dt}
\left(
\frac{\partial L}{\partial\dot\theta}
\right)
-
\frac{\partial L}{\partial\theta}
\right]
-
\frac{\partial L}{\partial t}.
$$

Il termine tra parentesi quadre è nullo per l'equazione di
Eulero–Lagrange e, nel nostro caso,

$$
\frac{\partial L}{\partial t}=0.
$$

Quindi

$$
\boxed{
\frac{d\mathcal E}{dt}=0
}
$$

e pertanto

$$
\boxed{
\mathcal E=\text{costante}.
}
$$

---

### 2 — Nel pendolo questa quantità è l'energia meccanica

Poiché

$$
\frac{\partial L}{\partial\dot\theta}
=
m\ell^2\dot\theta,
$$

si ha

$$
\mathcal E
=
m\ell^2\dot\theta^2
-
\left[
\frac12m\ell^2\dot\theta^2
-
mg\ell(1-\cos\theta)
\right].
$$

Pertanto

$$
\boxed{
E
=
\frac12m\ell^2\dot\theta^2
+
mg\ell(1-\cos\theta)
}
$$

è costante lungo ogni traiettoria.

I due termini sono precisamente

$$
T=
\frac12m\ell^2\dot\theta^2
$$

e

$$
V=
mg\ell(1-\cos\theta),
$$

per cui

$$
\boxed{E=T+V.}
$$

La conservazione dell'energia non introduce quindi una nuova equazione
del moto: è una conseguenza della stessa dinamica descritta dalle
equazioni di Eulero–Lagrange.

---

### 3 — Verifica direttamente dall'equazione del moto

Possiamo verificarlo anche utilizzando

$$
\ddot\theta
+
\frac{g}{\ell}\sin\theta
=
0.
$$

Derivando l'energia:

$$
\frac{dE}{dt}
=
m\ell^2\dot\theta\,\ddot\theta
+
mg\ell\sin\theta\,\dot\theta.
$$

Raccogliendo $m\ell^2\dot\theta$,

$$
\frac{dE}{dt}
=
m\ell^2\dot\theta
\left(
\ddot\theta
+
\frac{g}{\ell}\sin\theta
\right).
$$

L'equazione del moto implica quindi

$$
\boxed{
\frac{dE}{dt}=0.
}
$$

---

### 4 — Le traiettorie del ritratto di fase come livelli di energia

Poiché $E$ è costante, ogni stato

$$
(\theta,\dot\theta)
$$

raggiunto durante il moto deve soddisfare

$$
\frac12m\ell^2\dot\theta^2
+
mg\ell(1-\cos\theta)
=
E.
$$

Possiamo isolare $\dot\theta^2$:

$$
\boxed{
\dot\theta^2
=
\frac{2E}{m\ell^2}
-
\frac{2g}{\ell}(1-\cos\theta).
}
$$

Questa relazione definisce direttamente le curve nel piano

$$
(\theta,\dot\theta).
$$

Il ritratto di fase del pendolo può quindi essere interpretato come
una famiglia di **curve a energia costante**.

Valori differenti di $E$ corrispondono a famiglie differenti di
traiettorie.

---

### 5 — Qual è l'energia della sella?

L'equilibrio superiore si trova in

$$
(\theta,\dot\theta)=(\pi,0).
$$

La sua energia è

$$
E_{\mathrm s}
=
\frac12m\ell^2(0)^2
+
mg\ell(1-\cos\pi).
$$

Poiché

$$
\cos\pi=-1,
$$

otteniamo

$$
\boxed{
E_{\mathrm s}=2mg\ell.
}
$$

Questo è il livello energetico critico.

---

### 6 — Oscillazione, separatrice e rotazione

Consideriamo ora tre casi.

Se

$$
E<2mg\ell,
$$

il sistema non possiede energia sufficiente per raggiungere la
configurazione superiore con $\dot\theta=0$.

La traiettoria presenta punti di inversione e il moto è oscillatorio:

$$
\boxed{
E<2mg\ell
\quad\Longrightarrow\quad
\text{oscillazione}.
}
$$

Se invece

$$
E>2mg\ell,
$$

quando il pendolo raggiunge $\theta=\pi$ possiede ancora velocità
angolare diversa da zero e può superare la configurazione superiore:

$$
\boxed{
E>2mg\ell
\quad\Longrightarrow\quad
\text{rotazione}.
}
$$

Il caso limite è

$$
\boxed{
E=2mg\ell.
}
$$

È precisamente il livello energetico della separatrice:

$$
\boxed{
\text{oscillazione}
\quad\Big|\quad
E=2mg\ell
\quad\Big|\quad
\text{rotazione}.
}
$$

---

### 7 — Forma analitica della separatrice

Poniamo quindi

$$
E=2mg\ell
$$

nell'equazione delle curve a energia costante:

$$
\frac12m\ell^2\dot\theta^2
+
mg\ell(1-\cos\theta)
=
2mg\ell.
$$

Dividendo per $m\ell$ e riordinando:

$$
\frac12\ell\dot\theta^2
=
g(1+\cos\theta).
$$

Pertanto

$$
\boxed{
\dot\theta^2
=
\frac{2g}{\ell}(1+\cos\theta).
}
$$

Questa è già un'equazione esatta della separatrice nel piano degli stati.

Utilizzando

$$
1+\cos\theta
=
2\cos^2\frac{\theta}{2},
$$

otteniamo anche

$$
\boxed{
\dot\theta^2
=
\frac{4g}{\ell}
\cos^2\frac{\theta}{2}.
}
$$

Nell'intervallo fondamentale

$$
-\pi\leq\theta\leq\pi,
$$

dove $\cos(\theta/2)\geq0$, i due rami possono quindi essere scritti come

$$
\boxed{
\dot\theta
=
\pm
2\sqrt{\frac{g}{\ell}}
\cos\frac{\theta}{2}.
}
$$

Abbiamo così ottenuto analiticamente la curva che nell'Attività 8
avevamo cercato attraverso l'integrazione numerica.

---

### 8 — La soglia cercata nell'Attività 8

Nell'Attività 8 il pendolo partiva dalla configurazione inferiore,

$$
\theta(0)=0,
$$

con velocità angolare iniziale positiva.

In quella configurazione

$$
V(0)=0,
$$

quindi l'energia iniziale è

$$
E_0
=
\frac12m\ell^2\dot\theta(0)^2.
$$

La condizione critica si ottiene imponendo

$$
E_0=E_{\mathrm s}.
$$

Quindi

$$
\frac12m\ell^2\dot\theta_{\mathrm c}^{\,2}
=
2mg\ell.
$$

La massa si semplifica e otteniamo

$$
\boxed{
\dot\theta_{\mathrm c}
=
2\sqrt{\frac{g}{\ell}}.
}
$$

Per il pendolo utilizzato nell'Attività 8,

$$
\ell=1.00\,\mathrm m,
\qquad
g=9.81\,\mathrm{m/s^2},
$$

si ottiene

$$
\boxed{
\dot\theta_{\mathrm c}
\simeq
6.26\ \mathrm{rad/s}.
}
$$

Il valore che avevamo cercato numericamente possiede quindi una
determinazione analitica precisa.

---

### 9 — Coerenza con la linearizzazione vicino alla sella

La descrizione energetica globale deve essere compatibile con ciò che
abbiamo trovato localmente attraverso gli autovalori.

Vicino a

$$
\theta=\pi
$$

poniamo

$$
\eta=\theta-\pi,
\qquad
|\eta|\ll1.
$$

Poiché

$$
1+\cos(\pi+\eta)
\simeq
\frac{\eta^2}{2},
$$

l'equazione della separatrice

$$
\dot\theta^2
=
\frac{2g}{\ell}(1+\cos\theta)
$$

diventa localmente

$$
\dot\eta^2
\simeq
\frac{g}{\ell}\eta^2.
$$

Quindi

$$
\boxed{
\dot\eta
\simeq
\pm
\sqrt{\frac{g}{\ell}}\eta.
}
$$

Sono precisamente le due direzioni locali ottenute dalla
linearizzazione della sella.

La descrizione energetica globale e l'analisi locale mediante
autovalori sono quindi due letture diverse ma perfettamente coerenti
della stessa dinamica.

---

---

### 10 — Chiusura della descrizione analitica

Abbiamo ora descritto il pendolo semplice ideale da diversi punti di vista:

- equazioni di Eulero–Lagrange;
- integrazione numerica;
- evoluzione temporale;
- spazio degli stati;
- equilibri e stabilità;
- linearizzazione;
- centro e sella;
- separatrice;
- conservazione dell'energia.

Questi strumenti non producono descrizioni indipendenti: sono
rappresentazioni differenti della **stessa dinamica**.

In particolare, la descrizione energetica ci ha permesso di ottenere
analiticamente sia la forma della separatrice,

$$
\dot\theta
=
\pm 2\sqrt{\frac{g}{\ell}}
\cos\frac{\theta}{2},
$$

sia la velocità angolare critica, per un rilascio da

$$
\theta(0)=0,
$$

$$
\boxed{
\dot\theta_{\mathrm c}
=
2\sqrt{\frac{g}{\ell}}.
}
$$

Nella cella successiva raccoglieremo in un unico ritratto di fase
le diverse famiglie di traiettorie e la separatrice, collegando
la costruzione numerica alla descrizione analitica appena ottenuta.

````python
#@title Sintesi — Ritratto di fase e separatrice { display-mode: "form" }

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display, HTML, clear_output


# ============================================================
# CONTROLLO
# ============================================================

btn_sintesi = widgets.Button(
    description='Mostra ritratto di fase e separatrice',
    button_style='success',
    layout=widgets.Layout(
        width='360px',
        height='40px'
    )
)

out_sintesi = widgets.Output()

display(
    widgets.VBox([
        btn_sintesi,
        out_sintesi
    ])
)


# ============================================================
# FUNZIONE PRINCIPALE
# ============================================================

def _mostra_sintesi(_):

    with out_sintesi:

        clear_output(wait=True)

        # ----------------------------------------------------
        # PENDOLO DI RIFERIMENTO
        # ----------------------------------------------------

        lval = 1.00
        gval = 9.81

        omega_crit = 2.0 * np.sqrt(gval / lval)


        # ----------------------------------------------------
        # BANNER AZZURRINO
        # ----------------------------------------------------

        display(
            HTML(
                f"""
                <div style="
                    background:#eef6ff;
                    color:#16324f;
                    border-left:5px solid #4a90e2;
                    padding:12px 15px;
                    margin:6px 0 16px 0;
                    border-radius:6px;
                    line-height:1.55;
                ">
                    <b>Pendolo di riferimento</b><br>

                    ℓ = {lval:.2f} m
                    &nbsp;&nbsp;&nbsp;&nbsp;

                    g = {gval:.2f} m/s²
                    &nbsp;&nbsp;&nbsp;&nbsp;

                    <b>Velocità critica da θ(0)=0:</b>

                    ω<sub>c</sub> = 2√(g/ℓ)
                    = <b>{omega_crit:.2f} rad/s</b>
                </div>
                """
            )
        )


        # ----------------------------------------------------
        # DOMINIO DEL PIANO SROTOLATO
        # ----------------------------------------------------

        theta_min = -4.0 * np.pi
        theta_max =  4.0 * np.pi

        theta = np.linspace(
            theta_min,
            theta_max,
            8000
        )


        # ----------------------------------------------------
        # FIGURA
        # ----------------------------------------------------

        fig, ax = plt.subplots(
            figsize=(11.8, 6.6)
        )


        # ====================================================
        # 1. SEPARATRICE
        # ====================================================
        #
        # E = 2 m g l
        #
        # omega² = (2g/l)(1 + cos(theta))
        #
        # Entrambi i rami appartengono alla STESSA separatrice.
        # ====================================================

        omega_sep_sq = (
            2.0 * gval / lval
            * (1.0 + np.cos(theta))
        )

        omega_sep = np.sqrt(
            np.maximum(
                omega_sep_sq,
                0.0
            )
        )

        ax.plot(
            theta,
            omega_sep,
            color='tab:blue',
            linewidth=2.4,
            zorder=4,
            label=r'Separatrice: $E=2mg\ell$'
        )

        ax.plot(
            theta,
            -omega_sep,
            color='tab:blue',
            linewidth=2.4,
            zorder=4
        )


        # ====================================================
        # 2. OSCILLAZIONI
        # ====================================================
        #
        # Usiamo due livelli energetici inferiori
        # alla separatrice.
        #
        # Le condizioni equivalgono a partire da
        # theta(0)=0 con omega0 = 2.5 e 4.5 rad/s.
        # ====================================================

        omega0_osc_list = [
            2.5,
            4.5
        ]

        first_osc = True

        for omega0 in omega0_osc_list:

            omega_sq = (
                omega0**2
                -
                2.0 * gval / lval
                * (1.0 - np.cos(theta))
            )

            # Dove omega² < 0 lo stato non è
            # energeticamente accessibile.
            omega_branch = np.where(
                omega_sq >= 0.0,
                np.sqrt(
                    np.maximum(
                        omega_sq,
                        0.0
                    )
                ),
                np.nan
            )

            ax.plot(
                theta,
                omega_branch,
                color='tab:green',
                linewidth=1.5,
                alpha=0.85,
                zorder=3,
                label=(
                    'Oscillazione'
                    if first_osc
                    else None
                )
            )

            ax.plot(
                theta,
                -omega_branch,
                color='tab:green',
                linewidth=1.5,
                alpha=0.85,
                zorder=3
            )

            first_osc = False


        # ====================================================
        # 3. ROTAZIONI
        # ====================================================
        #
        # Energia superiore alla soglia.
        #
        # Nel piano srotolato la traiettoria NON termina
        # dopo 2π: continua verso theta crescente o decrescente.
        #
        # Usiamo omega(0)=7 rad/s.
        # ====================================================

        omega0_rot = 7.0

        omega_rot_sq = (
            omega0_rot**2
            -
            2.0 * gval / lval
            * (1.0 - np.cos(theta))
        )

        omega_rot = np.sqrt(
            np.maximum(
                omega_rot_sq,
                0.0
            )
        )

        # Rotazione nel verso positivo
        ax.plot(
            theta,
            omega_rot,
            color='tab:purple',
            linewidth=1.6,
            alpha=0.85,
            zorder=2,
            label='Rotazione'
        )

        # Rotazione nel verso negativo
        ax.plot(
            theta,
            -omega_rot,
            color='tab:purple',
            linewidth=1.6,
            alpha=0.85,
            zorder=2
        )


        # ====================================================
        # 4. CENTRI
        # ====================================================
        #
        # theta = 2 k pi
        # ====================================================

        theta_centri = np.array([
            -4.0 * np.pi,
            -2.0 * np.pi,
             0.0,
             2.0 * np.pi,
             4.0 * np.pi
        ])

        omega_centri = np.zeros_like(
            theta_centri
        )

        ax.scatter(
            theta_centri,
            omega_centri,
            color='tab:orange',
            marker='o',
            s=60,
            zorder=9,
            label='Centro'
        )


        # ====================================================
        # 5. SELLE
        # ====================================================
        #
        # theta = (2k+1) pi
        #
        # Tutte le selle visibili vengono mostrate
        # come X rosse.
        # ====================================================

        theta_selle = np.array([
            -3.0 * np.pi,
            -1.0 * np.pi,
             1.0 * np.pi,
             3.0 * np.pi
        ])

        omega_selle = np.zeros_like(
            theta_selle
        )

        ax.scatter(
            theta_selle,
            omega_selle,
            color='red',
            marker='x',
            s=105,
            linewidths=2.5,
            zorder=10,
            label='Sella'
        )


        # ====================================================
        # 6. ASSI
        # ====================================================

        ax.set_xlim(
            theta_min,
            theta_max
        )

        ax.set_ylim(
            -8.0,
             8.0
        )

        ax.set_xlabel(
            r'$\theta$ [rad]',
            fontsize=11
        )

        ax.set_ylabel(
            r'$\dot{\theta}$ [rad/s]',
            fontsize=11
        )

        ax.set_title(
            'Ritratto di fase del pendolo ideale — piano srotolato',
            fontsize=14,
            pad=14
        )


        # ----------------------------------------------------
        # TACCHE IN MULTIPLI DI pi
        # ----------------------------------------------------

        xticks = np.arange(
            -4,
             5
        ) * np.pi

        xtick_labels = [
            r'$-4\pi$',
            r'$-3\pi$',
            r'$-2\pi$',
            r'$-\pi$',
            r'$0$',
            r'$\pi$',
            r'$2\pi$',
            r'$3\pi$',
            r'$4\pi$'
        ]

        ax.set_xticks(
            xticks
        )

        ax.set_xticklabels(
            xtick_labels
        )

        ax.axhline(
            0.0,
            linewidth=0.8,
            alpha=0.45
        )

        ax.grid(
            True,
            alpha=0.22
        )


        # ====================================================
        # 7. LEGENDA ESTERNA AL PLOT
        # ====================================================

        ax.legend(
            loc='center left',
            bbox_to_anchor=(1.02, 0.5),
            frameon=True,
            fontsize=9,
            borderaxespad=0.0
        )

        fig.subplots_adjust(
            left=0.09,
            right=0.78,
            bottom=0.13,
            top=0.90
        )


        # ====================================================
        # 8. NOTA SUL PIANO SROTOLATO
        # ====================================================

        ax.text(
            0.02,
            0.025,
            r'Nel piano srotolato $\theta$ non viene riportato modulo $2\pi$.',
            transform=ax.transAxes,
            fontsize=9,
            va='bottom',
            ha='left',
            bbox=dict(
                boxstyle='round,pad=0.30',
                facecolor='white',
                alpha=0.82
            )
        )


        plt.show()


# ============================================================
# COLLEGAMENTO DEL PULSANTE
# ============================================================

btn_sintesi.on_click(
    _mostra_sintesi
)
````

## **Attività 9 — Dal filmato alla misura sperimentale di $\theta(t)$**

Finora il moto del pendolo è stato ottenuto a partire dalle equazioni. Ora percorriamo la strada opposta: **partiamo da un filmato reale e ricostruiamo il moto misurato**.

La procedura sarà:

1. carica un filmato del pendolo;
2. scegli il **fotogramma iniziale** dell'analisi;
3. scegli il **fotogramma finale** dell'analisi;
4. sul fotogramma iniziale clicca una volta sul **perno**;
5. clicca una volta sul **centro del dado** all'estremità del pendolo.

In questo modo puoi analizzare l'intero filmato oppure isolare, per esempio, **una sola oscillazione completa**, durante la quale l'ampiezza cambia poco.

Il perno viene considerato **fisso nelle coordinate dell'immagine** per tutto l'intervallo scelto:

$$
P=(x_P,y_P)=\text{costante}.
$$

Questa è l'ipotesi appropriata se la videocamera non si sposta durante la ripresa. Il codice non cerca quindi nuovamente il perno in ogni fotogramma: usa sempre il punto scelto dallo studente.

Il secondo click determina la direzione iniziale del pendolo e la distanza perno--dado in pixel. Questa distanza viene poi mantenuta costante durante il tracking.

Useremo la stessa convenzione adottata nel resto del notebook:

$$
\theta=0
$$

quando il pendolo è verticale verso il basso; $\theta>0$ quando il dado si trova a destra della verticale passante per il perno.

Se il perno ha coordinate immagine $(x_P,y_P)$ e il dado $(x_D,y_D)$, poiché nelle immagini l'asse verticale cresce verso il basso,

$$
\boxed{
\theta=\operatorname{atan2}(x_D-x_P,\,y_D-y_P).
}
$$

Per ricavare $\theta(t)$ **non è ancora necessario conoscere la lunghezza reale del pendolo**. La lunghezza $\ell$ in metri servirà successivamente, quando confronteremo il moto misurato con quello previsto dalle equazioni di Eulero--Lagrange.

> **Indicazione pratica.** Scegli un fotogramma iniziale in cui perno, asta e dado siano ben visibili. Se vuoi confrontare il moto con un pendolo non dissipativo a ampiezza quasi costante, seleziona come intervallo circa un solo periodo completo.


````python
#@title  Carica il video, scegli l'intervallo e definisci perno e dado { display-mode: "form" }

import os
import cv2
import base64
import uuid
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display, clear_output, HTML, Markdown, Math
from google.colab import files as colab_files
from google.colab import output as colab_output

# ============================================================
# 1. CARICAMENTO DEL VIDEO
# ============================================================
print("Seleziona dal tuo computer il filmato del pendolo.")
uploaded = colab_files.upload()

if not uploaded:
    raise RuntimeError("Nessun video caricato.")

exp_video_path = next(iter(uploaded.keys()))

cap = cv2.VideoCapture(exp_video_path)
if not cap.isOpened():
    raise RuntimeError("Impossibile aprire il video caricato.")

exp_fps = float(cap.get(cv2.CAP_PROP_FPS))
exp_nframes = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
exp_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
exp_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
cap.release()

if exp_fps <= 0 or exp_nframes < 3:
    raise RuntimeError("Il video non contiene metadati temporali sufficienti.")

print(f"Video: {exp_video_path}")
print(f"Fotogrammi: {exp_nframes}")
print(f"Frame rate: {exp_fps:.3f} frame/s")
print(f"Durata: {exp_nframes/exp_fps:.3f} s")
print(f"Risoluzione: {exp_width} × {exp_height} pixel")


def _exp_read_frame(frame_index):
    cap = cv2.VideoCapture(exp_video_path)
    cap.set(cv2.CAP_PROP_POS_FRAMES, int(frame_index))
    ok, frame = cap.read()
    cap.release()

    if not ok:
        raise RuntimeError(
            f"Impossibile leggere il fotogramma {frame_index}."
        )

    return frame


def _exp_preview_interval(start_index, end_index):
    if end_index <= start_index:
        raise ValueError(
            "Il fotogramma finale deve essere successivo a quello iniziale."
        )

    frame_a = _exp_read_frame(start_index)
    frame_b = _exp_read_frame(end_index)

    rgb_a = cv2.cvtColor(frame_a, cv2.COLOR_BGR2RGB)
    rgb_b = cv2.cvtColor(frame_b, cv2.COLOR_BGR2RGB)

    fig, axes = plt.subplots(1, 2, figsize=(9.5, 7.0))

    axes[0].imshow(rgb_a)
    axes[0].axis('off')
    axes[0].set_title(
        f"Inizio — frame {start_index}\n"
        f"t = {start_index/exp_fps:.3f} s"
    )

    axes[1].imshow(rgb_b)
    axes[1].axis('off')
    axes[1].set_title(
        f"Fine — frame {end_index}\n"
        f"t = {end_index/exp_fps:.3f} s"
    )

    plt.tight_layout()
    plt.show()

    duration = (end_index-start_index)/exp_fps

    display(Math(
        rf"\Delta t_{{\rm analisi}}"
        rf"={duration:.3f}\,\mathrm{{s}},\qquad "
        rf"N_{{\rm frame}}={end_index-start_index+1}"
    ))


# ============================================================
# 2. SCELTA DEL FOTOGRAMMA INIZIALE E FINALE
# ============================================================
_default_start = int(
    min(exp_nframes-2, max(0, round(0.05*exp_nframes)))
)

_default_end = exp_nframes-1

exp_start_w = widgets.IntSlider(
    value=_default_start,
    min=0,
    max=max(0, exp_nframes-2),
    step=1,
    description='frame iniziale:',
    continuous_update=False,
    readout=True,
    layout=widgets.Layout(width='470px')
)

exp_start_manual_w = widgets.BoundedIntText(
    value=_default_start,
    min=0,
    max=max(0, exp_nframes-2),
    step=1,
    description='da tastiera:',
    style={'description_width': '95px'},
    layout=widgets.Layout(width='220px')
)

exp_start_link = widgets.jslink(
    (exp_start_w, 'value'),
    (exp_start_manual_w, 'value')
)

exp_end_w = widgets.IntSlider(
    value=_default_end,
    min=1,
    max=max(1, exp_nframes-1),
    step=1,
    description='frame finale:',
    continuous_update=False,
    readout=True,
    layout=widgets.Layout(width='470px')
)

exp_end_manual_w = widgets.BoundedIntText(
    value=_default_end,
    min=1,
    max=max(1, exp_nframes-1),
    step=1,
    description='da tastiera:',
    style={'description_width': '95px'},
    layout=widgets.Layout(width='220px')
)

exp_end_link = widgets.jslink(
    (exp_end_w, 'value'),
    (exp_end_manual_w, 'value')
)


def _exp_keep_interval_valid(change=None):
    s = int(exp_start_w.value)
    e = int(exp_end_w.value)

    if e <= s:
        if change is not None and change.get('owner') is exp_start_w:
            exp_end_w.value = min(exp_nframes-1, s+1)
        else:
            exp_start_w.value = max(0, e-1)


exp_start_w.observe(_exp_keep_interval_valid, names='value')
exp_end_w.observe(_exp_keep_interval_valid, names='value')

exp_preview_btn = widgets.Button(
    description='Mostra intervallo scelto'
)

exp_click_btn = widgets.Button(
    description='Conferma intervallo e seleziona perno/dado'
)

exp_frame_out = widgets.Output()
exp_select_out = widgets.Output()


def _exp_show_selected_interval(_=None):
    with exp_frame_out:
        clear_output(wait=True)

        s = int(exp_start_w.value)
        e = int(exp_end_w.value)

        if e <= s:
            print(
                "Il fotogramma finale deve essere successivo "
                "a quello iniziale."
            )
            return

        _exp_preview_interval(s, e)


def _exp_click_two_points(frame):
    """
    Selezione interattiva dei due punti sul fotogramma iniziale.

    Usiamo una visualizzazione relativamente grande per ridurre
    l'errore di click sul perno.
    """
    h, w = frame.shape[:2]

    target_w = min(760, w)
    scale = target_w / w
    target_h = max(1, int(round(h * scale)))

    small = cv2.resize(
        frame,
        (target_w, target_h),
        interpolation=cv2.INTER_AREA
    )

    ok, buf = cv2.imencode(
        '.jpg',
        small,
        [int(cv2.IMWRITE_JPEG_QUALITY), 95]
    )

    if not ok:
        raise RuntimeError(
            "Impossibile preparare il fotogramma per la selezione."
        )

    data_uri = (
        'data:image/jpeg;base64,' +
        base64.b64encode(buf).decode('ascii')
    )

    uid = 'pend_click_' + uuid.uuid4().hex

    js = f"""
    new Promise((resolve) => {{
      const old = document.getElementById('{uid}');
      if (old) old.remove();

      const box = document.createElement('div');
      box.id = '{uid}';
      box.style.margin = '10px 0';
      box.style.padding = '10px';
      box.style.border = '1px solid #bbb';
      box.style.display = 'inline-block';
      box.style.background = 'white';

      const msg = document.createElement('div');
      msg.style.fontWeight = '600';
      msg.style.marginBottom = '8px';
      msg.textContent = '1) Clicca con precisione sul PERNO del pendolo';
      box.appendChild(msg);

      const canvas = document.createElement('canvas');
      const img = new Image();

      box.appendChild(canvas);
      document.body.appendChild(box);

      const pts = [];

      img.onload = () => {{
        canvas.width = img.width;
        canvas.height = img.height;
        canvas.style.maxWidth = '760px';
        canvas.style.width = '100%';
        canvas.style.height = 'auto';

        const ctx = canvas.getContext('2d');
        ctx.drawImage(img, 0, 0);

        const handler = (ev) => {{
          const rect = canvas.getBoundingClientRect();

          const x = (
            (ev.clientX - rect.left) *
            canvas.width / rect.width
          );

          const y = (
            (ev.clientY - rect.top) *
            canvas.height / rect.height
          );

          pts.push({{x:x, y:y}});

          ctx.beginPath();
          ctx.arc(x, y, 8, 0, 2*Math.PI);
          ctx.lineWidth = 3;
          ctx.strokeStyle = (
            pts.length === 1 ? '#1f77b4' : '#ff7f0e'
          );
          ctx.stroke();

          if (pts.length === 1) {{
            msg.textContent =
              '2) Ora clicca con precisione sul CENTRO DEL DADO';
          }} else {{
            canvas.removeEventListener('click', handler);
            msg.textContent = 'Selezione completata.';
            resolve(pts);
          }}
        }};

        canvas.addEventListener('click', handler);
      }};

      img.src = '{data_uri}';
    }})
    """

    pts = colab_output.eval_js(js)

    if pts is None or len(pts) != 2:
        raise RuntimeError(
            "Selezione dei punti non completata."
        )

    pivot_small = np.array([
        float(pts[0]['x']),
        float(pts[0]['y'])
    ])

    bob_small = np.array([
        float(pts[1]['x']),
        float(pts[1]['y'])
    ])

    return pivot_small / scale, bob_small / scale


def _exp_select_geometry(_):
    global exp_start_frame, exp_end_frame
    global exp_pivot_px, exp_bob0_px
    global exp_radius_px, exp_theta_click

    with exp_select_out:
        clear_output(wait=True)

        exp_start_frame = int(exp_start_w.value)
        exp_end_frame = int(exp_end_w.value)

        if exp_end_frame <= exp_start_frame:
            print(
                "Il fotogramma finale deve essere successivo "
                "a quello iniziale."
            )
            return

        frame = _exp_read_frame(exp_start_frame)

        display(Markdown(
            "**Sul fotogramma iniziale clicca prima il perno "
            "e poi il centro del dado.**"
        ))

        display(Markdown(
            "Il **perno rimarrà fisso esattamente in questa "
            "coordinata** per tutti i fotogrammi analizzati."
        ))

        try:
            pivot, bob = _exp_click_two_points(frame)
        except Exception as exc:
            print("Selezione non riuscita:", exc)
            return

        exp_pivot_px = np.asarray(
            pivot,
            dtype=float
        )

        exp_bob0_px = np.asarray(
            bob,
            dtype=float
        )

        exp_radius_px = float(
            np.linalg.norm(
                exp_bob0_px-exp_pivot_px
            )
        )

        exp_theta_click = float(
            np.arctan2(
                exp_bob0_px[0]-exp_pivot_px[0],
                exp_bob0_px[1]-exp_pivot_px[1]
            )
        )

        if exp_radius_px < 30:
            print(
                "I due punti risultano troppo vicini. "
                "Ripeti la selezione."
            )
            return

        rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        fig, ax = plt.subplots(
            figsize=(6.5, 9.0)
        )

        ax.imshow(rgb)

        ax.plot(
            exp_pivot_px[0],
            exp_pivot_px[1],
            'o',
            markersize=10,
            label='perno fisso'
        )

        ax.plot(
            exp_bob0_px[0],
            exp_bob0_px[1],
            'o',
            markersize=10,
            label='dado iniziale'
        )

        ax.plot(
            [exp_pivot_px[0], exp_bob0_px[0]],
            [exp_pivot_px[1], exp_bob0_px[1]],
            linewidth=2
        )

        ax.axis('off')
        ax.legend()
        ax.set_title(
            f"Geometria iniziale — frame {exp_start_frame}"
        )

        plt.show()

        display(Math(
            rf"P=({exp_pivot_px[0]:.1f},"
            rf"{exp_pivot_px[1]:.1f})\,\mathrm{{pixel}}"
        ))

        display(Math(
            rf"R_{{\rm immagine}}"
            rf"={exp_radius_px:.1f}\,\mathrm{{pixel}},\qquad "
            rf"\theta_{{\rm click}}"
            rf"={np.rad2deg(exp_theta_click):.2f}^\circ"
        ))

        display(Math(
            rf"n_{{\rm inizio}}={exp_start_frame},\qquad "
            rf"n_{{\rm fine}}={exp_end_frame},\qquad "
            rf"\Delta t_{{\rm analisi}}"
            rf"={(exp_end_frame-exp_start_frame)/exp_fps:.3f}\,"
            rf"\mathrm{{s}}"
        ))

        print(
            "Geometria e intervallo acquisiti. "
            "Ora esegui l'Attività 9.1."
        )


exp_preview_btn.on_click(
    _exp_show_selected_interval
)

exp_click_btn.on_click(
    _exp_select_geometry
)

display(widgets.VBox([
    widgets.HBox([
        exp_start_w,
        exp_start_manual_w
    ]),
    widgets.HBox([
        exp_end_w,
        exp_end_manual_w
    ]),
    widgets.HBox([
        exp_preview_btn,
        exp_click_btn
    ]),
    exp_frame_out,
    exp_select_out,
]))

_exp_show_selected_interval()
````

## **Attività 9.1 — Ricostruisci automaticamente $\theta(t)$, $\dot\theta(t)$, $\ddot\theta(t)$ e il diagramma degli stati**

Il perno scelto nell'Attività 9 viene mantenuto **fisso**:

$$
P=(x_P,y_P)=\text{costante}.
$$

Il problema di tracking riguarda quindi soltanto la direzione del pendolo e la posizione del dado.

Per rendere il riconoscimento più robusto, il codice usa contemporaneamente tre informazioni:

1. la presenza dell'**asta mobile** lungo una possibile direzione;
2. la presenza del **dado mobile** nella zona in cui dovrebbe trovarsi l'estremità del pendolo;
3. la continuità temporale del moto, usando gli ultimi due angoli per prevedere approssimativamente la posizione nel fotogramma successivo.

Asta e dado vengono distinti dagli elementi statici della scena mediante un **fondo temporale mediano** costruito sull'intervallo di fotogrammi scelto.

Il raggio immagine

$$
R_{\rm immagine}
$$

rimane quello misurato tra il click sul perno e il click iniziale sul dado. Il punto del dado ricostruito è quindi sempre proiettato sulla circonferenza di centro $P$ e raggio $R_{\rm immagine}$.

Per questa prima analisi manteniamo i dati **grezzi**. Una volta ottenuti i campioni

$$
\theta_i=\theta(t_i),
$$

calcoliamo direttamente

$$
\dot\theta_i
\simeq
\frac{\theta_{i+1}-\theta_{i-1}}{2\Delta t}
$$

e

$$
\ddot\theta_i
\simeq
\frac{\theta_{i+1}-2\theta_i+\theta_{i-1}}{\Delta t^2}.
$$

Non viene applicato alcun filtro in questa attività.

Prima dei grafici vengono mostrati il primo e l'ultimo fotogramma e alcuni dei fotogrammi in cui il tracking ha avuto la confidenza minore. **Controlla che il punto blu del perno resti nello stesso luogo e che il punto arancione segua realmente il dado.**


````python
#@title Attività 9.1 — Tracking ibrido asta + dado e dati sperimentali grezzi { display-mode: "form" }

import cv2
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output, Markdown, Math

if (
    'exp_pivot_px' not in globals()
    or 'exp_bob0_px' not in globals()
    or 'exp_end_frame' not in globals()
):
    raise RuntimeError(
        "Esegui prima l'Attività 9, scegli l'intervallo "
        "e seleziona perno e dado."
    )

_n_interval = int(
    exp_end_frame-exp_start_frame+1
)

exp_track_btn = widgets.Button(
    description='Avvia ricostruzione automatica'
)

exp_track_progress = widgets.IntProgress(
    value=0,
    min=0,
    max=max(1, _n_interval),
    description='tracking:',
    bar_style='',
    layout=widgets.Layout(width='520px')
)

exp_track_out = widgets.Output()


def _exp_build_temporal_background(
    video_path,
    start_frame,
    end_frame,
    n_samples=31
):
    """
    Fondo temporale mediano limitato all'intervallo selezionato.

    Gli elementi statici restano nel fondo. Asta e dado, che cambiano
    posizione durante l'oscillazione, tendono invece a essere rimossi
    dalla mediana temporale.
    """
    cap = cv2.VideoCapture(video_path)

    start_frame = int(start_frame)
    end_frame = int(end_frame)

    available = max(
        1,
        end_frame-start_frame+1
    )

    n_samples = int(min(
        max(9, n_samples),
        available
    ))

    indices = np.unique(
        np.linspace(
            start_frame,
            end_frame,
            n_samples
        ).astype(int)
    )

    gray_frames = []

    for idx in indices:
        cap.set(
            cv2.CAP_PROP_POS_FRAMES,
            int(idx)
        )

        ok, frame = cap.read()

        if not ok:
            continue

        gray_frames.append(
            cv2.cvtColor(
                frame,
                cv2.COLOR_BGR2GRAY
            )
        )

    cap.release()

    if len(gray_frames) < 3:
        raise RuntimeError(
            "Non riesco a costruire il fondo temporale del video."
        )

    return np.median(
        np.stack(gray_frames, axis=0),
        axis=0
    ).astype(np.uint8)


def _exp_prepare_tracking_frame(
    frame,
    background_gray
):
    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )

    hsv = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2HSV
    )

    saturation = hsv[:, :, 1]

    motion = cv2.absdiff(
        gray,
        background_gray
    )

    # Leggera regolarizzazione spaziale del solo indicatore di movimento.
    # Non modifica theta(t): serve soltanto a rendere più stabile
    # il riconoscimento dell'oggetto nel singolo fotogramma.
    motion = cv2.GaussianBlur(
        motion,
        (5, 5),
        0
    )

    return gray, saturation, motion


def _exp_sample_disk_mean(
    image,
    x0,
    y0,
    radius
):
    """
    Media dei pixel in un piccolo disco.
    """
    h, w = image.shape[:2]

    r = int(max(2, round(radius)))

    x_min = max(0, int(round(x0))-r)
    x_max = min(w, int(round(x0))+r+1)

    y_min = max(0, int(round(y0))-r)
    y_max = min(h, int(round(y0))+r+1)

    if x_max <= x_min or y_max <= y_min:
        return 0.0

    yy, xx = np.mgrid[
        y_min:y_max,
        x_min:x_max
    ]

    mask = (
        (xx-x0)**2 +
        (yy-y0)**2
        <= r**2
    )

    if np.count_nonzero(mask) < 4:
        return 0.0

    return float(
        np.mean(
            image[y_min:y_max, x_min:x_max][mask]
        )
    )


def _exp_score_angle(
    prepared,
    pivot,
    radius_px,
    theta
):
    """
    Punteggio ibrido:
      - evidenza dell'asta lungo la direzione candidata;
      - evidenza del dado nell'intorno dell'estremità candidata.

    Il secondo termine è particolarmente importante quando l'asta passa
    vicino a linee statiche dello sfondo.
    """
    gray, saturation, motion = prepared

    h, w = gray.shape[:2]

    s = np.sin(theta)
    c = np.cos(theta)

    # ------------------------------------------------------------
    # A) EVIDENZA DELL'ASTA
    # ------------------------------------------------------------
    r_min = max(
        18.0,
        0.08*radius_px
    )

    r_max = max(
        r_min+10.0,
        0.84*radius_px
    )

    rr = np.arange(
        r_min,
        r_max,
        2.5
    )

    base_x = pivot[0] + rr*s
    base_y = pivot[1] + rr*c

    # Campioniamo una striscia trasversale invece di una sola linea.
    offsets = np.array([
        -5., -3., -1., 0.,
        1., 3., 5.
    ])

    xx = np.rint(
        base_x[:, None] +
        offsets[None, :]*c
    ).astype(int)

    yy = np.rint(
        base_y[:, None] -
        offsets[None, :]*s
    ).astype(int)

    valid = (
        (xx >= 0) & (xx < w) &
        (yy >= 0) & (yy < h)
    )

    if np.count_nonzero(valid) < 30:
        return -np.inf

    evidence = np.zeros_like(
        xx,
        dtype=float
    )

    g = gray[
        yy[valid],
        xx[valid]
    ].astype(float)

    sat = saturation[
        yy[valid],
        xx[valid]
    ].astype(float)

    mov = motion[
        yy[valid],
        xx[valid]
    ].astype(float)

    motion_evidence = np.clip(
        (mov-5.0)/65.0,
        0.0,
        1.0
    )

    darkness = np.clip(
        (160.0-g)/110.0,
        0.0,
        1.0
    )

    neutrality = np.clip(
        (120.0-sat)/120.0,
        0.0,
        1.0
    )

    rod_evidence = (
        darkness*neutrality
    )

    evidence[valid] = (
        0.78*motion_evidence +
        0.22*rod_evidence
    )

    along_rod = np.max(
        evidence,
        axis=1
    )

    rod_score = float(
        0.72*np.mean(along_rod) +
        0.28*np.mean(
            along_rod > 0.30
        )
    )

    # ------------------------------------------------------------
    # B) EVIDENZA DEL DADO ALL'ESTREMITÀ
    # ------------------------------------------------------------
    bob_x = (
        pivot[0] +
        radius_px*s
    )

    bob_y = (
        pivot[1] +
        radius_px*c
    )

    bob_radius = float(
        np.clip(
            0.075*radius_px,
            7.0,
            18.0
        )
    )

    bob_motion = (
        _exp_sample_disk_mean(
            motion,
            bob_x,
            bob_y,
            bob_radius
        )
        / 255.0
    )

    bob_gray = (
        _exp_sample_disk_mean(
            gray,
            bob_x,
            bob_y,
            bob_radius
        )
    )

    bob_darkness = float(
        np.clip(
            (185.0-bob_gray)/130.0,
            0.0,
            1.0
        )
    )

    bob_score = float(
        0.82*np.clip(
            bob_motion/0.24,
            0.0,
            1.0
        )
        +
        0.18*bob_darkness
    )

    # Il dado pesa molto: evita di agganciare aste o supporti statici
    # che casualmente hanno la direzione corretta.
    return float(
        0.54*rod_score +
        0.46*bob_score
    )


def _exp_search_angle(
    prepared,
    pivot,
    radius_px,
    center_theta,
    span_deg,
    coarse_step_deg=0.45,
    fine_step_deg=0.06,
    prediction_penalty=0.025
):
    center_deg = np.rad2deg(
        center_theta
    )

    candidates = np.deg2rad(
        np.arange(
            center_deg-span_deg,
            center_deg+span_deg+
            0.5*coarse_step_deg,
            coarse_step_deg
        )
    )

    scores = np.array([
        _exp_score_angle(
            prepared,
            pivot,
            radius_px,
            th
        )
        for th in candidates
    ])

    delta = np.angle(
        np.exp(
            1j*(
                candidates-center_theta
            )
        )
    )

    penalized = (
        scores
        -
        prediction_penalty*(
            delta /
            np.deg2rad(
                max(span_deg, 1e-6)
            )
        )**2
    )

    j = int(
        np.argmax(penalized)
    )

    theta_coarse = float(
        candidates[j]
    )

    fine = np.deg2rad(
        np.arange(
            np.rad2deg(theta_coarse)-0.55,
            np.rad2deg(theta_coarse)+0.55+
            0.5*fine_step_deg,
            fine_step_deg
        )
    )

    fine_scores = np.array([
        _exp_score_angle(
            prepared,
            pivot,
            radius_px,
            th
        )
        for th in fine
    ])

    fine_delta = np.angle(
        np.exp(
            1j*(
                fine-center_theta
            )
        )
    )

    fine_penalized = (
        fine_scores
        -
        prediction_penalty*(
            fine_delta /
            np.deg2rad(
                max(span_deg, 1e-6)
            )
        )**2
    )

    k = int(
        np.argmax(fine_penalized)
    )

    return (
        float(fine[k]),
        float(fine_scores[k])
    )


def _exp_best_angle(
    prepared,
    pivot,
    radius_px,
    previous_theta,
    previous_previous_theta,
    fps
):
    """
    Predizione a velocità angolare quasi costante fra tre frame.

    Se la confidenza locale è bassa, effettua una ricerca di recupero
    più ampia, sempre vincolata attorno alla direzione prevista.
    """
    if previous_previous_theta is None:
        predicted_theta = float(
            previous_theta
        )
    else:
        dtheta = np.angle(
            np.exp(
                1j*(
                    previous_theta -
                    previous_previous_theta
                )
            )
        )

        max_predicted_step = np.deg2rad(
            float(
                np.clip(
                    1050.0/max(fps, 1.0),
                    12.0,
                    20.0
                )
            )
        )

        dtheta = float(
            np.clip(
                dtheta,
                -max_predicted_step,
                +max_predicted_step
            )
        )

        predicted_theta = float(
            previous_theta+dtheta
        )

    local_span_deg = float(
        np.clip(
            1000.0/max(fps, 1.0),
            14.0,
            26.0
        )
    )

    theta_hat, score = (
        _exp_search_angle(
            prepared,
            pivot,
            radius_px,
            predicted_theta,
            local_span_deg
        )
    )

    # Recupero automatico se la combinazione asta+dado è poco convincente.
    if score < 0.24:
        theta_wide, score_wide = (
            _exp_search_angle(
                prepared,
                pivot,
                radius_px,
                predicted_theta,
                58.0,
                coarse_step_deg=0.9,
                fine_step_deg=0.08,
                prediction_penalty=0.012
            )
        )

        if score_wide > score:
            theta_hat = theta_wide
            score = score_wide

    return (
        float(theta_hat),
        float(score)
    )


def _exp_raw_derivatives(
    theta_values,
    dt
):
    theta_values = np.asarray(
        theta_values,
        dtype=float
    )

    n = len(theta_values)

    if n < 3:
        raise RuntimeError(
            "Servono almeno tre fotogrammi "
            "per calcolare le derivate."
        )

    omega = np.empty_like(
        theta_values
    )

    alpha = np.empty_like(
        theta_values
    )

    omega[1:-1] = (
        theta_values[2:] -
        theta_values[:-2]
    )/(2.0*dt)

    omega[0] = (
        theta_values[1] -
        theta_values[0]
    )/dt

    omega[-1] = (
        theta_values[-1] -
        theta_values[-2]
    )/dt

    alpha[1:-1] = (
        theta_values[2:]
        -
        2.0*theta_values[1:-1]
        +
        theta_values[:-2]
    )/(dt**2)

    alpha[0] = (
        theta_values[2]
        -
        2.0*theta_values[1]
        +
        theta_values[0]
    )/(dt**2)

    alpha[-1] = (
        theta_values[-1]
        -
        2.0*theta_values[-2]
        +
        theta_values[-3]
    )/(dt**2)

    return omega, alpha


def _exp_pick_check_indices(
    scores,
    fps,
    n_panels=6
):
    """
    Mostra sempre inizio e fine e aggiunge i fotogrammi
    di confidenza più bassa, evitando pannelli quasi duplicati.
    """
    scores = np.asarray(
        scores,
        dtype=float
    )

    n = len(scores)

    chosen = [0]

    if n > 1:
        chosen.append(n-1)

    min_sep = max(
        1,
        int(round(0.30*fps))
    )

    order = np.argsort(scores)

    for idx in order:
        idx = int(idx)

        if all(
            abs(idx-j) >= min_sep
            for j in chosen
        ):
            chosen.append(idx)

        if len(chosen) >= n_panels:
            break

    if len(chosen) < n_panels:
        for idx in np.linspace(
            0,
            n-1,
            n_panels
        ).astype(int):
            idx = int(idx)

            if idx not in chosen:
                chosen.append(idx)

            if len(chosen) >= n_panels:
                break

    return np.array(
        sorted(chosen[:n_panels]),
        dtype=int
    )


def _exp_show_tracking_check(
    sample_indices
):
    cap = cv2.VideoCapture(
        exp_video_path
    )

    fig, axes = plt.subplots(
        2,
        3,
        figsize=(11, 9)
    )

    axes = np.asarray(
        axes
    ).ravel()

    for ax, local_i in zip(
        axes,
        sample_indices
    ):
        absolute_frame = (
            exp_start_frame +
            int(local_i)
        )

        cap.set(
            cv2.CAP_PROP_POS_FRAMES,
            absolute_frame
        )

        ok, frame = cap.read()

        if not ok:
            ax.axis('off')
            continue

        th = exp_theta_raw[
            int(local_i)
        ]

        bob = (
            exp_pivot_px
            +
            exp_radius_px*np.array([
                np.sin(th),
                np.cos(th)
            ])
        )

        rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        ax.imshow(rgb)

        # Il perno è deliberatamente lo STESSO punto in tutti i frame.
        ax.plot(
            exp_pivot_px[0],
            exp_pivot_px[1],
            'o',
            markersize=6,
            label='perno fisso'
        )

        ax.plot(
            bob[0],
            bob[1],
            'o',
            markersize=6,
            label='dado ricostruito'
        )

        ax.plot(
            [
                exp_pivot_px[0],
                bob[0]
            ],
            [
                exp_pivot_px[1],
                bob[1]
            ],
            linewidth=1.6
        )

        ax.set_title(
            rf"$t={exp_t_raw[int(local_i)]:.2f}$ s, "
            rf"$\theta={np.rad2deg(th):.1f}^\circ$"
            "\n"
            rf"conf.={exp_tracking_score[int(local_i)]:.2f}"
        )

        ax.axis('off')

    cap.release()

    fig.suptitle(
        'Controllo tracking: inizio, fine e fotogrammi più difficili'
    )

    plt.tight_layout()
    plt.show()


def _exp_run_tracking(_):
    global exp_t_raw
    global exp_theta_raw
    global exp_omega_raw
    global exp_alpha_raw
    global exp_bob_xy_raw
    global exp_tracking_score
    global exp_background_gray

    with exp_track_out:
        clear_output(wait=True)
        exp_track_progress.value = 0

        print(
            "Costruzione del fondo temporale mediano "
            "nell'intervallo selezionato..."
        )

        exp_background_gray = (
            _exp_build_temporal_background(
                exp_video_path,
                exp_start_frame,
                exp_end_frame,
                n_samples=31
            )
        )

        print(
            "Fondo costruito. "
            "Avvio del tracking ibrido asta + dado..."
        )

        cap = cv2.VideoCapture(
            exp_video_path
        )

        cap.set(
            cv2.CAP_PROP_POS_FRAMES,
            int(exp_start_frame)
        )

        theta_list = []
        score_list = []

        previous_theta = float(
            exp_theta_click
        )

        previous_previous_theta = None

        processed = 0
        n_to_process = int(
            exp_end_frame -
            exp_start_frame + 1
        )

        while processed < n_to_process:
            ok, frame = cap.read()

            if not ok:
                break

            prepared = (
                _exp_prepare_tracking_frame(
                    frame,
                    exp_background_gray
                )
            )

            if processed == 0:
                # Il primo theta è quello definito direttamente dai click.
                theta_hat = float(
                    exp_theta_click
                )

                score = _exp_score_angle(
                    prepared,
                    exp_pivot_px,
                    exp_radius_px,
                    theta_hat
                )
            else:
                theta_hat, score = (
                    _exp_best_angle(
                        prepared,
                        exp_pivot_px,
                        exp_radius_px,
                        previous_theta,
                        previous_previous_theta,
                        exp_fps
                    )
                )

            theta_list.append(
                theta_hat
            )

            score_list.append(
                score
            )

            previous_previous_theta = (
                previous_theta
            )

            previous_theta = (
                theta_hat
            )

            processed += 1

            if processed % 5 == 0:
                exp_track_progress.value = min(
                    processed,
                    exp_track_progress.max
                )

        cap.release()

        exp_track_progress.value = min(
            processed,
            exp_track_progress.max
        )

        if len(theta_list) < 3:
            raise RuntimeError(
                "Tracking insufficiente: "
                "sono stati ricostruiti meno di tre fotogrammi."
            )

        exp_theta_raw = np.unwrap(
            np.asarray(
                theta_list,
                dtype=float
            )
        )

        exp_tracking_score = np.asarray(
            score_list,
            dtype=float
        )

        dt = 1.0/exp_fps

        exp_t_raw = (
            np.arange(
                len(exp_theta_raw),
                dtype=float
            )*dt
        )

        (
            exp_omega_raw,
            exp_alpha_raw
        ) = _exp_raw_derivatives(
            exp_theta_raw,
            dt
        )

        exp_bob_xy_raw = np.column_stack([
            exp_pivot_px[0]
            + exp_radius_px*np.sin(
                exp_theta_raw
            ),
            exp_pivot_px[1]
            + exp_radius_px*np.cos(
                exp_theta_raw
            )
        ])

        display(
            Markdown(
                "### Controllo del tracking"
            )
        )

        print(
            f"Fotogrammi analizzati: "
            f"{len(exp_theta_raw)}"
        )

        print(
            f"Frame assoluti: "
            f"{exp_start_frame} — "
            f"{exp_start_frame+len(exp_theta_raw)-1}"
        )

        print(
            f"Intervallo temporale analizzato: "
            f"0 — {exp_t_raw[-1]:.3f} s"
        )

        print(
            "Perno fisso alle coordinate: "
            f"({exp_pivot_px[0]:.1f}, "
            f"{exp_pivot_px[1]:.1f}) pixel"
        )

        print(
            f"Punteggio tracking mediano: "
            f"{np.median(exp_tracking_score):.3f}"
        )

        low_fraction = np.mean(
            exp_tracking_score < 0.24
        )

        print(
            f"Fotogrammi a bassa confidenza: "
            f"{100.0*low_fraction:.1f}%"
        )

        if (
            np.median(
                exp_tracking_score
            ) < 0.24
            or low_fraction > 0.15
        ):
            print(
                "ATTENZIONE: una parte del tracking "
                "appare poco affidabile. "
                "Controlla soprattutto i fotogrammi "
                "a bassa confidenza mostrati sotto."
            )

        sample_indices = (
            _exp_pick_check_indices(
                exp_tracking_score,
                exp_fps,
                n_panels=6
            )
        )

        _exp_show_tracking_check(
            sample_indices
        )

        display(
            Markdown(
                "### Dati sperimentali grezzi"
            )
        )

        fig, axes = plt.subplots(
            2,
            2,
            figsize=(12, 8)
        )

        axes[0,0].plot(
            exp_t_raw,
            exp_theta_raw,
            linewidth=1.2
        )

        axes[0,0].set_xlabel(
            r'$t$ [s]'
        )

        axes[0,0].set_ylabel(
            r'$\theta$ [rad]'
        )

        axes[0,0].set_title(
            r'$\theta_{\rm exp}(t)$ — dato grezzo'
        )

        axes[0,0].grid(
            True,
            alpha=0.25
        )

        axes[0,1].plot(
            exp_t_raw,
            exp_omega_raw,
            linewidth=1.0
        )

        axes[0,1].set_xlabel(
            r'$t$ [s]'
        )

        axes[0,1].set_ylabel(
            r'$\dot\theta$ [rad/s]'
        )

        axes[0,1].set_title(
            r'$\dot\theta_{\rm exp}(t)$ — differenza finita'
        )

        axes[0,1].grid(
            True,
            alpha=0.25
        )

        axes[1,0].plot(
            exp_t_raw,
            exp_alpha_raw,
            linewidth=0.9
        )

        axes[1,0].set_xlabel(
            r'$t$ [s]'
        )

        axes[1,0].set_ylabel(
            r'$\ddot\theta$ [rad/s$^2$]'
        )

        axes[1,0].set_title(
            r'$\ddot\theta_{\rm exp}(t)$ — dato grezzo'
        )

        axes[1,0].grid(
            True,
            alpha=0.25
        )

        axes[1,1].plot(
            exp_theta_raw,
            exp_omega_raw,
            linewidth=1.0
        )

        axes[1,1].set_xlabel(
            r'$\theta$ [rad]'
        )

        axes[1,1].set_ylabel(
            r'$\dot\theta$ [rad/s]'
        )

        axes[1,1].set_title(
            'Diagramma degli stati sperimentale — grezzo'
        )

        axes[1,1].grid(
            True,
            alpha=0.25
        )

        plt.tight_layout()
        plt.show()

        display(
            Math(
                rf"\theta_{{\rm exp}}(0)="
                rf"{exp_theta_raw[0]:.5f}\,"
                rf"\mathrm{{rad}}"
                rf"={np.rad2deg(exp_theta_raw[0]):.2f}^\circ"
            )
        )

        display(
            Math(
                rf"\dot{{\theta}}_{{\rm exp}}(0)="
                rf"{exp_omega_raw[0]:.5f}\,"
                rf"\mathrm{{rad\,s^{{-1}}}}"
            )
        )

        display(
            Markdown(
                "**Osservazione:** i dati mostrati qui "
                "non sono filtrati. La derivazione numerica "
                "amplifica anche piccole irregolarità del tracking, "
                "soprattutto in $\\ddot\\theta(t)$. "
                "Nell'Attività 9.2 vedremo separatamente "
                "la ricostruzione filtrata."
            )
        )


exp_track_btn.on_click(
    _exp_run_tracking
)

display(
    widgets.VBox([
        exp_track_btn,
        exp_track_progress,
        exp_track_out
    ])
)
````

<details>
<summary>
<font size="4"><b>Approfondimento facoltativo — Perché usiamo le differenze finite centrate?</b></font>
&nbsp;&nbsp;<i>(clicca per aprire)</i>
</summary>

<br>

Supponiamo di conoscere $\theta(t)$ soltanto negli istanti equispaziati

$$
t_i=i\,\Delta t.
$$

Indichiamo

$$
\theta_i=\theta(t_i).
$$

La formula più immediata per approssimare la derivata sarebbe la differenza in avanti

$$
\frac{\theta_{i+1}-\theta_i}{\Delta t}.
$$

Per capire la sua accuratezza sviluppiamo $\theta_{i+1}$ in serie di Taylor attorno a $t_i$:

$$
\theta_{i+1}
=
\theta_i
+
\dot\theta_i\Delta t
+
\frac{1}{2}\ddot\theta_i\Delta t^2
+
O(\Delta t^3).
$$

Dividendo per $\Delta t$,

$$
\frac{\theta_{i+1}-\theta_i}{\Delta t}
=
\dot\theta_i
+
\frac{1}{2}\ddot\theta_i\Delta t
+
O(\Delta t^2).
$$

L'errore principale è quindi proporzionale a $\Delta t$:

$$
\boxed{
\frac{\theta_{i+1}-\theta_i}{\Delta t}
=
\dot\theta_i+O(\Delta t).
}
$$

Consideriamo ora anche il punto precedente:

$$
\theta_{i-1}
=
\theta_i
-
\dot\theta_i\Delta t
+
\frac{1}{2}\ddot\theta_i\Delta t^2
-
\frac{1}{6}\theta_i^{(3)}\Delta t^3
+
O(\Delta t^4).
$$

Per il punto successivo,

$$
\theta_{i+1}
=
\theta_i
+
\dot\theta_i\Delta t
+
\frac{1}{2}\ddot\theta_i\Delta t^2
+
\frac{1}{6}\theta_i^{(3)}\Delta t^3
+
O(\Delta t^4).
$$

Sottraendo le due espressioni, i termini pari si cancellano:

$$
\theta_{i+1}-\theta_{i-1}
=
2\dot\theta_i\Delta t
+
\frac{1}{3}\theta_i^{(3)}\Delta t^3
+
O(\Delta t^5).
$$

Pertanto

$$
\boxed{
\dot\theta_i
\simeq
\frac{\theta_{i+1}-\theta_{i-1}}{2\Delta t}
}
$$

con errore

$$
\boxed{O(\Delta t^2).}
$$

La differenza centrata è quindi, a parità di passo temporale, più accurata della differenza in avanti.

Per la seconda derivata sommiamo invece i due sviluppi:

$$
\theta_{i+1}+\theta_{i-1}
=
2\theta_i
+
\ddot\theta_i\Delta t^2
+
O(\Delta t^4).
$$

Da cui

$$
\boxed{
\ddot\theta_i
\simeq
\frac{\theta_{i+1}-2\theta_i+\theta_{i-1}}{\Delta t^2}
}
$$

ancora con errore di ordine

$$
O(\Delta t^2).
$$

Queste formule migliorano l'**errore di discretizzazione**, ma non eliminano l'errore sperimentale presente nei valori misurati di $\theta_i$. Anzi, la derivazione numerica tende a rendere più visibili le piccole irregolarità dei dati: questo effetto diventa particolarmente evidente nella seconda derivata.

In questa attività lasciamo volutamente i dati grezzi. Un eventuale trattamento del rumore verrà considerato solo in una fase successiva.

<br>

</details>

## **Attività 9.2 — Riduci il rumore nelle derivate e osserva la struttura dell'accelerazione**

Nell'Attività 9.1 abbiamo volutamente derivato direttamente i dati grezzi.

La seconda derivata amplifica fortemente anche piccoli errori nella misura di $\theta(t)$. Per questo gli spike di $\ddot\theta(t)$ possono dominare la scala verticale e nascondere la struttura fisica del segnale.

Ora applichiamo un filtro **Savitzky--Golay**: in una piccola finestra temporale si approssima localmente $\theta(t)$ con un polinomio e si derivano analiticamente i polinomi locali.

Lo studente può modificare:

- il numero di fotogrammi della finestra, che deve essere dispari;
- l'ordine del polinomio locale.

Il dato grezzo non viene eliminato: conserviamo separatamente

$$
\theta_{\rm raw}(t),
\qquad
\dot\theta_{\rm raw}(t),
\qquad
\ddot\theta_{\rm raw}(t)
$$

e

$$
\theta_{\rm filt}(t),
\qquad
\dot\theta_{\rm filt}(t),
\qquad
\ddot\theta_{\rm filt}(t).
$$

Nel grafico dell'accelerazione la scala verticale viene scelta usando il segnale filtrato: gli eventuali spike grezzi possono quindi uscire dalla finestra visibile, ma **non vengono cancellati dai dati**.

Per oscillazioni sufficientemente ampie, osserva con attenzione la forma di $\ddot\theta(t)$ vicino ai massimi e ai minimi: il pendolo non è armonico e possono comparire i caratteristici **doppi picchi**.


````python
#@title Attività 9.2 — Filtraggio delle derivate sperimentali { display-mode: "form" }

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from scipy.signal import savgol_filter
from IPython.display import display, clear_output, Markdown, Math

if 'exp_theta_raw' not in globals():
    raise RuntimeError(
        "Esegui prima l'Attività 9.1 e ricostruisci i dati sperimentali grezzi."
    )

if len(exp_theta_raw) < 7:
    raise RuntimeError(
        "Servono almeno sette campioni per applicare questo filtro."
    )

# ============================================================
# CONTROLLI: slider + inserimento da tastiera
# ============================================================
_filt_slider_layout = widgets.Layout(width='470px')
_filt_text_layout = widgets.Layout(width='220px')
_filt_text_style = {'description_width': '95px'}

_max_window = min(
    61,
    len(exp_theta_raw) if len(exp_theta_raw) % 2 == 1
    else len(exp_theta_raw)-1
)

_default_window = min(15, _max_window)
if _default_window % 2 == 0:
    _default_window -= 1

window_w = widgets.IntSlider(
    value=_default_window,
    min=5,
    max=_max_window,
    step=2,
    description='finestra [frame]:',
    continuous_update=True,
    readout=True,
    layout=_filt_slider_layout
)

window_manual_w = widgets.BoundedIntText(
    value=_default_window,
    min=5,
    max=_max_window,
    step=2,
    description='da tastiera:',
    style=_filt_text_style,
    layout=_filt_text_layout
)

window_link = widgets.jslink(
    (window_w, 'value'),
    (window_manual_w, 'value')
)

poly_w = widgets.IntSlider(
    value=3,
    min=2,
    max=5,
    step=1,
    description='ordine polinomio:',
    continuous_update=True,
    readout=True,
    layout=_filt_slider_layout
)

poly_manual_w = widgets.BoundedIntText(
    value=3,
    min=2,
    max=5,
    step=1,
    description='da tastiera:',
    style=_filt_text_style,
    layout=_filt_text_layout
)

poly_link = widgets.jslink(
    (poly_w, 'value'),
    (poly_manual_w, 'value')
)

filter_btn = widgets.Button(
    description='Applica filtro e calcola derivate'
)

filter_out = widgets.Output()


def _apply_experimental_filter(_):
    global exp_theta_filt, exp_omega_filt, exp_alpha_filt
    global exp_filter_window, exp_filter_polyorder

    with filter_out:
        clear_output(wait=True)

        window = int(window_w.value)
        polyorder = int(poly_w.value)

        if window % 2 == 0:
            print(
                "La finestra deve contenere un numero dispari di fotogrammi. "
                "Scegli, per esempio, 11, 15, 21, ..."
            )
            return

        if window > len(exp_theta_raw):
            print("La finestra è più lunga della serie sperimentale.")
            return

        if polyorder >= window:
            print(
                "L'ordine del polinomio deve essere minore "
                "del numero di punti della finestra."
            )
            return

        dt = 1.0/exp_fps

        exp_filter_window = window
        exp_filter_polyorder = polyorder

        # Fit polinomiale locale e sue derivate.
        exp_theta_filt = savgol_filter(
            exp_theta_raw,
            window_length=window,
            polyorder=polyorder,
            deriv=0,
            delta=dt,
            mode='interp'
        )

        exp_omega_filt = savgol_filter(
            exp_theta_raw,
            window_length=window,
            polyorder=polyorder,
            deriv=1,
            delta=dt,
            mode='interp'
        )

        exp_alpha_filt = savgol_filter(
            exp_theta_raw,
            window_length=window,
            polyorder=polyorder,
            deriv=2,
            delta=dt,
            mode='interp'
        )

        display(Markdown("### Parametri del filtraggio"))

        display(Math(
            rf"N_{{\rm finestra}}={window},\qquad "
            rf"p={polyorder},\qquad "
            rf"\Delta t={dt:.5f}\,\mathrm{{s}}"
        ))

        display(Math(
            rf"T_{{\rm finestra}}"
            rf"=N_{{\rm finestra}}\Delta t"
            rf"\simeq {window*dt:.3f}\,\mathrm{{s}}"
        ))

        fig, axes = plt.subplots(2, 2, figsize=(12, 8))

        # theta
        axes[0,0].plot(
            exp_t_raw,
            exp_theta_raw,
            linewidth=0.8,
            alpha=0.35,
            label='grezzo'
        )
        axes[0,0].plot(
            exp_t_raw,
            exp_theta_filt,
            linewidth=1.6,
            label='filtrato'
        )
        axes[0,0].set_xlabel(r'$t$ [s]')
        axes[0,0].set_ylabel(r'$\theta$ [rad]')
        axes[0,0].set_title(r'$\theta(t)$: grezzo e filtrato')
        axes[0,0].grid(True, alpha=0.25)
        axes[0,0].legend()

        # omega
        axes[0,1].plot(
            exp_t_raw,
            exp_omega_raw,
            linewidth=0.7,
            alpha=0.30,
            label='differenza finita grezza'
        )
        axes[0,1].plot(
            exp_t_raw,
            exp_omega_filt,
            linewidth=1.5,
            label='derivata filtrata'
        )
        axes[0,1].set_xlabel(r'$t$ [s]')
        axes[0,1].set_ylabel(r'$\dot\theta$ [rad/s]')
        axes[0,1].set_title(r'$\dot\theta(t)$')
        axes[0,1].grid(True, alpha=0.25)
        axes[0,1].legend()

        # alpha: scala determinata dal segnale filtrato
        axes[1,0].plot(
            exp_t_raw,
            exp_alpha_raw,
            linewidth=0.55,
            alpha=0.22,
            label='grezzo'
        )
        axes[1,0].plot(
            exp_t_raw,
            exp_alpha_filt,
            linewidth=1.5,
            label='filtrato'
        )

        edge = max(2, window//2)
        central = exp_alpha_filt[
            edge:-edge if edge < len(exp_alpha_filt) else None
        ]

        if len(central) < 5:
            central = exp_alpha_filt

        scale = np.nanpercentile(
            np.abs(central),
            99.5
        )

        if np.isfinite(scale) and scale > 0:
            axes[1,0].set_ylim(
                -1.15*scale,
                +1.15*scale
            )

        axes[1,0].set_xlabel(r'$t$ [s]')
        axes[1,0].set_ylabel(r'$\ddot\theta$ [rad/s$^2$]')
        axes[1,0].set_title(
            r'$\ddot\theta(t)$ — scala centrata sul segnale filtrato'
        )
        axes[1,0].grid(True, alpha=0.25)
        axes[1,0].legend()

        # diagramma degli stati
        axes[1,1].plot(
            exp_theta_raw,
            exp_omega_raw,
            linewidth=0.7,
            alpha=0.25,
            label='grezzo'
        )
        axes[1,1].plot(
            exp_theta_filt,
            exp_omega_filt,
            linewidth=1.4,
            label='filtrato'
        )
        axes[1,1].set_xlabel(r'$\theta$ [rad]')
        axes[1,1].set_ylabel(r'$\dot\theta$ [rad/s]')
        axes[1,1].set_title(
            'Diagramma degli stati sperimentale'
        )
        axes[1,1].grid(True, alpha=0.25)
        axes[1,1].legend()

        plt.tight_layout()
        plt.show()

        display(Markdown(
            "**Nota sulla scala di $\\ddot\\theta$:** "
            "il grafico è ingrandito sulla struttura del segnale filtrato. "
            "Gli spike del dato grezzo che eccedono questa scala restano "
            "nei dati, ma vengono intenzionalmente lasciati fuori dalla "
            "finestra verticale del grafico."
        ))

        display(Math(
            rf"\theta_{{\rm filt}}(0)"
            rf"={exp_theta_filt[0]:.5f}\,\mathrm{{rad}},\qquad "
            rf"\dot{{\theta}}_{{\rm filt}}(0)"
            rf"={exp_omega_filt[0]:.5f}\,\mathrm{{rad\,s^{{-1}}}}"
        ))


filter_btn.on_click(_apply_experimental_filter)

display(widgets.VBox([
    widgets.HBox([window_w, window_manual_w]),
    widgets.HBox([poly_w, poly_manual_w]),
    filter_btn,
    filter_out
]))
````

## **Attività 9.3 — Confronta il moto sperimentale con la soluzione delle equazioni di Eulero--Lagrange**

Ora confrontiamo il moto ricostruito dal video con il modello del pendolo semplice ideale.

Per il modello teorico lo studente deve fissare:

$$
\ell,
\qquad
\theta(0),
\qquad
\dot\theta(0).
$$

La lunghezza $\ell$ è la **lunghezza reale del pendolo**, misurata dal perno al centro del dado. Non deve essere ricavata dalla lunghezza in pixel del filmato.

Il modello teorico è quello ottenuto precedentemente dalle equazioni di Eulero--Lagrange:

$$
\boxed{
\ddot\theta+\frac{g}{\ell}\sin\theta=0.
}
$$

Il codice integra numericamente il sistema

$$
\begin{cases}
\dot\theta=\omega,\\[2mm]
\dot\omega=-\dfrac{g}{\ell}\sin\theta,
\end{cases}
$$

sullo stesso intervallo temporale del filmato e sovrappone teoria ed esperimento per

$$
\theta(t),
\qquad
\dot\theta(t),
\qquad
\ddot\theta(t),
$$

oltre al diagramma degli stati

$$
(\theta,\dot\theta).
$$

Per il confronto puoi scegliere se usare i dati sperimentali **filtrati** oppure, per confronto didattico, i dati **grezzi**. Il notebook mostra anche le condizioni iniziali ricavate dal filmato; puoi inserirle manualmente oppure copiarle nei controlli con il pulsante dedicato.


````python
#@title Attività 9.3 — Confronto esperimento / Eulero–Lagrange { display-mode: "form" }

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from scipy.integrate import solve_ivp
from IPython.display import display, clear_output, Markdown, Math

if 'exp_theta_raw' not in globals():
    raise RuntimeError(
        "Esegui prima l'Attività 9.1 e ricostruisci i dati sperimentali."
    )

if 'exp_theta_filt' not in globals():
    raise RuntimeError(
        "Esegui prima l'Attività 9.2 e costruisci i dati sperimentali filtrati."
    )

# ============================================================
# CONTROLLI: slider + inserimento da tastiera
# ============================================================
_common_slider_layout = widgets.Layout(width='460px')
_common_text_layout = widgets.Layout(width='220px')
_common_text_style = {'description_width': '95px'}

dataset_cmp_w = widgets.ToggleButtons(
    options=[
        ('filtrati', 'filtered'),
        ('grezzi', 'raw')
    ],
    value='filtered',
    description='dati exp:'
)

ell_exp_w = widgets.FloatSlider(
    value=0.50,
    min=0.05,
    max=2.00,
    step=0.01,
    description='ℓ [m]:',
    continuous_update=True,
    readout=True,
    readout_format='.2f',
    layout=_common_slider_layout
)

ell_exp_manual_w = widgets.BoundedFloatText(
    value=0.50,
    min=0.05,
    max=2.00,
    step=0.001,
    description='da tastiera:',
    style=_common_text_style,
    layout=_common_text_layout
)

ell_exp_link = widgets.jslink(
    (ell_exp_w, 'value'),
    (ell_exp_manual_w, 'value')
)

_theta_exp0_deg = float(
    np.rad2deg(exp_theta_filt[0])
)

_omega_exp0 = float(
    exp_omega_filt[0]
)

theta_cmp_w = widgets.FloatSlider(
    value=float(np.clip(_theta_exp0_deg, -175.0, 175.0)),
    min=-175.0,
    max=175.0,
    step=0.1,
    description='θ(0) [°]:',
    continuous_update=True,
    readout=True,
    readout_format='.1f',
    layout=_common_slider_layout
)

theta_cmp_manual_w = widgets.BoundedFloatText(
    value=float(np.clip(_theta_exp0_deg, -175.0, 175.0)),
    min=-175.0,
    max=175.0,
    step=0.01,
    description='da tastiera:',
    style=_common_text_style,
    layout=_common_text_layout
)

theta_cmp_link = widgets.jslink(
    (theta_cmp_w, 'value'),
    (theta_cmp_manual_w, 'value')
)

omega_cmp_w = widgets.FloatSlider(
    value=float(np.clip(_omega_exp0, -15.0, 15.0)),
    min=-15.0,
    max=15.0,
    step=0.05,
    description='theta_punto(0) [rad/s]:',
    style={'description_width': '175px'},
    continuous_update=True,
    readout=True,
    readout_format='.2f',
    layout=widgets.Layout(width='560px')
)

omega_cmp_manual_w = widgets.BoundedFloatText(
    value=float(np.clip(_omega_exp0, -15.0, 15.0)),
    min=-15.0,
    max=15.0,
    step=0.001,
    description='da tastiera:',
    style=_common_text_style,
    layout=_common_text_layout
)

omega_cmp_link = widgets.jslink(
    (omega_cmp_w, 'value'),
    (omega_cmp_manual_w, 'value')
)


copy_exp_btn = widgets.Button(
    description='Usa condizioni iniziali misurate'
)

compare_btn = widgets.Button(
    description='Simula e confronta'
)

compare_out = widgets.Output()


def _selected_experimental_series():
    if dataset_cmp_w.value == 'filtered':
        return (
            exp_theta_filt,
            exp_omega_filt,
            exp_alpha_filt,
            'esperimento — filtrato'
        )

    return (
        exp_theta_raw,
        exp_omega_raw,
        exp_alpha_raw,
        'esperimento — grezzo'
    )


def _copy_measured_initial_conditions(_):
    theta_exp, omega_exp, _, _ = _selected_experimental_series()

    theta_cmp_w.value = float(np.clip(
        np.rad2deg(theta_exp[0]),
        theta_cmp_w.min,
        theta_cmp_w.max
    ))

    omega_cmp_w.value = float(np.clip(
        omega_exp[0],
        omega_cmp_w.min,
        omega_cmp_w.max
    ))


def _compare_experiment_theory(_):
    with compare_out:
        clear_output(wait=True)

        theta_exp, omega_exp, alpha_exp, exp_label = (
            _selected_experimental_series()
        )

        lval = float(ell_exp_w.value)
        theta0 = np.deg2rad(float(theta_cmp_w.value))
        omega0 = float(omega_cmp_w.value)
        gval = 9.81

        # Se la funzione di verifica delle sezioni precedenti è disponibile,
        # controlliamo che T e V inseriti dallo studente corrispondano al
        # pendolo semplice ideale prima del confronto.
        if '_verifica_pendolo_esatto' in globals():
            try:
                _verifica_pendolo_esatto()
            except Exception as exc:
                print(
                    "Il modello inserito nelle sezioni precedenti "
                    "non coincide con il pendolo semplice ideale:"
                )
                print(exc)
                return

        def rhs(tt, z):
            th, om = z
            return [
                om,
                -(gval/lval)*np.sin(th)
            ]

        sol = solve_ivp(
            rhs,
            (0.0, float(exp_t_raw[-1])),
            [theta0, omega0],
            t_eval=exp_t_raw,
            method='DOP853',
            rtol=1e-10,
            atol=1e-12
        )

        if not sol.success:
            print(
                "Integrazione teorica non riuscita:",
                sol.message
            )
            return

        theta_th = sol.y[0]
        omega_th = sol.y[1]
        alpha_th = -(gval/lval)*np.sin(theta_th)

        display(Markdown("### Parametri del confronto"))

        display(Math(
            rf"\ell={lval:.4f}\,\mathrm{{m}},\qquad "
            rf"\theta(0)={np.rad2deg(theta0):.2f}^\circ,\qquad "
            rf"\dot{{\theta}}(0)={omega0:.4f}\,"
            rf"\mathrm{{rad\,s^{{-1}}}}"
        ))

        display(Math(
            rf"\theta_{{\rm exp}}(0)"
            rf"={np.rad2deg(theta_exp[0]):.2f}^\circ,\qquad "
            rf"\dot{{\theta}}_{{\rm exp}}(0)"
            rf"={omega_exp[0]:.4f}\,"
            rf"\mathrm{{rad\,s^{{-1}}}}"
        ))

        fig, axes = plt.subplots(2, 2, figsize=(12, 8))

        axes[0,0].plot(
            exp_t_raw,
            theta_exp,
            linewidth=1.2,
            label=exp_label
        )
        axes[0,0].plot(
            exp_t_raw,
            theta_th,
            linewidth=2.0,
            label='Eulero–Lagrange'
        )
        axes[0,0].set_xlabel(r'$t$ [s]')
        axes[0,0].set_ylabel(r'$\theta$ [rad]')
        axes[0,0].set_title(r'Confronto $\theta(t)$')
        axes[0,0].grid(True, alpha=0.25)
        axes[0,0].legend()

        axes[0,1].plot(
            exp_t_raw,
            omega_exp,
            linewidth=1.1,
            label=exp_label
        )
        axes[0,1].plot(
            exp_t_raw,
            omega_th,
            linewidth=2.0,
            label='Eulero–Lagrange'
        )
        axes[0,1].set_xlabel(r'$t$ [s]')
        axes[0,1].set_ylabel(r'$\dot\theta$ [rad/s]')
        axes[0,1].set_title(r'Confronto $\dot\theta(t)$')
        axes[0,1].grid(True, alpha=0.25)
        axes[0,1].legend()

        axes[1,0].plot(
            exp_t_raw,
            alpha_exp,
            linewidth=1.0,
            label=exp_label
        )
        axes[1,0].plot(
            exp_t_raw,
            alpha_th,
            linewidth=2.0,
            label='Eulero–Lagrange'
        )
        axes[1,0].set_xlabel(r'$t$ [s]')
        axes[1,0].set_ylabel(
            r'$\ddot\theta$ [rad/s$^2$]'
        )
        axes[1,0].set_title(
            r'Confronto $\ddot\theta(t)$'
        )
        axes[1,0].grid(True, alpha=0.25)
        axes[1,0].legend()

        axes[1,1].plot(
            theta_exp,
            omega_exp,
            linewidth=1.1,
            label=exp_label
        )
        axes[1,1].plot(
            theta_th,
            omega_th,
            linewidth=2.0,
            label='Eulero–Lagrange'
        )
        axes[1,1].set_xlabel(r'$\theta$ [rad]')
        axes[1,1].set_ylabel(
            r'$\dot\theta$ [rad/s]'
        )
        axes[1,1].set_title(
            'Confronto nel piano degli stati'
        )
        axes[1,1].grid(True, alpha=0.25)
        axes[1,1].legend()

        plt.tight_layout()
        plt.show()

        display(Markdown(
            "**Da osservare:** il modello teorico descrive un pendolo ideale "
            "senza dissipazione. Il filmato reale può mostrare smorzamento, "
            "piccole imperfezioni geometriche e rumore di misura. "
            "Il confronto permette quindi di distinguere gli effetti del "
            "modello dinamico da quelli sperimentali."
        ))


copy_exp_btn.on_click(
    _copy_measured_initial_conditions
)

compare_btn.on_click(
    _compare_experiment_theory
)

display(Markdown("### Condizioni iniziali misurate dal video"))

display(Math(
    rf"\theta_{{\rm raw}}(0)"
    rf"={np.rad2deg(exp_theta_raw[0]):.2f}^\circ,\qquad "
    rf"\dot{{\theta}}_{{\rm raw}}(0)"
    rf"={exp_omega_raw[0]:.4f}\,\mathrm{{rad\,s^{{-1}}}}"
))

display(Math(
    rf"\theta_{{\rm filt}}(0)"
    rf"={np.rad2deg(exp_theta_filt[0]):.2f}^\circ,\qquad "
    rf"\dot{{\theta}}_{{\rm filt}}(0)"
    rf"={exp_omega_filt[0]:.4f}\,\mathrm{{rad\,s^{{-1}}}}"
))

display(widgets.VBox([
    dataset_cmp_w,
    widgets.HBox([ell_exp_w, ell_exp_manual_w]),
    widgets.HBox([theta_cmp_w, theta_cmp_manual_w]),
    widgets.HBox([omega_cmp_w, omega_cmp_manual_w]),
    widgets.HBox([copy_exp_btn, compare_btn]),
    compare_out
]))
````
